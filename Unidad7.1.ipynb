## <font color=#38bdf8>El Módulo "os" en Python</font>

El módulo <span style="color:orange">os</span> es una biblioteca estandar que proporciona una interfaz para interactuar con el sistema operativo.

Permite realizar operaciones relacionadas con la gestión de archivos, directorios, rutas, variables de entorno y otras funcionalidades.

### <font color=#38bdf8>Características más comunes</font>

+ Manipulación de rutas y de nombres de archivos. (<span style="color:limegreen">os.path</span>)

+ Realizar listados de archivos y directorios. (<span style="color:limegreen">os.listdir()</span>)

+ Creación y eliminación de directorios. (<span style="color:limegreen">os.mkdir()</span> y <span style="color:limegreen">os.rmdir()</span>).

+ Manipulación de variables de entorno. (<span style="color:limegreen">os.getenv()</span> y <span style="color:limegreen">os.environ()</span>)

+ Recuperar información sobre el sistema.

+ Cambiar de directorio actual. (<span style="color:limegreen">os.chdir()</span>)

+ Funciones relacionadas con procesos. (<span style="color:limegreen">os.fork()</span> y <span style="color:limegreen">os.kill()</span>)

### <font color=#38bdf8>Funcionalidades Claves</font>

+ Gestión de Rutas y Archivos:

    + <span style="color:limegreen">os.path</span>: Proporciona funciones para manipular rutas y nombres de archivos de ma- nera independiente del sistema operativo.

    + <span style="color:limegreen">os.path.join()</span>: Combina partes de una ruta en una sola ruta.

    + <span style="color:limegreen">os.path.exists()</span>: Verifica si una ruta o archivo existe.

    + <span style="color:limegreen">os.listdir()</span>: Lista los archivos y directorios en un directorio específico.

    + <span style="color:limegreen">os.mkdir()</span>: Crea un nuevo directorio.

    + <span style="color:limegreen">os.remove(path)</span>: Elimina un archivo.

    + <span style="color:limegreen">os.rename(source, destination)</span>: Mueve un archivo del origen al destino.

    + <span style="color:limegreen">os.rmdir(dirpath)</span>: Borra el directorio especificado.

    + <span style="color:limegreen">os.removedirs(dirpath)</span>: Borra el directorio y todos los subdirectorios dentro de él recursivamente.

    + <span style="color:limegreen">os.stat()</span>: Obtiene el estado de una archivo.
    
+ Variables de Entorno:

    + <span style="color:limegreen">os.getenv()</span>: Obtiene el valor de una variable de entorno específica.

    + <span style="color:limegreen">os.environ</span>: Proporciona acceso a todas las variables de entorno del sistema.

+ Interacción con el Sistema Operativo:

    + <span style="color:limegreen">os.system()</span>: Ejecuta un comando del sistema.

    + <span style="color:limegreen">os.startfile()</span>: Abre un archivo con la aplicación predeterminada.

    + <span style="color:limegreen">os.chdir()</span>: Cambia el directorio de trabajo actual.
    
+ Información del Sistema:

    + <span style="color:limegreen">os.name</span>: Devuelve el nombre del sistema operativo (por ejemplo, ’posix’ en sistemas tipo Unix).

    + <span style="color:limegreen">os.uname()</span>: Proporciona información detallada sobre el sistema operativo (solo en sistemas Unix).

+ Manipulación de Procesos:

    + <span style="color:limegreen">os.fork()</span>: Crea un nuevo proceso duplicado.

    + <span style="color:limegreen">os.kill()</span>: Envía señales a procesos.

    + <span style="color:limegreen">os.wait()</span>: Espera a que un proceso hijo termine.
    
    + <span style="color:limegreen">os.pid()</span>: Obtiene el PID actual.

+ Manipulación de Directorios:

    + <span style="color:limegreen">os.chdir()</span>: Cambia el directorio de trabajo actual.

    + <span style="color:limegreen">os.getcwd()</span>: Obtiene el directorio de trabajo actual.

    + <span style="color:limegreen">os.listdir(dirpath)</span>: Muestra el contenido del directorio indicado.

    + <span style="color:limegreen">os.mkdir(dirpath)</span>: Crea un directorio.

    + <span style="color:limegreen">os.makedirs(dirpath)</span>: Crea el directorio y todos los directorios primarios requeridos recursivamente.

+ Funciones de Ruta Avanzadas:

    + <span style="color:limegreen">os.path.abspath()</span>: Convierte una ruta relativa en una ruta absoluta.
    
    + <span style="color:limegreen">os.path.basename()</span>: Obtiene el nombre base de un archivo o ruta.

    + <span style="color:limegreen">os.path.dirname()</span>: Obtiene el directorio de una ruta.

+ Manipulación de Permisos y Propietarios:

    + <span style="color:limegreen">os.chmod()</span>: Cambia los permisos de un archivo.

    + <span style="color:limegreen">os.chown()</span>: Cambia el propietario y el grupo de un archivo.
    
    + <span style="color:limegreen">os.getuid()</span>: Obtiene el ID del usuario actual.

En resumen, el módulo <span style="color:orange">os</span> en Python es esencial para interactuar con el sistema operativo en el que se ejecuta el programa, lo que permite realizar tareas como la manipulación de archivos, la gestión de directorios y la obtención de información sobre el sistema.

### <font color=#38bdf8>Acceso a Funciones del Sistema Operativo</font>

El módulo <span style="color:orange">os</span> proporciona un contenedor para módulos específicos de la plataforma, tales como "posix", "nt" y "mac". La interfaz de programación para las funciones disponibles en todas las plataformas debe ser la misma, por lo que usar el módulo <span style="color:orange">os</span> ofrece cierta medida de portabilidad. Sin embargo, no todas las funciones están disponibles en todas las plataformas.

Muchas de las funciones de gestión de procesos descritas no están disponibles para el sistema operativo Windows.

El módulo <span style="color:orange">os</span> consta principalmente de funciones para crear y administrar procesos en ejecución o contenido del sistema de archivos (archivos y directorios), además de algunos otras funcionalidades.

#### <font color=#38bdf8>Examinar el contenido del Sistema de Archivos</font>

Para preparar una lista de los contenidos de un directorio en el sistema de archivos, se usa <span style="color:limegreen">os.listdir()</span>.

```python
# Archivo os_listdir.py
# Uso: python3 os_listdir.py .
import os
import sys
print(sorted(os.listdir(sys.argv[1])))
```

El valor de retorno es una lista sin clasificar de todos los miembros nombrados del directorio dado (.). No hace distinción entre archivos, subdirectorios o enlaces simbólicos.

La función <span style="color:limegreen">walk()</span> accede a un directorio de forma recursiva y para cada subdirectorio genera una tupla que contiene la ruta del directorio, cualquier subdirectorio de esa ruta y una lista de los nombres de los archivos en ese directorio.

```python
# Archivo os_walk.py
# Uso: python3 os_walk.py .
import os
import sys

# Si no se da un path a la lista, se usa /tmp
if len(sys.argv) == 1:
    root = '/tmp'
else:
    root = sys.argv[1]

for dir_name, sub_dirs, files in os.walk(root):
    print(dir_name)
    # Hacer que los nombres de los subdirectorios se destaquen con /
    sub_dirs = [n + '/' for n in sub_dirs]
    # Mezcla los contenidos de los directorios
    contents = sub_dirs + files
    contents.sort()
    # Mostrar el contenido
    for c in contents:
        print('  {}'.format(c))
    print()
```

Si se necesita más información que los nombres de los archivos, es probable que sea más eficiente usar <span style="color:limegreen">os.scandir()</span> porque se recopila más información en una llamada al sistema cuando se escanea el directorio.

```python
# Archivo os_scandir.py
# Uso: python3 os_scandir.py .
import os
import sys

for entry in os.scandir(sys.argv[1]):
    if entry.is_dir():
        typ = 'dir'
    elif entry.is_file():
        typ = 'archivo'
    elif entry.is_symlink():
        typ = 'link'
    else:
        typ = 'desconocido'
    print(f'{entry.name} {typ}')
```

#### <font color=#38bdf8>Administrar Permisos del Sistema de Archivos</font>

Se puede acceder a información detallada sobre un archivo usando <span style="color:limegreen">os.stat()</span> o <span style="color:limegreen">os.lstat()</span> (para verificar el estado de algo que podría ser un enlace simbólico).

```python
# Archivo os_stat.py
# Uso: python3 os_stat.py nada | nombre_archivo
import os
import sys
import time

if len(sys.argv) == 1:
    filename = __file__
else:
    filename = sys.argv[1]

stat_info = os.stat(filename)

print(f'os.stat({filename}):')
print('  Tamaño:', stat_info.st_size)
print('  Permisos:', oct(stat_info.st_mode))
print('  Propietario:', stat_info.st_uid)
print('  Dispositivo:', stat_info.st_dev)
print('  Creado      :', time.ctime(stat_info.st_ctime))
print('  Ultima modificación:', time.ctime(stat_info.st_mtime))
print('  Ultimo acceso:', time.ctime(stat_info.st_atime))
```

Intente pasar diferentes nombres de archivo en la línea de comando al archivo <span style="color:magenta">os_stat.py</span>.

En sistemas tipo Unix, los permisos de archivo se pueden cambiar usando <span style="color:limegreen">os.chmod()</span>, pasando el modo como un entero. Los valores posibles de modo pueden construirse usando constantes definidas en el módulo <span style="color:orange">stat</span>. El siguiente ejemplo alterna el bit de permiso de ejecución del usuario:

```python
# Archivo os_stat_chmod.py
# Uso: python3 os_stat_chmod.py
import os
import stat

filename = 'os_stat_chmod_ejemplo.txt'
if os.path.exists(filename):
    os.unlink(filename)
with open(filename, 'wt') as f:
    f.write('contenidos')

# Determina los permisos que esta usando stat
existing_permissions = stat.S_IMODE(os.stat(filename).st_mode)

if not os.access(filename, os.X_OK):
    print('Sumando permisos de ejecución')
    new_permissions = existing_permissions | stat.S_IXUSR
else:
    print('Borrando permisos de ejecución')
    # usar xor para borrar los permisis de ejecución del usuario
    new_permissions = existing_permissions ^ stat.S_IXUSR

os.chmod(filename, new_permissions)
```

La secuencia de comandos asume que tiene los permisos necesarios para modificar el modo del archivo cuando se ejecuta.

La función <span style="color:limegreen">os.access()</span> se puede usar para probar los derechos de acceso que un proceso tiene para un archivo.

```python
# Archivo os_access.py
# Uso: python3 os_access.py
import os

print('Probando:', __file__)
print('Existe?:', os.access(__file__, os.F_OK))
print('Se puede Leer?:', os.access(__file__, os.R_OK))
print('Se puede Escribir?:', os.access(__file__, os.W_OK))
print('Se puede Ejecutar?:', os.access(__file__, os.X_OK))
```

La documentación de la biblioteca para <span style="color:limegreen">os.access()</span> incluye dos advertencias especiales. Primero, no tiene mucho sentido invocar a <span style="color:limegreen">os.access()</span> para probar si se puede abrir un archivo antes de invocar a <span style="color:limegreen">open()</span> en él. 

Hay una pequeña, pero real, ventana de tiempo entre las dos invocaciones durante la cual los permisos en el archivo podrían cambiar. 

La otra advertencia se aplica principalmente a los sistemas de archivos en red que extienden la semántica de permisos POSIX. Algunos tipos de sistemas de archivos pueden responder a la invocación POSIX de que un proceso tiene permiso para acceder a un archivo, luego informan una falla cuando el intento se realiza usando <span style="color:limegreen">open()</span> por alguna razón no probada a través de la llamada POSIX. Con todo, es mejor invocar a <span style="color:limegreen">open()</span> con el modo requerido y detectar el <span style="color:limegreen">IOError</span> generado si hay un problema.

#### <font color=#38bdf8>Crear y Eliminar Directorios</font>

Existen varias funciones para trabajar con directorios en el sistema de archivos, incluida la creación, el listado de contenidos y su eliminación.

```python
#Archivo os_directories.py
# Uso: python3 os_directories.py
import os
dir_name = 'os_directories_ejemplo'
print('Creando...', dir_name)
os.mkdir(dir_name)

file_name = os.path.join(dir_name ,'ejemplo_dir.txt')
print(file_name)
print('Creando...', file_name)
with open(file_name , 'wt') as f:
    f.write('Archivo␣de␣ejemplo')
```

Hay dos conjuntos de funciones para crear y eliminar directorios. Al crear un nuevo directorio con <span style="color:limegreen">os.mkdir()</span>, todos los directorios principales ya deben existir. Al eliminar un directorio con <span style="color:limegreen">os.rmdir()</span>, solo se elimina el directorio hoja (la última parte de la ruta). En contraste, <span style="color:limegreen">os.makedirs()</span> y <span style="color:limegreen">os.removedirs()</span> operan en todos los nodos en la ruta. <span style="color:limegreen">os.makedirs()</span> creará cualquier parte de la ruta que no exista, y <span style="color:limegreen">os.removedirs()</span> eliminará todos los directorios principales, siempre que estén vacíos.

#### <font color=#38bdf8>Enlaces Simbólicos</font>

Para las plataformas y los sistemas de archivos que admiten los enlaces simbólicos, existen varias funciones para trabajar con ellos.

```python
# Archivo os_simlinks.py
# Uso: python3 os_simlinks.py
import os

link_name = '/tmp/' + os.path.basename(__file__)
print(link_name)

print(f'Creando enlace (link) {link_name} -> {__file__}')
os.symlink(__file__ , link_name)

stat_info = os.lstat(link_name)
print('Permisos:', oct(stat_info.st_mode))

print('Apunta a:', os.readlink(link_name))

# Cleanup
# #os.unlink(link_name)
```

Usar <span style="color:limegreen">os.symlink()</span> para crear un enlace simbólico y <span style="color:limegreen">os.readlink()</span> para leerlo y determinar el archivo original al que apunta el enlace. La función <span style="color:limegreen">os.lstat()</span> es como <span style="color:limegreen">os.stat()</span>, pero opera en enlaces simbólicos.

#### <font color=#38bdf8>Renombrar un Archivo Existente</font>

Reemplazar o renombrar un archivo existente no es trivial y puede exponer a las aplicaciones a condiciones complejas. Las funciones <span style="color:limegreen">os.rename()</span> y <span style="color:limegreen">os.replace()</span> implementan algoritmos seguros para estas acciones, utilizando operaciones atómicas en sistemas compatibles con POSIX.


```python
# Archivo os_rename_replace.py
# Uso: python3 os_reanme_replace.py
import glob
import os

with open('renombrar_inicio.txt', 'w') as f:
    f.write('comenzando como renombrar_inicio.txt')

print('Comenzando:', glob.glob('renombrar*.txt'))

os.rename('renombrar_inicio.txt', 'renombrar_final.txt')
print('Despues de renombrar:', glob.glob('renombrar*.txt'))

with open('renombrar_final.txt', 'r') as f:
    print('Contenidos:', repr(f.read()))

with open('renombrar_nuevo_contenidos.txt', 'w') as f:
    f.write('finalizando con contenidos de renombrar_nuevo_contenido.txt')

os.replace('renombrar_nuevo_contenidos.txt', 'renombrar_final.txt')

with open('renombrar_final.txt', 'r') as f:
    print('Despues del reemplazo:', repr(f.read()))

# for name in glob.glob('renombrar*.txt'):
#     os.unlink(name)
```

#### <font color=#38bdf8>Determinar y Cambiar el Propietario de un Proceso</font>

El siguiente conjunto de funciones proporcionadas por <span style="color:orange">os</span> se utiliza para determinar y cambiar las identificaciones del propietario de un proceso. Estas son utilizadas con mayor frecuencia por los autores de demonios o programas especiales del sistema que necesitan cambiar el nivel de permiso en lugar de ejecutarse como <span style="color:magenta">root</span>. Esta sección no trata de explicar todos los detalles intrincados de la seguridad de Unix, los propietarios de procesos, etc. 

El siguiente ejemplo muestra la información real y efectiva del usuario y del grupo para un proceso, y luego cambia los valores efectivos. Esto es similar a lo que un demonio necesitaría hacer cuando se inicia como <span style="color:magenta">root</span> durante un arranque del sistema, para reducir el nivel de privilegio y ejecutarse como un usuario diferente.

<span style="color:yellow">Nota</span>: Antes de ejecutar el ejemplo, cambiar los valores <span style="color:limegreen">TEST_GID</span> y <span style="color:limegreen">TEST_UID</span> para que coincidan con un usuario real definido en el sistema.

```python
# Archivo os_process_user_example.py
# Uso: python3 os_process_user_example.py
import os

TEST_GID = 501
TEST_UID = 20


def show_user_info():
    print('Usuario (actual/efectivo)  : {} / {}'.format(
        os.getuid(), os.geteuid()))
    print('Grupo (actual/efectivo) : {} / {}'.format(
        os.getgid(), os.getegid()))
    print('Grupos Actuales   :', os.getgroups())


print('ANTES DEL CAMBIO:')
show_user_info()
print()

try:
    os.setegid(TEST_GID)
except OSError:
    print('ERROR: No se pudo cambiar el grupo efectivo. '
          'Ejecutar de nuevo como root.')
else:
    print('CAMBIO DE GRUPO:')
    show_user_info()
    print()

try:
    os.seteuid(TEST_UID)
except OSError:
    print('ERROR: No se pudo cambiar el usuario efectivo. '
          'Ejecutar de nuevo como root.')
else:
    print('CAMBIO DE USUARIO:')
    show_user_info()
    print()
```

Los valores no cambian porque cuando no se ejecuta como <span style="color:magenta">root</span>, un proceso no puede cambiar su valor de propietario efectivo. Cualquier intento de establecer la identificación de usuario efectiva o la identificación de grupo en otra cosa que no sea la del usuario actual provoca un <span style="color:limegreen">OSError</span>. 

Ejecutar la misma secuencia de comandos usando <span style="color:magenta">sudo</span> para que comience con privilegios de <span style="color:magenta">root</span> es una historia diferente.

#### <font color=#38bdf8>Gestionar el Entorno de un Proceso</font>

Otra característica del sistema operativo expuesta a un programa a través del módulo <span style="color:orange">os</span> es el entorno. Las variables establecidas en el entorno son visibles como strings que se pueden leer a través de <span style="color:limegreen">os.environ</span> o <span style="color:limegreen">os.getenv()</span>.

Las variables de entorno se usan comúnmente para valores de configuración, como rutas de búsqueda, ubicaciones de archivos e indicadores de depuración. 

Este ejemplo muestra cómo recuperar una variable de entorno y pasar un valor a un proceso secundario.

```python
# Archivo os_environ_example.py
# Uso:python3 -u os_environ_example.py
import os

print('Valor Inicial:', os.environ.get('TESTVAR', None))
print('Proceso hijo:')
os.system('echo $TESTVAR')
os.environ['TESTVAR'] = 'ESTE VALOR SE CAMBIO'
print()

print('Valor cambiado:', os.environ['TESTVAR'])
print('Proceso hijo:')
os.system('echo $TESTVAR')

del os.environ['TESTVAR']
print()
print('Valor borrado:', os.environ.get('TESTVAR', None))
print('Proceso hijo:')
os.system('echo $TESTVAR')
```

#### <font color=#38bdf8>Administrar el Directorio de Trabajo de un Proceso</font>

Los sistemas operativos con sistemas de archivos jerárquicos poseen el concepto de *directorio de trabajo actual*: el directorio en el sistema de archivos del proceso que usa como ubicación de inicio cuando se accede a los archivos con rutas relativas. El directorio de trabajo actual se puede recuperar con <span style="color:limegreen">os.getcwd()</span> y se puede cambiar con <span style="color:limegreen">os.chdir()</span>.

```python
# Archivo os_cwd_example.py
# Uso: python3 os_cwd_example.py
import os

print('Comenzando:', os.getcwd())

print('Moviendo al directorio padre:', os.pardir)
os.chdir(os.pardir)

print('Después de moverse:', os.getcwd())
```

#### <font color=#38bdf8>Ejecutar Comandos Externos</font>

<font color=orange>Advertencia</font>: Muchas de estas funciones que trabajan con procesos tienen una portabilidad limitada. Para una forma más consistente de trabajos con procesos de manera independiente de la plataforma, estudiar el módulo <span style="color:orange">mod.subprocess</span> en su lugar.

La forma básica de ejecutar un comando, sin interactuar con él, es con el método <span style="color:limegreen">os.system()</span>. Toma un solo argumento string, que es la línea de comando que debe ejecutar un subproceso que ejecuta un shell.

```python
# Archivo os_system_example.py
# Uso: python3 -u os_system_example.py
import os

# Comando
p =os.system('pwd')
print(p)
```

El valor de retorno de <span style="color:limegreen">os.system()</span> es el valor de salida del shell que ejecuta el programa, empaquetado en un número de 16 bits, donde el byte mas significativo es el estado de la salida y el byte menos significativo es el número de señal que causó la muerte del proceso, o cero.

Dado que el comando se pasa directamente al shell para su procesamiento, se puede incluir la sintaxis del shell, como el <span style="color:magenta">globbing</span> o las variables de entorno.

```python
# Archivo os_system_shell.py
# Uso: python3 -u os_system_shell.py
import os

# Command con expansion del shell
os.system('echo $TMPDIR')
```

La variable de entorno <span style="color:magenta">$TMPDIR</span> en este string se expande cuando el shell ejecuta la línea de comando.

A menos que el comando se ejecute explícitamente en segundo plano, la invocación a <span style="color:limegreen">os.system()</span> se bloquea hasta que se complete. La entrada, salida y error estándar del proceso secundario están vinculados a las secuencias apropiadas que posee la persona que invoca de forma predeterminada, pero se pueden redirigir utilizando la sintaxis del shell.

```python
# Archivo os_system_background.py
# Uso: python3 -u os_system_background.py
import os
import time

print('Invocando...')
os.system('date; (sleep 3; date) &')

print('Durmiendo...')
time.sleep(5)
```
