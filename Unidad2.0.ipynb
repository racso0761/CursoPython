**JupyterLab** es una interfaz de notebooks de última generación.
JupyterLab es el entorno de desarrollo interactivo web más reciente para notebooks, código y datos. Su interfaz flexible permite a los usuarios configurar y organizar flujos de trabajo en ciencia de datos, computación científica, periodismo computacional y aprendizaje automático. Su diseño modular permite extensiones para ampliar y enriquecer la funcionalidad.

Además de proporcionar una plataforma de codificación interactiva, <b>Jupyter Notebook</b> admite tanto celdas de código como de texto. Las celdas de texto permiten el formato <b>Markdown</b>. Así que puede utilizar texto sin formato, imágenes, ecuaciones matemáticas LaTex y mucho más para explicar el flujo de trabajo de su proyecto.

Los archivos producidos por esta aplicación, llamados <b>jupyter notebooks</b> o simplemente <b>notebooks</b> (<i>cuadernos</i>), tienen la extensión <b>.ipynb</b>, y se puede acceder a ellos desde la aplicación de <b>Jupyter Notebook</b> o desde alguna otra que los soporte (<i>VSCode</i>, <i>Google Colab</i>, etc.). También aunque no es lo recomendable se puede trabajar en ellos desde cualquier aplicación de edición de texto ya que se tratan en realidad de archivos de texto en formato <i>JSON</i>. En este último caso se pierde el formato visual que proporciona la aplicación <b>Jupyter Notebook</b>.

La aplicación **Jupyter Notebook** consta en realidad de dos componentes:

+ <b>Kernel</b>: es la parte que se encarga de ejecutar el código que contiene el <i>notebook</i>. En principio el kernel que viene por defecto ejecuta código <i>Python</i>, aunque también se pueden instalar kernels para otros lenguajes.
+ <b>Dashboard</b>: es la parte que permite mostrar los <i>notebooks</i> (renderizar y ejecutar el código) que se disponen en una computadora, y se usa también para gestionar los kernels.</li>

Tanto el ambiente de desarrollo integrado <i>VSCode</i> como <i>Google Colab<i> permiten hacer uso de los <i>notebooks</i> para producir texto entrelazado con código Python ejecutable y colaborar en grupos de trabajo.

Cuando la tarea es desarrollar scripts de Python, tales como una aplicación particular siempre es mejor utilizar un IDE (Ambiente de desarrollo integrado) tal cómo <i>VSCode</i> o <i>PyCharm</i>. Este último es de pago aunque existe una versión restringida de uso gratuito. En cambio <i>VSCode</i> al ser de código libre se puede usar gratuitamente.

Si por el contrario lo que se desea hacer es trabajar en <i>Ciencia de Datos</i>, <i>Machine Learning</i>, <i>Deep Learning</i> o <i>Aprendizaje Estadístico</i> la opción de utilizar <b>Jupyter Notebooks</b> es la ideal.

La creación y uso de <b>Jupyter Notebooks</b> tal y como se comentó anteriormente se puede hacer desde <i>VSCode</i> o desde <i>Google Colab</i>, en este último caso la ventaja principal es que se trabaja en la nube y no se necesita instalar nada en la computadora local.

En las próximas secciones se desarrollará un breve introducción a <i>Google Colab</i> y al manejo de los notebooks como ambientes de desarrollo.

## <font color=#38bdf8>Introducción</font>

<i>Google Colaboratory</i> o simplemente <i>Colab</i>, es un entorno de trabajo gratuito en la nube que permite escribir y ejecutar código <i>Python</i> en un simple navegador web sin requerir ninguna configuración adicional ni pre-requisitos, no se necesita instalar absolutamente nada en la computadora local y todo lo que se produzca se puede almacenar en la nube. 

<i>Colab</i> está basado en <b>Jupyter Notebook</b>, por lo cual los <i>notebooks</i> o cuadernos de <i>Colab</i> permiten combinar código ejecutable y texto enriquecido (<i>Markdown</i>) en un único documento, junto con imágenes, código HTML, \LaTeX y muchos elementos más. Sin embargo, los <i>notebooks</i> que se creen en <i>Colab</i> son almacenados en una cuenta de <i>Google Drive</i>, es decir, los <i>notebooks</i> de <i>Colab</i> son <i>notebooks</i> de <b>Jupyter</b> alojados en la nube. Además, <i>Colab</i> crea cuadernos en la versión 3.x de <i>Python</i> (o en la versión 2.x, aunque esto no se recomienda) y el ambiente de trabajo ya viene con muchas bibliotecas o paquetes pre-instalados listos para utilizar, pero también permite instalar otros.

## <font color=#38bdf8>Características Principales</font>

La característica más importante que distingue a <i>Colab</i> de otros servicios gratuitos en la nube es que proporciona acceso gratuito (limitado) al uso de <i>GPU</i>s. Una <i>GPU</i> es una <i>Unidad de Procesamiento Gráfico</i> que es un coprocesador dedicado al procesamiento de gráficos u operaciones de punto flotante. Esta posibilidad de hacer uso gratuito de <i>GPU</i>s, eventualmente, podría liberar a la computadora de tener que llevar a cabo un trabajo demasiado costoso tanto en tiempo como en procesamiento e incluso permitiría realizar ese trabajo si la computadora no cuenta con recursos suficientes.

Otro de los beneficios que tiene <i>Colab</i> lo indica su propio nombre, ``Colaboratory``, es decir, colaborativo, y permite realizar trabajos en la nube, almacenarlos en <i>Google Drive</i> y compartir documentos si se necesita trabajar en equipo. Se pueden compartir fácilmente los <i>notebooks</i> para que otras personas realicen comentarios o los editen, como se haría en <i>Google Docs</i>.

## <font color=#38bdf8>¿Por Que utilizar Google Colab?</font>

1. <i>Ejecución en la Nube</i>: <i>Colab</i> permite ejecutar notebooks directamente en la nube, lo que significa que no se necesita configurar una computadora propia o entorno local.
2. <i>Sin configuración</i>: Se puede ejecutar código <i>Python</i> directamente en un navegador.
3. <i>Acceso gratuito a GPUs</i>: Ideal para aprendizaje automático y tareas relacionadas que requieren alto poder computacional.
4. <i>Fácil de compartir</i>: Se pueden compartir los proyectos como se hace con documentos de <i>Google Docs</i> o <i>Google Sheets</i>.
5. <i>Integración con Google Drive</i>: Almacena todo su trabajo directamente en <i>Google Drive</i>.
6. <i>Interactivo</i>: Permite una ejecución interactiva, similar a <b>Jupyter Notebook</b>, es decir entrelazar texto y código <i>Python</i> ejecutable.
7. <i>Soporte de Varios Formatos</i>: Además de <i>Python</i>, <i>Colab</i> soporta otros formatos y lenguajes de programación como <i>Markdown</i>, lo que permite crear documentos con un formato atractivo y explicativo.
8. <i>Integración con GitHub</i>: Puede guardar los <i>notebooks</i> de <i>Colab</i> directamente en repositorios de <i>GitHub</i> y también puede cargar <i>notebooks</i> desde <i>GitHub</i> a <i>Colab</i>.
9. <i>Instalar paquetes</i>: Se pueden instalar paquetes adicionales usando la sentencia ``!pip install nombre_del_paquete``.
10. <i>Entorno Educativo</i>: Dada su facilidad de uso y acceso gratuito, <i>Colab</i> es ampliamente utilizado en educación, especialmente en cursos en donde el entrelazado de texto y código ejecutable es importante.
11. <i>Colaboración en Tiempo Real</i>: Al igual que otros productos de <i>Google</i>, múltiples personas pueden colaborar en un <i>notebook</i> en tiempo real, lo que facilita el trabajo en equipo.
12. <i>Celdas de texto</i>: Se pueden agregar celdas de texto para incluir explicaciones, imágenes o ecuaciones usando el lenguaje <i>Markdown</i>.

## <font color=#38bdf8>El Menú de Google Colab</font>

El menú principal de *Google Colab* ofrece una gran variedad de opciones que permiten interactuar y controlar un *notebook*. El menú posee las siguientes opciones:

+ <i>Archivo</i>:
	+ <i>Nuevo notebook</i>: Crea un notebook en blanco.
	+ <i>Subir notebook</i>: Permite subir un notebook desde una computadora a la nube.
	+ <i>Abrir notebook</i>: Abre un notebook existente desde <i>Google Drive</i>, <i>GitHub</i> o una <i>URL</i>.
	+ <i>Guardar</i>: Permite guardar los cambios actuales en un notebook.
	+ <i>Guardar una copia en Drive</i>: Permite guardar una copia del notebook actual en <i>Google Drive</i>.
	+ <i>Descargar archivo</i> <b>.ipynb</b>: Permite descargar a la computadora el notebook en formato <b>.ipynb</b>.
	+ <i>Descargar archivo</i> <b>.py</b>: Permite descargar el notebook en formato de script de <i>Python</i>.
	+ <i>Imprimir</i>: Permite realizar una impresión del notebook.
	+ <i>Cerrar y Detener</i>: Cierra el notebook y detiene cualquier celda en ejecución.</li>
+ <i>Editar</i>:
	+ <i>Deshacer</i>: Deshace la última acción.
	+ <i>Rehacer</i>: Rehace la última acción deshecha.
	+ <i>Cortar celda</i>: Corta la celda seleccionada.
	+ <i>Copiar celda</i>: Copia la celda seleccionada.
	+ <i>Pegar celda</i>: Pega la celda copiada o cortada.
	+ <i>Buscar y reemplazar</i>: Busca y reemplaza texto en el notebook.
	+ <i>Configuración del notebook</i>: Accede a la configuración del notebook, como cambiar el tipo de acelerador de hardware.
+ <i>Ver</i>:
	+ <i>Tabla de contenidos</i>: Muestra una tabla de contenidos basada en los encabezados del notebook.
	+ <i>Mostrar/Ocultar código</i>: Muestra u oculta todas las celdas de código en el notebook.
	+ <i>Mostrar/Ocultar salida</i>: Muestra u oculta la salida de todas las celdas de código.
+ <i>Insertar</i>:
	+ <i>Código</i>: Inserta una nueva celda de código debajo de la celda actual.
	+ <i>Texto</i>: Inserta una nueva celda de texto debajo de la celda actual.
+ <i>Entorno de Ejecución</i>:
	+ <i>Ejecutar todo</i>: Ejecuta todas las celdas del notebook en orden.
	+ <i>Ejecutar antes</i>: Ejecuta todas las celdas que están antes de la celda actual.
	+ <i>Ejecutar selección</i>: Ejecuta la celda o celdas seleccionadas.
	+ <i>Interrumpir ejecución</i>: Detiene la ejecución de la celda que se está ejecutando.
	+ <i>Reiniciar entorno de ejecución</i>: Reinicia el kernel, lo que borra todas las variables y estados.
	+ <i>Cambiar tipo de entorno de ejecución</i>: Te permite cambiar la configuración, como seleccionar una GPU o TPU.
+ <i>Herramientas</i>:
	+ <i>Configuración</i>: Accede a las preferencias y configuraciones del notebook.
	+ <i>Comandos</i>: Muestra una lista de comandos que puedes usar en *Colab*.
	+ <i>Extensiones</i>: Permite gestionar y agregar extensiones.
+ <i>Ayuda</i>: Contiene enlaces a documentación, tutoriales y otros recursos útiles para trabajar con <i>Colab</i>.

## <font color=#38bdf8>El Menú Lateral de Google Colab</font>

Las pestañas en el menú de lateral izquierdo de *Google Colab* ayudan a navegar, buscar, gestionar variables y archivos, y acceder a fragmentos de código útiles, lo que facilita y mejora la experiencia de trabajo en la plataforma.

1. <i>Índice (Table of Contents)</i>: Esta pestaña muestra un índice o tabla de contenidos del notebook. Cada título en el notebook (creado usando <i>Markdown</i>) aparecerá en esta sección, permitiendo navegar rápidamente a diferentes secciones del notebook. Es especialmente útil para notebooks largos.

2. <i>Búsqueda (Search)</i>: Esta pestaña permite buscar específicamente en el notebook. Se pueden introducir términos de búsqueda y <i>Colab</i> resaltará y llevará a las ocurrencias de esos términos en el notebook. Es similar a la función de búsqueda en la mayoría de los editores de texto.

3. <i>Variables (Code snippets)</i>: En esta opción se pueden ver todas las variables que están actualmente en memoria en el entorno de ejecución. Muestra el nombre de la variable, su tipo y su valor actual. Es útil para hacer un seguimiento de las variables que se han definido y para depurar el código. Además, en esta pestaña, se pueden encontrar fragmentos de código predefinidos que se pueden insertar directamente en el notebook. Estos fragmentos incluyen tareas comunes como cargar datos o visualizar gráficos.

4. <i>Archivos (Files)</i>: Esta pestaña muestra el sistema de archivos del entorno de ejecución de <i>Colab</i>. Aquí se pueden ver y gestionar archivos que se han cargado o generado. También permite cargar archivos desde una computadora o descargar archivos a una computadora. Además, se puede montar <i>Google Drive</i> en el entorno de ejecución de <i>Colab</i>, lo que permite acceder a los archivos de <i>Google Drive</i> directamente.

Ahora se está en condiciones de contestar la pregunta, <i>¿Qué es un notebook?</i>

Un <span style="color:limegreen">notebook</span> es un documento que está compuesto por celdas en las que se puede incluir:

+ Texto en formato <span style="color:limegreen">Markdown</span> (incluyendo fórmulas).
+ Elementos multimedia.
+ Código Python ejecutable (también otros lenguajes).

En  <span style="color:limegreen">Markdown</span>, la salida de la celda es la renderización del texto. En código Python, la salida de la celda es el resultado de las sentencias incluidas en la celda.

> <span style="color=magenta">Nota</span>: Los <span style="color:limegreen">notebooks</span> o cuadernos son básicamente archivos de texto en formato <span style="color:limegreen">JSON</span> con extensión <span style="color:limegreen">.ipynb</span> (que proviene de *IPython Notebook*).  

El entorno de trabajo de <i>Google Colab</i> tiene la forma que se muestra en la Figura:

![Colab10](.\Latex_Curso_v8_2026\Unidad12a\Colab10.jpg)

En la parte superior izquierda se encuentra el nombre del notebook y su extensión <b>.ipynb</b>, que proviene de <i>IPython Notebook</i> y es un formato que permite ejecutar notebooks tanto en <i>IPython</i>, como en <b>Jupyter</b> y <i>Colab</i>. Por defecto, se utiliza el nombre <i>UntitledXX.ipynb</i>. Para cambiarlo, se debe seleccionar y escribir el nombre deseado en el cuadro de edición, como se muestra en la Figura:

![Colab11](.\Latex_Curso_v8_2026\Unidad12a\Colab11.jpg)

### <font color=#38bdf8>Celdas de Código</font>

Una celda de código es la unidad mínima de ejecución dentro de un notebook, es decir, es donde se incluye código para ejecutarlo. Para ejecutar el código dentro de una celda se pulsa el botón con el icono <b>play</b> que se encuentra en la izquierda o pulsar las teclas <b>Control + Enter</b> (ejecutar celda) o <b>Shift + Enter</b> (ejecutar celda y saltar a la siguiente). Después de la ejecución, debajo de la celda se mostrará el resultado correspondiente (si es que la celda lo tiene). Para comenzar a probarlo se escribirá la tradicional instrucción ``Hola mundo`` (ver Figura).

![Colab15](.\Latex_Curso_v8_2026\Unidad12a\Colab15.jpg)

### <font color=#38bdf8>Celdas de Texto</font>

Aunque en realidad <span style="color:limegreen">Markdown</span> también se considera un lenguaje que tiene la finalidad de permitir crear contenido de una manera sencilla de escribir, y que en todo momento mantenga un diseño legible, así que para simplificar se puede considerar a <span style="color:limegreen">Markdown</span> como un método de escritura.

## <font color=#38bdf8>Sintaxis de Markdown</font>

Ahora desarrollaremos la sintaxis de <span style="color:limegreen">Markdown</span> y los conceptos básicos para escribir texto utilizando este lenguaje de marcado. De hecho, al final del mismo estarás perfectamente capacitado para empezar a utilizar este lenguaje en tus escritos, publicaciones y prácticos.

<font color=#38bdf8>Párrafos y Saltos de Línea</font>

Para generar un nuevo párrafo en Markdown simplemente se separa el texto mediante una línea en blanco.

Al igual que sucede con HTML, Markdown no soporta dobles líneas en blanco, así que si se intenta estas se convertirán en una sola al procesarse.

Para realizar un salto de línea y empezar una frase en una línea siguiente dentro del mismo párrafo, se tendrá que pulsar dos veces la barra espaciadora antes de pulsar una vez <i>Enter</i>.

Por ejemplo si se quisiera escribir un poema Haiku quedaría así:

"Andando con sus patitas mojadas,  
el gorrión  
por la terraza de madera"  

Donde cada línea tiene dos espacios en blanco al final.

<font color=#38bdf8>Encabezados</font>

Las <span style="color:limegreen">#</span> numerales son uno de los métodos utilizados en Markdown para crear encabezados. Se debe utilizar añadiendo uno por cada nivel.

Es decir,

<span style="color:limegreen">#</span> Encabezado 1

<span style="color:limegreen">##</span> Encabezado 2

<span style="color:limegreen">###</span> Encabezado 3

<span style="color:limegreen">####</span> Encabezado 4

<span style="color:limegreen">#####</span> Encabezado 5

<span style="color:limegreen">######</span> Encabezado 6

por ejemplo:
# Práctica de Laboratorio 1

## Planteo del problema

## Situación 1

La situacion ...

## Situación 2

La situación ...

## <font color=#38bdf8>Citas</font>

Las citas se generar utilizando el carácter mayor que <span style="color:limegreen">></span> al comienzo del bloque de texto. por ejemplo:

<span style="color:limegreen">\></span> Un país, una civilización se puede juzgar por la forma en que trata a sus animales.  — \*Mahatma Gandhi\*

que da por resultado:

> Un país, una civilización se puede juzgar por la forma en que trata a sus animales.  — <i>Mahatma Gandhi</i>

Si la cita en cuestión se compone de varios párrafos, deberás añadir el mismo símbolo <span style="color:limegreen">></span> al comienzo de cada uno de ellos.

<span style="color:limegreen">\></span> Creo que los animales ven en el hombre un ser igual a ellos que ha perdido de forma extraordinariamente peligrosa el sano intelecto animal.

<span style="color:limegreen">\></span> Es decir, que ven en él al animal irracional, al animal que ríe, al animal que llora, al animal infeliz. — \*Friedrich Nietzsche\*

También se pueden colocar dos símbolos (<span style="color:limegreen">>></span>) para crear citas anidadas.

> Creo que los animales ven en el hombre un ser igual a ellos que ha perdido de forma extraordinariamente peligrosa el sano intelecto animal.

> Es decir, que ven en él al animal irracional, al animal que ríe, al animal que llora, al animal infeliz. — <i>Friedrich Nietzsche</i>

<font color=#38bdf8>Estilo de Texto</font>

El Lenguaje <i>Markdown</i> utiliza asteriscos (* ó _) para enfatizar texto.

Simplemente se tienen que encerrar las palabras u oraciones con éstos símbolos para conseguir letras en *cursiva* (\*cursiva\*) o **negrita** (\*\*negrita\*\*), como se indica en la siguiente Tabla:

Enfasis | Símbolo
---|---
Negrita | ** ó _ _
Cursiva | * ó _
Cursiva y Negrita | *** ó _ _ _

Por ejemplo:

    __Este texto está en negrita__

    _Este texto está en cursiva_

    ***Este texto está en cursiva y negrita***

Cuyo resultado es:

**Este texto está en negrita**

_Este texto está en cursiva_

***Este texto está en cursiva y negrita***

### <font color=#38bdf8>Otra Forma de Enfatizar Texto</font>

En <i>Markdown</i> también es posible enfatizar texto a través del uso de colores, para esto se debe escribir el texto en formato <i>HTML</i> según la siguiente sintaxis:

    <span style="color:red">Este texto debe aparecer en rojo</span>

que da por resultado:

<span style="color:red">Este texto debe aparecer en rojo</span>

## <font color=#38bdf8>Listas</font>

Existen dos tipos de Listas:

+ Listas con bullets

+ Listas numeradas

### <font color=#38bdf8>Listas con bullets</font>

Se utiliza el símbolo <span style="color:limegreen">+</span> para crear un item dentro de una lista.
Para generar listas anidadas, simplemente se tiene que añadir cuatro espacios en blanco antes del siguiente <span style="color:limegreen">+</span>.

    + Elemento de lista 1

    + Elemento de lista 2
    
        + Elemento de lista 3

    + Elemento de lista 4

        + Elemento de lista 5

            + Elemento de lista 6

Que da por resultado:

+ Elemento de lista 1
+ Elemento de lista 2
    + Elemento de lista 3
+ Elemento de lista 4
    + Elemento de lista 5
        + Elemento de lista 6

### <font color=#38bdf8>Listas Numeradas</font>

Para crear listas ordenadas (numeradas) se debe utilizar la sintaxis de tipo: <span style="color:limegreen">número</span> seguido de un punto, <span style="color:limegreen">(.)</span>. Al igual que ocurre con las listas con bullets, también se pueden anidar o combinar.

    1. Elemento de lista 1
    1.  Elemento de lista 2  
        + Elemento de lista 3
        + Elemento de lista 4
            1. Elemento de lista 5
            2. Elemento de lista 6

Este ejemplo produce:
1. Elemento de lista 1
1.  Elemento de lista 2  
    + Elemento de lista 3
    + Elemento de lista 4
        1. Elemento de lista 5
        2. Elemento de lista 6

## <font color=#38bdf8>Links o Enlaces</font>

Añadir enlaces a una publicación, más que común, hoy en día es algo casi obligatorio. Con <i>Markdown</i> se puede realizar de la siguiente manera.

Se crean escribiendo la palabra o texto del enlace entre <span style="color:limegreen">[ ]</span> corchetes, y el link en cuestión entre <span style="color:limegreen">( )</span> paréntesis. Esto se puede realizar en línea con el texto o en un párrafo separado. También se puede colocar en negrita o cursiva e incluir en los encabezados.

Por ejemplo:

    [**Sitio Oficial de Python**](https://www.python.org/)

que produce:

[**Sitio Oficial de Python**](https://www.python.org/)

Otro tipo de enlaces son las referencias a secciones dentro del texto que se está escribiendo. 

    Aquí hay [un enlace a otro lugar][otro lugar].
    Aquí hay [otro enlace más][otro-enlace].
    Y de nuevo regresamos al [primer enlace][otro lugar].

    [otro lugar]: https://colab.research.google.com/
    [otro-enlace]: https://www.python.org/

Que se renderiza como:

Aquí hay [un enlace a otro lugar][otro lugar].  
Aquí hay [otro enlace más][otro-enlace].  
Y de nuevo regresamos al [primer enlace][otro lugar].  

[otro lugar]: https://colab.research.google.com/  
[otro-enlace]: https://www.python.org/  

Las "referencias" anteriores son los segundos pares de corchetes: [otro lugar] y [otro-enlace]. Al final de un documento en Markdown, esos corchetes están definidos de forma apropiada como enlaces a sitios externos. Una ventaja del tipo de enlace en referencia es que múltiples enlaces al mismo sitio sólo necesitan ser actualizados una sola vez. Por ejemplo, si decidimos hacer que todo el [otro lugar] enlace a otro sitio, sólo tendremos que cambiar un solo enlace de la referencia.

Los enlaces de referencia no aparecen en el archivo final renderizado desde el formato Markdown. Están definidos dándoles el mismo nombre de etiqueta encerrada entre corchetes, seguido de dos puntos y seguido del enlace.
