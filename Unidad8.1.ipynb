# La POO y Python

Como ya se ha estudiado, una clase es un nuevo tipo de datos que usted o alguna otra persona crea para resolver un problema particular. Una vez que se ha creado una clase, cualquier persona puede utilizarla sin conocer los detalles de su funcionamiento, o incluso la forma en que se ha construido. Se va hacer uso de las clases como si sólo fueran otro tipo de datos predefinido disponible para su uso en un programa.

Los objetos idénticos, exceptuando su estado durante la ejecución del programa, se agrupan en clases de objetos y en Python se usa la palabra reservada <span style="color:limegreen">class</span> para su definición. Crear nuevos tipos de datos (clases) es un concepto fundamental en la programación orientada a objetos. <span style="color:magenta">Se debe puntualizar que en Python no existe el concepto de clases o tipos de datos abstractos</span>. Los tipos de datos (clases) trabajan casi exactamente como tipos predefinidos: se pueden crear variables de ese tipo (llamados objetos o instancias) y manipular estas variables (llamado envío de mensajes o peticiones; es decir se envía un mensaje y el objeto decide qué hacer con él).

Por eso, lo que hace realmente un programa orientado a objetos en Python es crear nuevos tipos de datos. Prácticamente todos los lenguajes de programación orientados a objetos usan la palabra reservada <span style="color:limegreen">class</span>. Cuando vea la palabra <span style="color:limegreen">class</span>, piense en <span style="color:limegreen">type</span> y viceversa. Pero, ¿cómo se consigue que un objeto haga algo, como utilizar objetos? Debe haber una forma de hacer una petición al objeto para que haga algo, como completar una operación, dibujar algo en la pantalla, etc. Y cada objeto puede satisfacer sólo ciertas peticiones. Las peticiones a las que puede responder un objeto están definidas por su intefaz.

La interfaz establece las peticiones que se pueden hacer a un objeto particular. Sin embargo, se debe codificar en algún sitio para satisfacer esta petición. Ésta, junto con los datos ocultos, constituyen la implementación. Desde el punto de vista de la programación procedural, no es complicado. Una clase tiene una función asociada para cada posible petición, y cuando se hace una petición particular a un objeto, se invoca a esa función. Este proceso normalmente se resume diciendo que se ha <span style="color:magenta">enviado un mensaje (se ha hecho una petición)</span> a un objeto, y el objeto sabe qué hacer con este mensaje (ejecuta un código). Cada clase se representa con una caja, con el nombre de la clase en la parte superior, los atributos que necesite describir en la parte central de la caja, y los métodos (las funciones o comportamientos que pertenecen a este objeto, que reciben cualquier mensaje que se envíe al objeto) en la parte inferior de la caja.

## Inicialización y Destrucción de Objetos

En Python el concepto de inicialización y limpieza es esencial para facilitar el uso de las bibliotecas y eliminar muchos de los errores sutiles que ocurren cuando el programador que usa esas bibliotecas olvida cumplir con sus actividades.

### Constructor

De acuerdo a lo ya estudiado, se sabe que los objetos son el "equivalente en software" de las cosas que nos rodean en el mundo real, y es lógico entender el por qué de la necesidad de que algo o alguien se encargue de dar, a los atributos de un objeto, los valores iniciales que se requieren para poder enfrentar su ciclo de vida.

Toda clase correctamente diseñada dispone de un grupo de elementos fundamentales que hacen a su funcionalidad y operación básica. El principal de ellos es un método muy particular que se denomina <span style="color:magenta">Constructor de la Clase (o simplemente constructor)</span> y su función es inicializar los atributos propios de cada objeto en el momento que se crea. El constructor de una clase en Python permite garantizar que los atributos de un nuevo tipo de datos ("objetos de la clase") siempre se inicialicen correctamente.

En Python, el nombre del constructor es <span style="color:limegreen">\_\_init\_\_</span> y es invocado automáticamente en el momento de la creación del objeto.

Como cualquier función, el constructor puede tomar argumentos que permitan especificar cómo ha de crearse el objeto, dados unos valores iniciales. Los argumentos del constructor son una especie de garantía de que todas las partes del objeto se inicializan con valores apropiados. La tarea del constructor es asignar a los atributos internos de cada objeto los valores iniciales adecuados. Otra característica es que siempre el primer argumento del constructor debe ser <span style="color:limegreen">self</span>, que hace referencia al propio objeto.

Por ejemplo, si la clase <span style="color:yellow">Forma</span> tiene un constructor que toma como argumento tres parámetros: un string que se refiere al color de la forma y dos valores enteros que se refieren a la posición <span style="color:yellow">x</span> e <span style="color:yellow">y</span> en el plano, entonces para crear un objeto <span style="color:yellow">figu</span> del tipo <span style="color:yellow">Forma</span> se debe hacer:

```python
#Definición de la clase Forma
class Forma():
    def __init__(self, color, x, y): # Constructor
        # Inicialización de atributo
        self.x = x
        self.y = y
        self.color = color
```

```python
# Instancia un objeto figu de color rojo
# en la posición x=10 e y=20 
figu = Forma('Rojo', 10, 20)
#...
```

## Encapsulamiento

El <span style="color:magenta">encapsulamiento</span> en POO es un concepto que hace referencia al ocultamiento del estado interno de un objeto hacia el exterior. Es decir, <span style="color:magenta">encapsular</span> consiste en hacer que los atributos o métodos internos de un objeto no se puedan acceder ni modificar desde el exterior directamente, sino que se pueda hacer <span style="color:magenta">solo por los servicios propuestos por el objeto (la interfaz)</span>.

El encapsulamiento permite definir niveles de visibilidad para los elementos de la clase. Estos niveles de visibilidad definen los derechos de acceso a los datos en función de si se accede a ellos mediante un método de la propia clase, de una clase heredada o de cualquier otra clase.

> Python <span style="color:red">NO POSEE</span> niveles de visibilidad para los atributos o métodos. Por defecto <span style="color:red">no oculta</span> los atributos y métodos de una clase, siempre son <span style="color:magenta">públicos</span>. Se pueden definir atributos y/o métodos <span style="color:magenta">protegidos</span> o <span style="color:magenta">privados</span> pero solo como una convención.

```python
# Ejemplo: POO 8

class Vehiculo():
    def __init__(self, nombre, color):
        self.nombre = nombre
        self.color = color 
        self.largo = 3.4
        self.ancho = 1.8
        print(f'Soy {self.nombre}, de color {self.color}  y he sido creado')

    #define un método
    def arrancar(self):
        print(f'Soy {self.nombre} y he arrancado')
```

```python
# Creación (instanciación) de un Objeto de tipo Vehículo
auto = Vehiculo('Ford', 'Rojo') 
print(auto.color)
print(f' Largo: {auto.largo}, Ancho: {auto.ancho}')
auto.arrancar() # invocación de un método
```

```python
auto.nombre = 'Chevrolet'  # acceso directo, no existe protección
print('Ahora el nombre es: ', auto.nombre)
auto.arrancar()
```

La manera de proteger los atributos y/o métodos en Python es la siguiente:

+ Un atributo o un método que comience con un símbolo de subrayado indica que dicho nombre es de uso interno y se denomina <span style="color:magenta">protegido</span>. Al importar un módulo con <span style="color:limegreen">from modulo import \*</span>, ningún nombre que comience con <span style="color:limegreen">_</span> será importado.

+ La manera de hacer que un atributo o un método sea <span style="color:magenta">privado</span> es anteponer al nombre un doble símbolo de subrayado. En realidad es un mecanismo engañoso ya que en realidad Python no posee niveles de protección y lo deja como responsabilidad del programador. Python usa <span style="color:magenta">name mangling</span> que consiste en modificar el nombre del atributo o método privado anteponiéndole un símbolo de subrayado y el nombre de la clase. Esto lo hace con el objetivo de resolver conflicto de nombres entre clases.

```python
# Ejemplo: POO 9
# Encapsulamiento
class Saludo():
    def __init__(self, nom: str, apel: str):
        self._nom = nom
        self.__apel = apel
        print(f'Soy {self._nom}, {self.__apel} y he sido instanciado')

    def __del__(self):
        print('Este es mi fin')

    def saludar(self):
        print(f'Hola {self._nom} {self.__apel}. Como estas?')
```

```python
sal = Saludo('Alberto', 'Ramírez') # Instanciación del objeto sal
sal.saludar()
```

```python
print('Accediendo a un atributo: ', sal._nombre)
print('Accediendo a un atributo privado: ', sal.__apellido)
```

```python
# Name Mangling
print('Accediendo a un atributo privado desde el exterior: ', sal._Saludo__apel)
```

```python
# Ejemplo: POO 10
class Vehiculo():
    def __init__(self, nombre, color):
        self.__nombre = nombre
        self.color = color 
        self.largo = 4
        self.ancho = 2.4
        print(f'Soy {self.__nombre} y he sido creado')
    
    #define un método privado
    def __arrancar(self):
        print(f'Soy {self.__nombre}, tengo el color {self.color} y he arrancado')
```

## Atributos de Objeto vs Atributos de Clase

La diferencia fundamental entre estos tipos de atributos es que los <span style="color:magenta">atributos de clase</span> son
compartidos por todas las instancias de esa clase, mientras que los <span style="color:magenta">atributos de instancia</span> son particulares para cada objeto creado con esa clase. Por lo tanto, los atributos de instancia son para datos únicos y propios para cada objeto y los atributos de clase son para aquellos valores que deban ser compartidos por todas las instancias de esa clase.

+  Los <span style="color:magenta">atributos de instancia</span> son creados usando la sintaxis <span style="color:limegreen">self.attr</span>. Como se ha dicho son locales para esa instancia y por lo tanto solo accesibles desde una de ellas. Es decir debe existir un
objeto instanciado para acceder al atributo.

 + En cuanto a su utilidad, genéricamente podemos decir que permiten
    definir un determinado <span style="color:magenta">"estado"</span> de ese objeto particular que lo
    diferencia de otros objetos de su clase. Por ejemplo, el atributo
    <span style="color:yellow">matrícula</span> en una clase <span style="color:yellow">Automóvil</span>.

 + Son almacenados en un diccionario (diccionario de instancia u objeto) al que se puede acceder mediante
<span style="color:limegreen">NombredeInstancia.\_\_dict\_\_</span>.

+ Los <span style="color:magenta">atributos de clase</span> se definen fuera de cualquier método
  de la clase, normalmente justo debajo del string de documentación.


  + Se pueden referenciar directamente desde la clase, es decir no
    necesitan de un objeto instanciado.
  +   Se pueden referenciar desde cualquier instancia de la clase.
  + Se usan comúnmente para establecer valores predeterminados comunes a todas las instancias, definir constantes, etc.
  + Son almacenados en un diccionario distinto (el diccionario de la clase) e
    independiente al usado para los atributos de instancia. Este diccionario de clase se puede consultar mediante <span style="color:limegreen">NombredeClase.\_\_dict\_\_</span> o desde una instancia con <span style="color:limegreen">NombredeInstancia.\_\_class\_\_.\_\_dict\_\_</span>.
 
  + Cuando se intenta asignar un nuevo valor al atributo de clase usando <span style="color:limegreen">instancia.atributo</span> o desde un método de instancia por medio de <span style="color:limegreen">self.atributo</span> no se modifica el atributo de clase, se crea un atributo de instancia con el mismo nombre y se le asigna el nuevo valor.


```python
# Ejemplo: POO 11
class Vehiculo():
    identidad = 'Soy un Vehiculo' # Atributo de clase
    def __init__(self, nombre): # Constructor
        self.nombre = nombre # Atributo de instancia

    # define un método
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')
```

```python
# Acceso desde la clase sin existencia de objeto
print(Vehiculo.identidad)
```

```python
# Instanciación de objeto
moto = Vehiculo('Honda')
print(moto.nombre)
print(moto.identidad)
```

```python
# Instanciación de otro objeto
automovil = Vehiculo('Ford')
print(automovil.nombre)
print(automovil.identidad)
```

```python
# Acceso y modificación de atributo de clase
Vehiculo.identidad = 'Soy un Vehículo distinto'
print(automovil.identidad, '\t', moto.identidad)
```

```python
# Modificación de atributo de clase desde una instancia
automovil.identidad = 'Soy un Vehículo de los costosos'
print(automovil.identidad, '\t', moto.identidad)
```

```python
print(f'\nDiccionario de instancia:\n{automovil.__dict__}')
print(f'\nDiccionario de Clase:\n {Vehiculo.__dict__}')
print(f'\nDiccionario de Clase a través de instancia:\n{automovil.__class__.__dict__}')
```

## Métodos de Objeto vs Métodos de Clase


+  Un <span style="color:magenta">método de instancia</span> es un método que modifica o accede al
  estado del objeto al que hace referencia. Recibe <span style="color:limegreen">self</span> como
  primer argumento, el cual se convierte en el propio objeto sobre el
  qué se está trabajando.

+  Un <span style="color:magenta">método de clase</span> es un método que modifica o accede al estado
  de la clase a la que hace referencia. Recibe <span style="color:limegreen">cls</span> como primer
  argumento, el cual se convierte en la propia clase sobre la que se
  está trabajando. La identificación de estos métodos se completa
  aplicando el decorador <span style="color:limegreen">@classmethod</span>.

```python
# Ejemplo: POO 12
class Vehiculo():
    
    contador = 0 # Atributo de clase
    
    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')

    @classmethod
    def cantidad_vehiculos(cls):
        print(f'{cls.contador} vehiculos creados hasta este momento!')
```

```python
# Creación de objetos
b1 = Vehiculo('Honda')
b2 = Vehiculo('Yamaha')
b1 = Vehiculo('Suzuki')
```

## Métodos Estáticos

+ Un <span style="color:magenta">método estático</span> es un método que no debería modificar el
  estado del objeto ni de la clase. No recibe ningún argumento especial
  (ni <span style="color:limegreen">self</span>, ni <span style="color:limegreen">cls</span>).

+  La identificación de estos métodos se completa aplicando el decorador
  <span style="color:limegreen">@staticmethod</span> al método.

Los métodos estáticos son útiles cuando se quiere definir una funcionalidad que pertenece a una clase pero que no depende de ningún atributo de instancia o de clase.

A continuación, se presentan algunas situaciones en las que se pueden utilizar los métodos estáticos:

+ Cuando se quiere implementar una funcionalidad que no depende de
  ningún atributo de la clase ni de una instancia en particular de la
  clase. Por ejemplo, un método que realiza una operación matemática
  simple, como calcular el valor absoluto de un número.

+ Cuando se quiere proporcionar una función auxiliar que se utiliza en
  varias partes del programa y que no está específicamente relacionada
  con alguna instancia de la clase. En este caso, se puede definir el
  método estático en la clase para tener una mejor organización del
  código.

+ Cuando se quiere crear un método de la clase que no requiere acceso a
  ningún atributo de instancia ni de clase, pero que se relaciona con la
  lógica de la clase. Por ejemplo, se puede utilizar un método estático
  para crear una función de ayuda que comprueba la validez de un string
  de entrada en relación con la clase.

```python
# Ejemplo: POO 13
class Vehiculo():

    contador = 0 # Atributo de clase

    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')

    @staticmethod
    def get_vehiculos_posibles():
        return ['moto','automóvil','tren','avión','bicicleta','barco']
```

## Métodos Decorados

+ Es posible que, según el escenario, se requiera <span style="color:magenta">decorar</span> ciertos
  métodos de una clase.

+ Esto es posible siguiendo la misma estructura de <span style="color:magenta">decoradores</span>
  que ya se ha estudiado, pero con ciertos matices.

+ En el ejemplo que sigue se crea un decorador que audita las acciones de un objeto para saber qué acción ha realizado.

```python
# Ejemplo: POO 14
class Vehiculo():
    contador = 0 # Atributo de clase
    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia
        self.pos = [0, 0] # Posicion inicial

    @staticmethod
    def auditar(method):
        def wrapper(self, *args, **kwargs):
            print(f'Vehiculo {self.nombre} ejecutando {method.__name__}')
            return method(self, *args, **kwargs)
        return wrapper

    @auditar
    def mover(self, x, y):
        self.pos[0] += x
        self.pos[1] += y
    
    @auditar
    def reset(self):
        self.pos = [0, 0]
```

```python
moto = Vehiculo('Honda')
moto.mover(10,10)
```

```python
print(moto.pos)
```

```python
moto.reset()
print(moto.pos)
```

## Métodos Mágicos

Una clase puede implementar ciertas operaciones que se invocan mediante una sintaxis especial (como operaciones aritméticas o subíndices y divisiones) definiendo métodos con nombres especiales. Este es el enfoque de Python para \emph{la sobrecarga de operadores}, permitiendo a las clases definir su propio comportamiento con respecto a los operadores del lenguaje.

+ Cuando se escribe el resultado de la operación <span style="color:yellow">'hola mundo' * 3</span> ¿cómo sabe el objeto string <span style="color:yellow">'hola mundo'</span> lo que debe hacer para multiplicarse con el objeto entero <span style="color:yellow">3?</span>. O dicho de otra forma, ¿cuál es la implementación del operador <span style="color:limegreen">\*</span> para <span style="color:limegreen">str</span> e <span style="color:limegreen">int</span>? En valores numéricos puede parecer evidente (siguiendo los operadores matemáticos), pero no es así para otros objetos.

+ La solución que proporciona Python para estas (y otras) situaciones
  son los <span style="color:magenta">métodos mágicos</span> también llamados métodos especiales.

+ Los métodos mágicos comienzan y terminan por doble símbolo de
  subrayado <span style="color:limegreen">\_\_</span>. Uno de los métodos mágicos más importantes es el
  constructor de una clase: <span style="color:limegreen">\_\_init\_\_()</span>.

+ La redefinición de operadores utilizando los métodos mágicos se
  denomina <span style="color:magenta">sobrecarga de operadores</span>.

+ Existe un método mágico (operador sobrecargado) asociado a cada operador (que se puede
  personalizar). Por ejemplo la comparación de dos objetos se realiza
  con el método <span style="color:limegreen">\_\_eq\_\_()</span>.

![Métodos Mágicos](./Figuras/Clases_metodosmagicos.png)

```python
# Ejemplo: POO 15
# from __future__ import annotations
class Vehiculo():
    def __init__(self, nombre, color, cant_cilindros):
        self.nombre = nombre # Atributo de instancia
        self.color = color
        self.cilindros = cant_cilindros

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')
    
    def __eq__(self, otro):
        return self.nombre == otro.nombre

    def __add__(self, other):
        n_nombre = self.nombre + '-' + other.nombre
        n_color = self.color + other.color
        n_cilindros = (self.cilindros + other.cilindros) / 2.
        return Vehiculo(n_nombre, n_color, n_cilindros)

    def __str__(self):
        return f'Vehiculo Marca: {self.nombre}, Color: {self.color}, Cilindros: {self.cilindros}'
```

```python
# Creación de objetos
b1 = Vehiculo('VW_Amarok', 'Gris', 4)
b2 = Vehiculo('VW_Amarok', 'Rojo', 6)
b3 = Vehiculo('Ford_Ranger', 'Azul', 4)
```

```python
# Invocación a __eq__
print(b1 == b2)
print(b1 == b3)
print(b1.__eq__(b3))
```

```python
# Invocación a __add__
b4 = b1 + b3
print(b4.nombre, b4.color, b4.cilindros)
```

## Gestores de Contexto

+ Otra de las aplicaciones interesantes de los métodos mágicos es la de los <span style="color:magenta">gestores de contexto</span>.

+ Un gestor de contexto permite aplicar una serie de acciones a la entrada y a la salida de un contexto definido.

+ Hay dos métodos que son utilizados para implementar los gestores de contexto:

    + <span style="color:limegreen">\_\_enter\_\_()</span>: Acciones que se llevan a cabo al entrar al contexto.
  
    + <span style="color:limegreen">\_\_exit\_\_(self, exception_type, exception_value, exception_traceback)</span>: Acciones que se llevan a cabo al salir del contexto.

```python
# Ejemplo: POO 16
# Medición de tiempos de ejecución

from time import time

class Timer():
    def __enter__(self):
        self.comienzo = time()

    def __exit__(self, tpo_ejec, valor_ejec, traza_ejec):
        # Se omitie el gestor de la excepcion
        self.final = time()
        tpo_ejec = self.final - self.comienzo
        print(f'Tiempo de ejecución [seg]: {tpo_ejec:.5f}')
```

```python
# Definimos un contexto
with Timer():
    for _ in range(1_000_000):
        x = 2 ** 20
```

```python
# Definimos otro contexto
with Timer():
    x = 0
    for _ in range(1_000_000):
        x += 2 ** 20
```

### Herencia Múltiple

+ Aunque no está disponible en todos los lenguajes de programación,

+ Si en una clase se hace referencia a un método o atributo que no

  que exista un conflicto en caso de que el método o el atributo buscado
  esté, a la vez, en varias clases base. En este caso, Python resuelve
  el conflicto a través del <span style="color:magenta">orden de resolución de métodos</span>.

+ Todas las clases en Python disponen de un método especial llamado
  <span style="color:limegreen">mro()</span> (<span style="color:magenta">"method resolution order"</span>) que devuelve una lista de las clases que se visitarían en caso de acceder a un método o a un
  atributo:

  + También se puede acceder a la misma información usando el atributo
    <span style="color:limegreen">\_\_mro\_\_</span>.
  
  + Todos los objetos en Python heredan, en primera instancia, de <span style="color:limegreen">Object</span>. Esto se puede comprobar con el correspondiente <span style="color:limegreen">mro()</span> de cada objeto.

```python
# Ejemplo: POO 22
# Herencia Múltiple
class Nombre:
    def __init__(self, nombre):
        self._nombre = nombre
```

## Mixins

Hay situaciones en las que interesa incorporar una <span style="color:magenta">clase base independiente</span> de la jerarquía establecida, y sólo a efectos de tareas auxiliares o transversales. Esta aproximación podría ayudar a evitar colisiones en los nombres de los métodos o atributos reduciendo la ambigüedad que añade la herencia múltiple. A estas clases auxiliares se las conoce como <span style="color:magenta">Mixins</span>.

Los mixins son clases que se definen en Python con la finalidad de poder integrar funcionalidad definida en una clase a otras clases diferentes.

Para esto se aprovecha que Python permite hacer uso de herencia múltiple, permitiendo agregar funcionalidad extra a clases sin la necesidad de tener que agregar nuevo código directamente en estas, sino que se hace utilizando la característica de la herencia.

Se llaman <span style="color:magenta">Mixin</span>, porque se mezclan (<span style="color:magenta">Mix-in</span>) en las clases ya existentes, entregándoles nueva funcionalidad sin alterar la estructura o jerarquía ya existente en ellas.

Imagine que se tienen 2 clases. <span style="color:yellow">Usuario</span> y <span style="color:yellow">Archivo</span>. Clases que, si bien no son del mismo tipo, comparten un método en común, el método <span style="color:yellow">to\_json()</span>.

```python
# Ejemplo: POO 24
import json 

class UsuarioBase():
    pass

class ArchivoBase():
    pass
```

```python
class Usuario(UsuarioBase):

    def __init__(self, email, password):
        self.email = email
        self.password = password

    def to_json(self):
        return json.dumps(self.__dict__)
```

```python
class Archivo(ArchivoBase):

    def __init__(self, path, extension):
        self.path = path
        self.extension = extension

    def to_json(self):
        return json.dumps(self.__dict__)
```

Para este ejemplo se tiene código duplicado. El método <span style="color:yellow">to\_json()</span> se encuentra en 2 clases diferentes. Lo cual no es algo bueno.

El tener código duplicado es un síntoma que el código se puede mejorar. En la gran mayoría de los casos realizar un buena abstracción del algoritmo solventará este problema. Entonces para solucionar esta duplicación de código se usan los <span style="color:magenta">Mixins</span>. La implementación con <span style="color:magenta">Mixins</span> toma la forma:

```python
# Ejemplo: POO 25
class JSONSerializableMixin:
    def to_json(self):
        return json.dumps(self.__dict__)
```

```python
class Usuario(UsuarioBase, JSONSerializableMixin):
    def __init__(self, email, password):
        self.email = email
        self.password = password
```

```python
class Archivo(ArchivoBase, JSONSerializableMixin):
    def __init__(self, path, extension):
        self.path = path
        self.extension = extension
```

Si bien no es necesario usar el sufijo <span style="color:magenta">Mixin</span>, es recomendable hacerlo para seguir un estandard de codificación y así el código será mucho más legible.

Utilizando la herencia múltiple que ofrece Python se puede extender dicho funcionamiento a una N cantidad de subclases. Con la clase <span style="color:magenta">Mixin</span> definida se pueden extender las funcionalidades, simplemente heredándola de ella.

## Iterables

+ Un objeto en Python se dice <span style="color:magenta">iterable</span> si implementa el protocolo de
  iteración. Este protocolo permite <span style="color:magenta">"entregar"</span> de a un valor por vez en forma de secuencia.

+ Hay muchos tipos de datos iterables en Python: <span style="color:magenta">strings</span>, <span style="color:magenta">listas</span>, <span style="color:magenta">tuplas</span>, <span style="color:magenta">conjuntos</span>, <span style="color:magenta">diccionarios</span>, etc.

+ Para ser un objeto iterable sólo es necesario implementar el método
  especial <span style="color:limegreen">\_\_iter\_\_()</span>. Este método debe proporcionar una
  referencia al objeto iterador que es quien se encargará de desarrollar
  el protocolo de iteración a través del método especial <span style="color:limegreen">\_\_next\_\_()</span>.

![Iterables](./Figuras/Protocolo_iteracion.png)

Cuando se utiliza una iteración <span style="color:limegreen">for</span> para recorrer los elementos de un iterable, ocurren varias cosas:

+ Se obtiene el objeto iterador del iterable.

+ Se realizan invocaciones sucesivas al método <span style="color:limegreen">\_\_next\_\_()</span> sobre dicho iterador para obtener cada elemento del iterable.

Se detiene la iteración cuando el iterador dispara la excepción <span style="color:limegreen">StopIteration</span>.

```python
# Ejemplo: POO 27
class Robot:
    def __init__(self, numero_serie):
        self.numero_serie = numero_serie * 5 # por que si
        
    def __repr__(self):
        return f'Robot: SN={self.numero_serie}.'
```

```python
class Fabrica:
    def __init__(self, numero_robots):
        self.robots = [Robot(str(i)) for i in 
                       range(numero_robots)]
        self.numero_robots = numero_robots
        self.puntero = 0
    
    def __iter__(self):
        return self # El iterador es el propio objeto
    
    def __next__(self):
        # Protocolo de la iteración
        if self.puntero >= self.numero_robots:
            raise StopIteration
        robot = self.robots[self.puntero]
        self.puntero += 1
        return robot
```

```python
# Funcionamiento
for robot in Fabrica(10):
    print(robot)
```

## Aspectos del Encapsulamiento en Python

Al momento de declarar una variable en el interior de una clase en un lenguaje de programación como "Java" o "C++" ya sea como un atributo o método se dispone de una palabra reservada para indicar si esta es <span style="color:magenta">privada (private)</span> o <span style="color:magenta">publica (public)</span>. En Python no existen los modificadores de acceso pero se logra indicar que un atributo es protegido o privado mediante el uso de los símbolos de subrayado (<span style="color:limegreen">\_</span>, <span style="color:limegreen">\_\_</span>).

El encapsulamiento permite agrupar datos y controlar su comportamiento en una clase. También permite controlar el acceso a los datos (atributos) y prevenir modificaciones no autorizadas.

Según Wikipedia, se define encapsulamiento como:

<span style="color:orange">Definición:</span>
<span style="color:magenta">En programación modular, y más específicamente en programación orientada a objetos, se denomina encapsulamiento al ocultamiento del estado de un objeto, es decir, el ocultamiento de los atributos de un objeto de manera que solo se pueda modificar mediante las operaciones definidas para ese objeto</span>.

<span style="color:magenta">Cada objeto está aislado del exterior y protege a los datos asociados de un objeto contra su modificación por quien no tenga derecho a acceder a ellos, eliminando efectos secundarios e interacciones</span>.

<span style="color:magenta">De esta forma el usuario de la clase puede obviar la implementación de los métodos y propiedades para concentrarse solo en cómo usarlos. Por otro lado se evita que el usuario pueda cambiar su estado de maneras imprevistas e incontrolada</span>.

Entonces, el encapsulamiento se define como la acción de <span style="color:magenta">ocultar el estado</span> o sea de <span style="color:magenta">ocultar los atributos de un objeto</span>. Se habla de ocultar los atributos o métodos, más específicamente de protegerlos para que solo puedan ser modificados mediante <span style="color:magenta">operaciones definidas</span> para ello. Esto asegura por ejemplo que <span style="color:magenta">no se pueda modificar un atributo si no es a través de un método que se haya creado específicamente para ello</span> y aquí es donde nacen los famosos métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span>, <span style="color:magenta">Deleter</span>, etc.

Anteriormente se han creado clases con sus atributos y métodos públicos, es decir, que son accesibles desde su instancia. Basta con crear un objeto a partir de esa clase y se podrá acceder y modificar libremente sus atributos. Y no solo desde su instancia sino también desde otra clase que herede de la anterior, por lo que un atributo que almacena información sensible puede ser modificado fácilmente.

Se debe tener en cuenta que en Python (por su filosofía) las propiedades y métodos privados no existen, por lo que el acceso a ellos es total. Este rasgo establece que en Python no es obligatorio usar encapsulamiento, pero puede ser útil en el caso que se encuentre con código de otro desarrollador que si lo aplique, lo que es bastante normal.

### Atributos Protegidos en Python

Los atributos que comienzan con un símbolo de subrayado <span style="color:limegreen">\_</span>, se consideran <span style="color:magenta">protegidos</span> en Python. A menudo se encontrará con algo como lo planteado en el siguiente ejemplo:

```python
# Ejemplo: POO 33
#!/usr/bin/env python
# -*- coding: utf-8 -*-
#
class Usuario (object):
    def __init__(self, nombre, clave):
        self.nombre = nombre
        self._clave = clave
```

```python
usuario1 = Usuario ("Roberto", "qwerty")
print (f'({usuario1.nombre}, {usuario1._clave})')
```

Analizando el ejemplo se observa que el atributo <span style="color:yellow">clave</span> esta precedido por un guión bajo, lo que indica que es un atributo protegido. Lo cual establece que solo puede ser accedido por esa clase y sus subclases, es decir, aquellas que hereden de esa clase. Se suele ver muy a menudo el uso de esta técnica como una buena practica para atributos o métodos de uso interno y también para evitar la colisión de los mismos nombres de métodos o atributos causado por herencia. Siempre y cuando se este hablando de programas que recurren a muchas clases, se observará con mucha frecuencia. De lo contrario, resulta, en realidad, innecesario.

### Atributos Privados en Python

Los atributos que comienzan con un doble símbolo de subrayado <span style="color:limegreen">\_\_</span>, se consideran <span style="color:magenta">privados</span> en Python. En el caso de un atributo privado se esta indicando que este solo podrá ser accedido o modificado si se especifica la clase precedida por un guión bajo seguida del atributo precedido por doble guión bajo (<span style="color:magenta">Name mangling</span>).

```python
# Ejemplo: POO 34
#!/usr/bin/env python
# -*- coding: utf-8 -*-
#
class Usuario (object):
    def __init__(self, nombre, clave):
        self.nombre = nombre
        self.__clave = clave
```

```python
usuario1 = Usuario ("Roberto", "qwerty")
print (f'({usuario1.nombre}, {usuario1.__clave})')
```

Para evitar la excepción que resulta de la ejecución del código anterior, se debe utilizar <span style="color:magenta">name mangling</span>, esto es anteponer al nombre del atributo un símbolo de subrayado con el nombre de la clase, de la siguiente manera:

```python
# Ejemplo: POO 35
#!/usr/bin/env python
# -*- coding: utf-8 -*-
#
class Usuario (object):
    def __init__(self, nombre, clave):
        self.nombre = nombre
        self.__clave = clave
```

```python
usuario1 = Usuario ("Roberto", "qwerty")
print (f'({usuario1.nombre}, {usuario1._Usuario__clave})')
```

Se puede concluir que se puede acceder a un atributo por más que sea privado y modificarlo de la misma manera. Pero no es lo que se <span style="color:magenta">considera correcto</span>. Por lo que para ello si se desean implementar métodos que permitan modificar estos atributos de la forma que se suele hacer en otros lenguajes donde se aplica <span style="color:magenta">encapsulamiento</span> se puede hacer utilizando los métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span>, <span style="color:magenta">Deleter</span> o mediante el uso del decorador <span style="color:limegreen">@property</span>.

### Los Métodos Getter y Setter

En esta publicación y en muchos artículos se encontrará el término <span style="color:magenta">código pitónico</span>, que hace referencia a buenas prácticas de programación en Python. 

Si se tiene conocimiento de un lenguaje de programación como "Java" o "C++", probablemente se esté acostumbrado a escribir métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span> y <span style="color:magenta">Deleter</span> para cada atributo de las clases. Estos métodos permiten acceder y modificar atributos privados mientras se mantiene el encapsulamiento. En Python, normalmente se expondrán los atributos como parte de la API (Interfaz de Programación de Aplicaciones) pública y se usarán propiedades cuando se necesiten atributos con comportamiento funcional.

Aunque las <span style="color:magenta">propiedades</span> (<span style="color:limegreen">@property</span>) son el camino a utilizar en Python, se pueden tener algunos inconvenientes prácticos. Debido a esto, se encontrarán algunas situaciones en las que los métodos <span style="color:magenta">Getters</span> y <span style="color:magenta">Setters</span> son preferibles a las propiedades.

Cuando se define una clase en programación orientada a objetos (POO), es probable que se tengan algunos atributos de instancia y otros de clase. Estos atributos son variables a las que se puede acceder a través de una instancia de la clase, a través de la clase o de ambas.

Los atributos contienen el estado interno de los objetos. En muchos casos, se necesitará acceder y modificar este estado, lo que implica acceder y modificar los atributos. Normalmente, existen al menos dos formas de acceder y modificar los atributos. 

En Python, se puede:

+ Acceder y cambiar él atributo directamente, debido a que son por definición <span style="color:magenta">públicos</span>.

+ Utilizar métodos para acceder al atributo y modificarlo.

Si se exponen los atributos de una clase a los usuarios, entonces esos atributos se convierten automáticamente en parte de la API pública de la clase. Es decir, serán atributos públicos, lo que significa que los usuarios podrán acceder y modificar directamente los valores de esos atributos en el código.

Tener un atributo que forma parte de la API de una clase se convertirá en un problema si se necesita modificar la implementación interna del propio atributo. Un ejemplo claro de este problema es cuándo se quiere convertir un atributo almacenado en uno calculado. Un atributo almacenado responderá inmediatamente a las operaciones de acceso y modificación simplemente recuperando y almacenando datos, mientras que un atributo calculado ejecutará cálculos antes de dichas operaciones.

El problema con los atributos normales es que no pueden tener una implementación interna porque son sólo variables. Por lo tanto, cambiar la implementación interna de un atributo requerirá convertir el atributo en un método, lo que probablemente modificará el código de los usuarios. ¿Por qué? Porque se tendrá que cambiar el acceso a los atributos y se tendrán que cambiar las operaciones de modificación en invocaciones a métodos en todo el código si se quiere que él mismo siga funcionando.

Para hacer frente a este tipo de problemas, algunos lenguajes de programación, como "Java" y "C++", requieren que se proporcionen métodos para manipular los atributos de las clases. Estos métodos se conocen comúnmente como métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span>.

### Que son los Métodos Getter y Setter

Los métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span> y <span style="color:magenta">Deleter</span> son bastante populares en muchos lenguajes de programación orientados a objetos. Por lo tanto, es bastante probable que ya se hayan estudiado. Como definición aproximada, se puede decir que los métodos <span style="color:magenta">Getters</span>, <span style="color:magenta">Setters</span> y <span style="color:magenta">Deleter</span> son:

+ <span style="color:magenta">Getter</span>: Un método que permite acceder (leer) a un atributo en una clase.

+ <span style="color:magenta">Setter</span>: Un método que permite asignar o modificar el valor de un atributo en una clase.

+ <span style="color:magenta">Deleter</span>: Un método que permite borrar o eliminar el valor de un atributo en una clase.

En la programación orientada a objetos, los patrones <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span> y <span style="color:magenta">Deleter</span> sugieren que los atributos públicos directos sólo deben utilizarse cuando se está seguro de que nunca se necesitará asignarles un comportamiento. Si por el contrario existe alguna posibilidad que un atributo cambie su implementación interna en algún momento, se deben utilizar los métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span> o <span style="color:magenta">Deleter</span>.

Implementar estos patrones requiere:

+ Hacer que los atributos no sean públicos.
+ Escribir los métodos <span style="color:magenta">Getter</span>, <span style="color:magenta">Setter</span> y <span style="color:magenta">Deleter</span> para cada atributo.

Por ejemplo, supóngase que se necesita escribir una clase <span style="color:yellow">Label</span> con los atributos <span style="color:yellow">text</span> y <span style="color:yellow">font</span>. Si se tuviera que utilizar métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> para gestionar estos atributos, entonces se escribiría la clase de la siguiente manera:

```python
# Ejemplo: POO 36

class Label:
    def __init__(self, text, font):
        self._text = text
        self._font = font

    def get_text(self):
        return self._text

    def set_text(self, value):
        self._text = value

    def get_font(self):
        return self._font

    def set_font(self, value):
        self._font = value
```

En este ejemplo, el constructor de la clase <span style="color:yellow">Label</span> toma dos argumentos, <span style="color:yellow">text</span> y <span style="color:yellow">font</span>. Estos argumentos se almacenan en los atributos de instancia no públicos <span style="color:yellow">.\_text</span> y <span style="color:yellow">.\_font</span>, respectivamente.

A continuación, se definen los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> para ambos atributos. Normalmente, los métodos <span style="color:magenta">Getter</span> devuelven el valor de un atributo, mientras que los métodos <span style="color:magenta">Setter</span> toman un nuevo valor y lo asignan al atributo subyacente.

```python
# Ejemplo: POO 37
#from label import Label

rotulo = Label("Hola Python", "Arial")
print(rotulo.get_text())
```

```python
rotulo.set_text("Ahora Otro Python")
print(rotulo.get_text())
print(rotulo.get_font())
```

La clase <span style="color:yellow">Label</span> oculta sus atributos del acceso público y en su lugar expone los métodos getter y setter <span style="color:yellow">get\_xx()</span> y <span style="color:yellow">set\_xx()</span>. Se pueden utilizar estos métodos cuando se necesite acceder o modificar los atributos de la clase, que no son públicos y, por lo tanto, no forman parte de la API de la clase.

### Como surgieron los Getter y Setter

Para entender el origen de estos métodos, se revisará el ejemplo de la clase <span style="color:yellow">Label</span>. Suponga que se desea almacenar automáticamente el texto en mayúsculas. Desafortunadamente, no se puede simplemente añadir este comportamiento a un atributo normal como <span style="color:yellow">text</span>. Sólo se puede añadir un comportamiento a través de métodos, pero convertir un atributo público en un método introducirá un cambio de última hora en la API.

Entonces, ¿qué se  puede hacer? Bueno, en Python, lo más probable es que se utilice una propiedad, como se estudiará mas adelante. Sin embargo, lenguajes de programación como "Java" y "C++" no soportan construcciones del tipo <span style="color:limegreen">@property</span>, o sus propiedades no son como las de Python.

Por eso estos lenguajes animan a que nunca se expongan los atributos como parte de las APIs públicas. En su lugar, se deben proporcionar métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span>, que ofrecen una forma rápida de cambiar la implementación interna de los atributos sin cambiar la API pública.

El encapsulamiento es otro tema fundamental relacionado con el origen de los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span>. Esencialmente, este principio se refiere a agrupar los datos con los métodos que operan sobre esos datos. De esta forma, las operaciones de acceso y modificación se harán a través de métodos exclusivamente.

El principio también tiene que ver con la restricción del acceso directo a los atributos de un objeto, lo que evitará exponer detalles de implementación o violar la invariancia del estado.

Para dotar a la clase <span style="color:yellow">Label</span> de la nueva funcionalidad requerida en "Java" o "C++", se deben utilizar métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> desde el principio. ¿Cómo se puede aplicar el patrón <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> para resolver el problema en Python?

Considérese la siguiente versión de la clase <span style="color:yellow">Label</span>:

```python
# Ejemplo: POO 38

class Label:
    def __init__(self, text, font):
        self.set_text(text)
        self.set_font(font)

    # métodos Getter y Setter
    def get_text(self):
        return self._text

    def set_text(self, value):
        self._text = value.upper()  # nuevo comportamiento
    
    def get_font(self):
        return self._font

    def set_font(self, value):
        self._font = value
```

En esta versión actualizada de la clase <span style="color:yellow">Label</span>, se proporcionan los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> para los atributos <span style="color:yellow">text</span> y <span style="color:yellow">font</span>. El atributo que contiene el texto no es público porque tiene un guión bajo inicial en su nombre, <span style="color:yellow">.\_text</span>. El método <span style="color:magenta">Setter</span> realiza la modificación de la entrada, convirtiendo el texto en mayúsculas.

Ahora se puede utilizar la clase <span style="color:yellow">Label</span> como en el siguiente fragmento de código:

```python
# Ejemplo: POO 39
#from label import Label

rotulo = Label("Frutas", "Arial")
print(rotulo.get_text())
```

```python
rotulo.set_text("Vegetales")
print(rotulo.get_text())
```

Muy bien. Hasta ahora, se ha añadido con éxito el comportamiento requerido al atributo <span style="color:yellow">text</span> de la clase <span style="color:yellow">Label</span>. Ahora el método <span style="color:magenta">Setter</span> tiene un verdadero objetivo en lugar de simplemente asignar un nuevo valor al atributo <span style="color:yellow">text</span>.

Aunque los patrones <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> son bastante común en otros lenguajes de programación, ese no es el caso en Python.

Añadir métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> a las clases puede aumentar considerablemente el número de líneas de código. Estos métodos también siguen un patrón repetitivo y aburrido que requerirá tiempo extra para completarse. Además, estos patrones son una fuente potencial de errores y además son demasiado tediosos. También se encontrará que la funcionalidad inmediata obtenida de todo este código adicional es a menudo casi nula.

Todo esto suena como algo que los desarrolladores de Python no desean hacer en su código. En Python, probablemente se escribirá la clase <span style="color:yellow">Label</span> como en el siguiente fragmento:

```python
# Ejemplo: POO 40

class Label:
    def __init__(self, text, font):
        self.text = text
        self.font = font
```

```python
# En forma directa
rotulo.text = "Vegetales"
print(rotulo.text)
```

Aquí, <span style="color:yellow">.text</span> y <span style="color:yellow">.font</span> son atributos públicos y están expuestos como parte de la API de la clase. Esto significa que los usuarios pueden cambiar su valor cuando lo deseen.

Hacer públicos los atributos, como en el ejemplo anterior, es una práctica común en Python. En estos casos, cambiar a patrones <span style="color:magenta">Getters</span> y <span style="color:magenta">Setters</span> introducirá cambios importantes. Entonces, ¿cómo lidiar con situaciones que requieren añadir comportamiento a los atributos? La forma pitónica de hacerlo es sustituir los atributos por propiedades.

### Usando Propiedades

La forma pitónica de agregar un comportamiento a un atributo es convertir el propio atributo en una propiedad. Las propiedades agrupan métodos para obtener, asignar y/o modificar, borrar y documentar los datos subyacentes. Por lo tanto, las propiedades son atributos especiales con un comportamiento adicional.

Las propiedades se pueden utilizar del mismo modo que los atributos normales. Cuando se accede a una propiedad, se invoca automáticamente a su método <span style="color:magenta">Getter</span> interno o adjunto. Del mismo modo, cuando se cambia la propiedad, se invoca a su método <span style="color:magenta">Setter</span> interno o adjunto. Este comportamiento proporciona los medios para agregar funcionalidad a los atributos sin introducir cambios en la API del código.

Como ejemplo de la manera en que las propiedades pueden ayudar a agregar comportamiento a los atributos, suponga que se necesita una clase <span style="color:yellow">Empleado</span> como parte de un sistema de gestión de empleados. Se comienza con la siguiente implementación básica:


```python
# Ejemplo: POO 41

class Empleado:
    def __init__(self, nombre, fecha_nac):
        self.nombre = nombre
        self.fecha_nac = fecha_nac

    # Implementacion...
```

El constructor de esta clase toma dos argumentos, el nombre y la fecha de nacimiento del empleado en cuestión. Estos atributos se almacenan directamente en dos atributos de instancia, <span style="color:yellow">.nombre</span> y <span style="color:yellow">.fecha\_nac</span>.

Se puede comenzar a utilizar la clase inmediatamente, por ejemplo con el siguiente fragmento de código:

```python
# Ejemplo: POO 42
#from empleado import Empleado

trabajador1 = Empleado("Juan", "2001-02-07")
print(f'Empleado: {trabajador1.nombre}')
print(f'Fecha de nacimiento: {trabajador1.fecha_nac}')
```

```python
trabajador1.nombre = "Juan Perez"
print(f'Empleado: {trabajador1.nombre}')
```

La clase <span style="color:yellow">Empleado</span> permite crear instancias que dan acceso directo a los atributos <span style="color:yellow">nombre</span> y <span style="color:yellow">fecha\_nac</span>. Tener en cuenta que también se pueden cambiar los atributos utilizando asignaciones directas.

A medida que el proyecto evoluciona, este adquiere nuevas funcionalidades. Por ejemplo, se necesita ahora almacenar el nombre del empleado en mayúsculas y convertir la fecha de nacimiento en un objeto de tipo <span style="color:limegreen">date</span>. Para cumplir con estas nuevas exigencias sin modificar la API con métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> para los atributos <span style="color:yellow">.nombre</span> y <span style="color:yellow">.fecha\_nac</span>, se pueden utilizar <span style="color:magenta">propiedades</span>. Las propiedades se pueden escribir de la siguiente manera:

+ Se utiliza el decorador <span style="color:limegreen">@property</span> y a continuación se define un método con el mismo nombre del atributo sin guiones. Este método se comporta como el método <span style="color:magenta">Getter</span>.

+ Se genera una propiedad que puede modificar el valor de un atributo definiendo un método con el nombre del atributo sin guiones y utilizando como decorador el nombre del atributo sin guiones seguido de <span style="color:limegreen">.setter</span>.


```python
# Ejemplo: POO 43
from datetime import date

class Empleado:
    def __init__(self, nombre, fecha_nac):
        self.nombre = nombre
        self.fecha_nac = fecha_nac
        
    @property
    def nombre(self):
        return self._nombre

    @nombre.setter
    def nombre(self, valor):
        self._nombre = valor.upper()

    @property
    def fecha_nac(self):
        return self._fecha_nac
        
    @fecha_nac.setter
    def fecha_nac(self, valor):
        self._fecha_nac = date.fromisoformat(valor)
```

En esta versión mejorada de la clase <span style="color:yellow">Empleado</span>, se convierte a <span style="color:yellow">.nombre</span> y <span style="color:yellow">.fecha\_nac</span> en propiedades utilizando el decorador <span style="color:limegreen">@property</span>. Ahora cada atributo tiene un método <span style="color:magenta">Getter</span> y un método <span style="color:magenta">Setter</span> internos con el nombre del propio atributo. Debido a las propiedades, los desarrolladores de Python tienden a diseñar las APIs de sus clases usando unas pocas directrices:

+ Utilizar atributos públicos siempre que sea apropiado, incluso si se espera que él atributo requiera un comportamiento funcional en el futuro.

+ Evitar definir métodos <span style="color:magenta">Setter</span> y <span style="color:magenta">Getter</span> para los atributos. Siempre se pueden convertir en propiedades si es necesario.

+ Utilizar propiedades cuando se necesite agregar comportamiento a los atributos y seguir utilizándolos como atributos normales en el código.

+ Evitar los efectos secundarios en las propiedades, porque nadie espera que operaciones cómo las asignaciones tengan efectos secundarios.

Las propiedades de Python son una potente herramienta. Por eso, los programadores tienden a usarlas en exceso. En general, sólo se debería usar propiedades cuando se necesite añadir un procesamiento extra sobre un atributo específico. Convertir todos los atributos en propiedades será una pérdida de tiempo. También puede implicar problemas de rendimiento y mantenimiento.

El siguiente ejemplo refuerza los conceptos sobre las propiedades en Python.

```python
# Ejemplo: POO 45
# Encapsulamiento        

class ListadoBebidas(): 
    bebidas_validas = ['Gin Tonic', 'Cerveza', 'Fernando']
    
    def __init__(self):
        self.bebida = 'Gin Tonic'

    @property
    def bebida(self):
        return f'La bebida oficial es {self.__bebida}'

    @bebida.setter
    def bebida(self, bebida):
        if bebida in ListadoBebidas.bebidas_validas:
            self.__bebida = bebida
        else:
            raise ValueError(f'La bebida {bebida} no está en el listado de bebidas válidas')
```

+ En el ejemplo se declaran dos atributos, uno llamado <span style="color:yellow">bebida</span> y una lista, atributo de clase llamada <span style="color:yellow">bebidas\_validas</span>.

+ Para recuperar la información de la variable <span style="color:yellow">bebida</span> se hace con el nombre del objeto y el nombre del método <span style="color:yellow">bebida()</span>. Como se aprecia, se puede manipular el resultado para mostrar la información de la manera deseada.

+ Para poder modificar el valor del atributo, se realiza una asignación directa <span style="color:yellow">fiesta.bebida = 'Limonada'</span>, esto invocará al método <span style="color:magenta">Setter</span> y se ejecutará la acción definida en él. En el caso del ejemplo se dispara una excepción, ya que la bebida que se suministra no está en el listado de bebidas válidas.

### Descriptores

Los descriptores son una característica avanzada de Python que permiten crear atributos con comportamientos funcionales adjuntos en las clases. Para crear un descriptor, se necesita usar el protocolo de descriptores, específicamente los métodos especiales <span style="color:limegreen">.\_\_get\_\_()</span> y <span style="color:limegreen">.\_\_set\_\_()</span>.

Los descriptores son similares a las propiedades. De hecho, una propiedad es un tipo especial de descriptor. Es decir, los descriptores normales son más potentes que las propiedades y pueden ser reutilizados a través de diferentes clases.

Para ilustrar cómo utilizar descriptores para crear atributos con comportamiento funcional, suponga que se necesita continuar desarrollando la clase <span style="color:yellow">Empleado</span>. Esta vez, se necesita agregar un atributo para almacenar la fecha en la que un empleado comenzó a trabajar para la empresa:


```python
# Ejemplo: POO 47

from datetime import date
class Empleado:
    def __init__(self, nombre, fecha_nac, fecha_inicio):
        self.nombre = nombre
        self.fecha_nac = fecha_nac
        self.fecha_inicio = fecha_inicio

    @property
    def nombre(self):
        return self._nombre

    @nombre.setter
    def nombre(self, valor):
        self._nombre = valor.upper()

    @property
    def fecha_nac(self):
        return self._fecha_nac

    @fecha_nac.setter
    def fecha_nac(self, valor):
        self._fecha_nac = date.fromisoformat(valor)

    @property
    def fecha_inicio(self):
        return self._fecha_inicio

    @fecha_inicio.setter
    def fecha_inicio(self, valor):
        self._fecha_inicio = date.fromisoformat(valor)
```

En esta actualización, se ha agregado otra propiedad a la clase <span style="color:yellow">Empleado</span>. Esta nueva propiedad le permitirá gestionar la fecha de inicio en la empresa de cada empleado. De nuevo, la propiedad convierte la fecha de un string a un objeto fecha.

Esta clase funciona como se esperaba. Sin embargo, comienza a hacerse repetitiva y aburrida. Así que se decide reescribir la clase. Si se observa se está haciendo la misma operación en ambos atributos relacionados con la fecha, entonces se decide utilizar un <span style="color:magenta">descriptor</span> para empaquetar la funcionalidad repetitiva:

```python
# Ejemplo: POO 48
from datetime import date

class Fecha:
    def __set_name__(self, owner, nombre):
        self.__nombre = nombre
        
    def __get__(self, instance, owner):
        print(f'Método __get__:{instance.__dict__[self.__nombre]}')
        return instance.__dict__[self.__nombre]
        
    def __set__(self, instance, valor):
        print(f'Método __set__')
        instance.__dict__[self.__nombre] = date.fromisoformat(valor)
```

```python
class Empleado:
    fecha_nac = Fecha() # Atributos de clase
    fecha_inicio = Fecha()

    def __init__(self, nombre, fecha_nac, fecha_inicio):
        self.nombre = nombre
        print(f'Constructor...')
        self.fecha_nac = fecha_nac
        self.fecha_inicio = fecha_inicio
        print(f'Fin Constructor...')
        
    @property
    def nombre(self):
        return self.__nombre

    @nombre.setter
    def nombre(self, valor):
        self.__nombre = valor.upper()
```

```python
tb1 = Empleado('Pepe Ebar', '1973-02-11', '2005-03-08')
fec = tb1.fecha_nac
print(tb1.fecha_nac)
print(tb1.fecha_inicio)
```

```python
tb1.fecha_nac = '2022-04-24'
print(tb1.fecha_nac)
```

```python
tb2 = Empleado('Carla Gutierrez', '1991-05-06', '2010-09-02')
print(tb2.fecha_nac)
print(tb1.fecha_nac)
```

Como se puede apreciar, este código es más limpio y menos repetitivo que su versión anterior. En esta actualización, se crea un descriptor <span style="color:yellow">Fecha</span> para gestionar los atributos relacionados con la fecha. El descriptor tiene un método <span style="color:limegreen">.\_\_set\_name\_\_()</span> que almacena automáticamente el nombre del atributo. También tiene métodos <span style="color:limegreen">.\_\_get\_\_()</span> y <span style="color:limegreen">.\_\_set\_\_()</span> que funcionan como <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> del atributo, respectivamente.

Las dos implementaciones de la clase <span style="color:yellow">Empleado</span> en esta sección funcionan de forma similar.

En general, si el programa posee un gran número de clases con definiciones de propiedades similares, entonces se debería considerar usar descriptores en su lugar.

### Los Métodos setattr y getattr

Otra forma de reemplazar los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> en Python es usar los métodos especiales <span style="color:limegreen">.\_\_setattr\_\_()</span> y <span style="color:limegreen">.\_\_getattr\_\_()</span> para gestionar los atributos. Considerar el siguiente ejemplo en el que se define una clase <span style="color:yellow">Punto</span>. La clase convierte automáticamente las coordenadas de entrada en números reales:

```python
# Ejemplo: POO 49
class Punto:

    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __getattr__(self, nombre: str):
        return self.__dict__[f'_{nombre}']

    def __setattr__(self, nombre, valor):
        self.__dict__[f'_{nombre}'] = float(valor)
```

El constructor de la clase <span style="color:yellow">Punto</span> toma dos coordenadas, <span style="color:yellow">x</span> e <span style="color:yellow">y</span>. El método <span style="color:limegreen">.\_\_getattr\_\_()</span> devuelve la coordenada representada por el nombre. Para ello, el método utiliza el diccionario del espacio de nombres de instancia, <span style="color:limegreen">.\_\_dict\_\_</span>. Tener en cuenta que el nombre final del atributo tendrá un guión bajo precediendo a lo que se pase en nombre. Python invoca automáticamente a <span style="color:limegreen">.\_\_getattr\_\_()</span> cada vez que se accede a un atributo de la clase <span style="color:yellow">Punto</span> usando la notación punto.

El método <span style="color:limegreen">.\_\_setattr\_\_()</span> agrega o actualiza atributos. En este ejemplo, <span style="color:limegreen">.\_\_setattr\_\_()</span> opera sobre cada coordenada y la convierte en un número real utilizando la función <span style="color:limegreen">float()</span>. De nuevo, Python invoca a <span style="color:limegreen">.\_\_setattr\_\_()</span> cada vez que ejecuta una operación de asignación sobre cualquier atributo de la clase contenedora.

El siguiente ejemplo ilustra cómo funciona esta clase en la práctica:

```python
# Ejemplo: POO 50

#from punto import Punto
punto = Punto(21, 42)
print(f'Coordenada x: {punto.x}')
print(f'Coordenada y: {punto.y}')
```

```python
punto.x = 84
print(f'Coordenada x: {punto.x}')
print(dir(punto))
```

La clase <span style="color:yellow">Punto</span> convierte automáticamente los valores de coordenadas en números reales. Se puede acceder a las coordenadas, <span style="color:yellow">x</span> e <span style="color:yellow">y</span>, como se haría con cualquier otro atributo normal. Sin embargo, las operaciones de acceso y modificación pasan por <span style="color:limegreen">.\_\_getattr\_\_()</span> y <span style="color:limegreen">.\_\_setattr\_\_()</span>, respectivamente.

Tener en cuenta que la clase <span style="color:yellow">Punto</span> permite acceder a las coordenadas como atributos públicos. Sin embargo, los almacena cómo atributos no públicos. Se puede confirmar este comportamiento con la función <span style="color:limegreen">dir()</span>.

El ejemplo de esta sección es un poco exótico, y probablemente no se utilizará algo similar en un código. Sin embargo, las herramientas que se han utilizado en el ejemplo permiten realizar validaciones o transformaciones en el acceso y modificación de atributos, al igual que lo hacen los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span>.

En cierto sentido, <span style="color:limegreen">.\_\_getattr\_\_()</span> y <span style="color:limegreen">.\_\_setattr\_\_()</span> son una especie de implementación genérica del patrón <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span>. Internamente, estos métodos funcionan como <span style="color:magenta">Getters</span> y <span style="color:magenta">Setters</span> que soportan el acceso regular a atributos en Python.

### Cuando Usar Getters y Setters

En la codificación de objetos del mundo real, se encontrarán algunos casos de uso en los que los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> pueden ser preferibles a las propiedades, aunque las propiedades son generalmente el camino a seguir en Python.

Por ejemplo, los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> pueden ser más adecuados para situaciones en las que se necesite:

+ Ejecutar transformaciones costosas (en el sentido de tiempo de ejecución) en el acceso o en la modificación de atributos.

+ Tomar argumentos extra y variables banderas.

+ Utilizar herencia.

+ Generar excepciones relacionadas con el acceso y modificación de atributos.

+ Facilitar la integración en equipos de desarrollo heterogéneos.

#### Evitar propiedades Lentas

Se debería evitar ocultar operaciones lentas detrás de una propiedad en Python. Los usuarios de las APIs esperan que el acceso y la modificación de atributos se realicen como el acceso y la modificación de variables normales. En otras palabras, los usuarios esperan que estas operaciones ocurran instantáneamente y sin efectos secundarios.

Alejarse demasiado de esa expectativa hará que la API resulte extraña y desagradable de usar, violando el <span style="color:magenta">principio de la menor sorpresa</span>.

Además, si los usuarios acceden y modifican repetidamente los atributos en una repetición, entonces el código puede implicar demasiada sobrecarga, lo que puede producir enormes e inesperados problemas de rendimiento.

Por el contrario, los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> hacen explícito que el acceso o la modificación de un atributo determinado se produce a través de una invocación a un método. De hecho, los usuarios sabrán que invocar a un método puede consumir tiempo, y el rendimiento del código puede variar significativamente debido a ello.

Explicitar estos hechos en las APIs puede ayudar a minimizar la sorpresa de los usuarios cuando accedan y modifiquen los atributos en el código.

En resumen, sí se va a utilizar una propiedad para gestionar un atributo, asegúrese de que los métodos detrás de la propiedad son eficientes y no causan efectos secundarios. Por el contrario, sí se está tratando con métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> lentos, entonces es mejor usar los métodos <span style="color:magenta">Getters</span> y <span style="color:magenta">Setters</span> tradicionales sobre las propiedades.

#### Argumento Bandera

A diferencia de las propiedades de Python, los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> permiten un acceso y modificación de atributos más flexible. Por ejemplo, suponga que se dispone de una clase <span style="color:yellow">Persona</span> con un atributo <span style="color:yellow">.fecha\_nac</span>. Este atributo debe ser constante durante la vida de una persona. Por lo tanto, se decide que el atributo sea de sólo lectura.

Sin embargo, como existe el error humano, se enfrentará a casos en los que alguien cometa un error al introducir la fecha de nacimiento de una persona determinada. Se puede resolver este problema proporcionando un método <span style="color:magenta">Setter</span> que tome un indicador <span style="color:yellow">forzar</span>, como en el ejemplo siguiente:

```python
# Ejemplo: POO 51

class Persona:
    def __init__(self, nombre, fecha_nac):
        self.nombre = nombre
        self._fecha_nac = fecha_nac

    def get_fecha_nac(self):
        return self._fecha_nac

    def set_fecha_nac(self, valor, forzar = False):
        if forzar:
            self._fecha_nac = valor
        else:
            raise AttributeError("No se puede modificar la fecha de nacimiento")
```

En este ejemplo se puede observar que se proporcionan los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> tradicionales para el atributo <span style="color:yellow">.fecha\_nac</span>. El método <span style="color:magenta">Setter</span> toma un argumento extra llamado <span style="color:yellow">forzar</span>, que permite forzar la modificación de la fecha de nacimiento de una persona.

Así es como funciona esta clase:

```python
# Ejemplo: POO 52

#from persona import Persona
pr1 = Persona("Juana Perez", "2000-11-29")
print(f'Persona: {pr1.nombre}')
print(f'Fecha de Nacimiento: {pr1.get_fecha_nac()}')
```

```python
# pr1.set_fecha_nac("2000-10-29") # Da mensaje de error
pr1.set_fecha_nac("2000-10-29", forzar = True)
print(f'Persona: {pr1.get_fecha_nac()}')
```

```python
try:
    pr1.set_fecha_nac("2000-10-12", forzar = True)
    print(f'Persona: {pr1.get_fecha_nac()}')
except AttributeError:
    print('Mi excepción')
```

Cuando se intenta modificar la fecha de nacimiento de <span style="color:yellow">pr1</span> utilizando <span style="color:magenta">.set\_fecha\_nac()</span> sin colocar <span style="color:yellow">forzar</span> en <span style="color:limegreen">True</span>, se obtiene una excepción del tipo <span style="color:limegreen">AttributeError</span> indicando que el atributo no puede ser modificado. Por el contrario, si se coloca <span style="color:yellow">forzar</span> en <span style="color:limegreen">True</span>, se podrá actualizar la fecha de nacimiento de <span style="color:yellow">pr1</span> para corregir cualquier error que se haya producido al introducir la fecha.

Es importante tener en cuenta que las propiedades de Python no aceptan argumentos extra en sus métodos <span style="color:magenta">Setter</span>. Sólo aceptan el nuevo valor a colocar o actualizar.

#### Uso de Herencia

Un problema con las propiedades de Python es que no funcionan bien en escenarios de herencia. Por ejemplo, suponga que se necesita extender o modificar el método <span style="color:magenta">Getter</span> de una propiedad en una subclase. En la práctica, no hay forma segura de hacerlo. No se puede simplemente sobreescribir el método <span style="color:magenta">Getter</span> y esperar que el resto de la funcionalidad de la propiedad siga siendo la misma que en la clase base.

Este problema se produce porque los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> están ocultos dentro de la propiedad. No se heredan independientemente sino como un todo. Por lo tanto, cuando se sobreescribe el método <span style="color:magenta">Getter</span> de una propiedad heredada de una clase padre, se sobreescribe toda la propiedad, incluyendo su método <span style="color:magenta">Setter</span> y el resto de sus componentes internos.

Como ejemplo, considere la siguiente jerarquía de clases:

```python
# Ejemplo: POO 53

class Persona:
    def __init__(self, nombre):
        self.nombre = nombre

    @property
    def nombre(self):
        return self._nombre

    @nombre.setter
    def nombre(self, valor):
        self._nombre = valor
```

```python
class Empleado(Persona):
    @property
    def nombre(self):
        return super().nombre.upper()
```

En este ejemplo, se anula el método <span style="color:magenta">Getter</span> de la propiedad <span style="color:yellow">.nombre</span> en <span style="color:yellow">Persona</span>. De esta manera, implícitamente se está sobrescribiendo toda la propiedad <span style="color:yellow">.nombre</span>, incluyendo su funcionalidad <span style="color:magenta">Setter</span>:

```python
# Ejemplo: POO 54

#from persona import Empleado
emp1 = Empleado("Juana")
print(f'Empleado: {emp1.nombre}')
```

```python
emp1.nombre = "Juana Perez"
```

En este caso <span style="color:yellow">.nombre</span> es una propiedad de sólo lectura porque el método <span style="color:magenta">Setter</span> de la clase base no fue heredado sino que fue sobreescrito por una propiedad completamente nueva. Para resolver este problema de herencia se deben usar los métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> tradicionales:

```python
# Ejemplo: POO 55

class Persona:
    def __init__(self, nombre):
        self._nombre = nombre

    def get_nombre(self):
        return self._nombre

    def set_nombre(self, valor):
        self._nombre = valor
```

```python
class Empleado(Persona):
    def get_nombre(self):
        return super().get_nombre().upper()
```

En esta versión de la clase <span style="color:yellow">Persona</span> se proporcionan métodos <span style="color:magenta">Getter</span> y <span style="color:magenta">Setter</span> independientes. <span style="color:yellow">Empleado</span> subclase de <span style="color:yellow">Persona</span>, sobreescribe el método <span style="color:magenta">Getter</span> para el atributo nombre. Este hecho no afecta al método <span style="color:magenta">Setter</span>, que <span style="color:yellow">Empleado</span> hereda correctamente de su clase base, <span style="color:yellow">Persona</span>.

Así es como funciona esta nueva versión de <span style="color:yellow">Empleado</span>:

```python
# Ejemplo: POO 56

#from persona import Empleado
juana = Empleado("Juana")
print(f'Empleado: {juana.get_nombre()}')
```

```python
juana.set_nombre("Juana Perez")
print(f'Empleado: {juana.get_nombre()}')
```

Ahora <span style="color:yellow">Empleado</span> es completamente funcional. El método <span style="color:magenta">Getter</span> anulado funciona como se espera. El método <span style="color:magenta">Setter</span> también funciona porque fue heredado exitosamente de <span style="color:magenta">Persona</span>.
