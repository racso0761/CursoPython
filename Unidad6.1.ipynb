### <font color=#38bdf8>Funciones Recursivas</font>

Las <span style="color:magenta">funciones recursivas</span> son funciones que se invocan a si mismas:

+ Consumen mas tiempo que el enfoque tradicional.

+ Existe un numero limitado de invocaciones recursivas para no consumir todos los recursos del sistema (el <span style="color:yellow">namespace</span>).

+ Se deben utilizar con mucha precaución y sólo en aquellos casos en el que el *algoritmo sea recursivo por naturaleza* y además poseer una condición de finalización.

Suponga se desea calcular el <span style="color:magenta">factorial</span> de un número que se define como:

$$
n! = \left\{ \begin{array}{l}
n * (n -1)!\\
0! = 1
\end{array} \right .
$$

```python
# Ejemplo: Funciones Recursivas 1
# El Factorial
def fac(n):
    if n == 0:
        return 1
    else:
        return n * fac(n -1)
```

```python
n = 5
print(f'Factorial de {n} = {fac(n)}')
```

```python
for i in range(0, 15):
    print(fac(i), end=' ')
```

Suponga ahora que se desea generar la <span style="color:magenta">serie de fibonacci</span> que se define como:

$$
fibonacci(n) = \left \{ \begin{array}{l}
fibonacci(n-1) + fibonacci(n -2)\\
fibonacci(0) = 0\\
fibonacci(1) = 1\\
\end{array} \right .
$$

```python
# Ejemplo: Funciones Recursivas 2

# La serie de Fibonacci
def fibonacci(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    return fibonacci(n - 1) + fibonacci(n - 2)
```

```python
serie = fibonacci(10)
print(f'Fibonacci de 10: {serie}')
```

```python
serie = fibonacci(20)
print(f'Fibonacci de 20: {serie}')
```

```python
for i in range(0, 20):
    print(fibonacci(i), end=' ')
```

### <font color=#38bdf8>Clausuras</font>

+ En Python, las funciones son <span style="color:magenta">ciudadanos de primera clase</span>. Esto quiere decir que es posible asignar a una variable una función, de igual forma, una función puede ser utilizada como argumento o puede ser retornada por otra función.

+ <h3 style="font-size: 14;">Una <span style="color:yellow">clausura</span> (del término inglés <span style="color:magenta">closure</span>) establece el uso de una <span style="color:magenta">función interior</span> que se genera dinámicamente y recuerda los valores de los argumentos con los que fue creada</h3>

+ En una clausura <span style="color:orange">se devuelve una función</span>, <span style="color:magenta">no una invocación a una función</span>.

```python
# Ejemplo: Clausuras 1

def hacer_multiplicador(n):
    def multiplicador(x):
        return x * n
    return multiplicador
```

```python
m3 = hacer_multiplicador(3)
```

```python
m5 = hacer_multiplicador(5)
```

```python
print(type(m3))
```

```python
print(m3(7))
```

```python
print(m5(8))
```

```python
# Ejemplo: Clausuras 2
# invocación directa
print(hacer_multiplicador(6)(5))
```

### <font color=#38bdf8>Decoradores</font>

+ Existen situaciones en las que se necesita modificar el comportamiento de funciones existentes pero sin alterar su código. Para estos casos es muy útil usar <span style="color:magenta">decoradores</span>.

+ Un <span style="color:magenta">decorador</span> es una función que recibe como argumento otra función, la modifica y devuelve una función diferente. Se podría interpretar como un caso particular de una <span style="color:magenta">clausura</span>.

+ El decorador es como una envoltura de la función original. 

Sintaxis básica:

```python
# Ejemplo: Decoradores 1
def mi_decorador(func):
    def wrapper(*args, **kwargs):
        # Algún código anterior a la invocación de func
        return func(*args, **kwargs)
    # Algún codigo previo a finalizar el decorador
    return wrapper
```

donde:
+ <span style="color:limegreen">mi_decorador</span>: es el nombre del decorador.

+ <span style="color:limegreen">wrapper</span>: es una función interior (convención de nombre).

+ <span style="color:limegreen">func</span>: función a decorar (convención de nombre).

+ <span style="color:limegreen">*args</span>: Argumentos posicionales (convención de nombre).

+ <span style="color:limegreen">**kwargs</span>: Argumentos nominales (convención de nombre).

```python
# Ejemplo: Decoradores 2

# Convierte el resultado numérico de una función a binario
def res_a_binario(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        return bin(resultado)
    return wrapper
```

```python
# función que calcula x^n
def mi_potencia(x: int, n: int) -> int:
    return x ** n
```

```python
print(mi_potencia(2, 3))
```

```python
print(mi_potencia(2, 10))
```

```python
# Se aplica el decorador
# res_a_binario es la funcion decoradora
# mi_potencia es la funcion a decorar
potencia_decorada = res_a_binario(mi_potencia)
```

```python
print(potencia_decorada(2, 3))
```

```python
print(potencia_decorada(2, 10))
```

#### <font color=#38bdf8>Uso de <span style="color:limegreen">@</span> para Decorar</font>

Para facilitar el uso de <span style="color:yellow">decoradores</span> Python ofrece el operador <span style="color:limegreen">@</span>.

+ Debe colocarse justo antes de la definición de la función que se desea decorar.
+ A continuación sin linea en blanco debe ir la definición de la función a decorar.

```python
# Ejemplo: Decorador 3

# Convierte el resultado numérico de una función a binario
def res_a_binario(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        return bin(resultado)
    return wrapper
```

```python
# función que calcula x^n
# Se aplica el decorador
@res_a_binario
def mi_potencia(x: int, n: int) -> int:
    return x ** n
```

```python
print(mi_potencia(2, 3))
```

```python
print(mi_potencia(2, 10))
```

#### <font color=#38bdf8>Decoradores y Argumentos</font>

+ Se ha visto un ejemplo de decorador que trabaja sobre el resultado de la función decorada, pero nada impide que se trabaje sobre los argumentos que se le pasan a la función decorada.

+ Suponga un escenario en el que se implementan funciones que trabajan con dos operandos y se desea asegurar de que esos operandos sean números enteros.

+ Se sabe positivamente que las funciones a decorar trabajan con dos operados (dos argumentos) entonces se puede definir la función interior <span style="color:limegreen">wrapper(value1, value2)</span> con dos argumentos, en vez de con un número indeterminado de los mismos.

```python
# Ejemplo: Decoradores 4

def asegura_entero(func):
    def wrapper(value1:int, value2: int, /) -> int | float | None:
        if isinstance(value1, int) and isinstance(value2, int):
            return func(value1, value2)
        return None
    return wrapper
```

```python
@ asegura_entero
def suma(a, b):
    return a + b
```

```python
# Comportamiento
resultado = suma(3, 4)
print(resultado)
```

```python
resultado = suma(3, 'a')
print(resultado)
```

```python
resultado = suma('a', 'b')
print(resultado)
```

```python
resultado = suma(3, 3.4)
print(resultado)
```

#### <font color=#38bdf8>Decoradores Múltiples</font>

+ Se puede aplicar más de un decorador a una función.

+ Se aplican desde adentro hacia afuera

```python
# Ejemplo: Decoradores 5

def mas5(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        return resultado + 5
    return wrapper
```

```python
def div2(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        return resultado // 2
    return wrapper
```

```python
# Se aplican los decoradores
@mas5
@div2
def prod(a, b):
    return(a * b)
```

```python
a = prod(3, 4) # ((3 * 4) // 2) + 5
print(a)
```

```python
# Ejemplo: Decoradores 6 (Como funciona)

def mas5(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        print(f'1: {resultado = }')
        print('Ahora mas5')
        return resultado + 5
    return wrapper
```

```python
def div2(func):
    def wrapper(*args, **kwargs):
        resultado = func(*args, **kwargs)
        print(f'2: {resultado = }')
        print('Ahora div2')
        return resultado // 2
    return wrapper
```

```python
# Se aplican los decoradores
@mas5
@div2
def prod(a, b):
    return(a * b)
```

```python
a = prod(3, 4) # ((3 * 4) // 2) + 5
print(a, '\n')
```

```python
# Como funciones:
def prod2(a, b):
    return a * b
```

```python
b = mas5(div2(prod2)) 
print(b(3, 4))
```

<h2 style="color:#38bdf8;color:#38bdf8;padding:25px;">4️⃣ Enfoque Funcional</h2>
</div>

+ Python es un lenguaje de <span style="color:yellow">programación multiparadigma</span>.

+ Uno de los paradigmas menos explotados en este lenguaje es la <span style="color:limegreen">programación funcional</span>.

La programación funcional en Python es un paradigma que trata de escribir código usando <span style="color:yellow">funciones puras</span>, evitando modificar el estado y reduciendo el uso de variables mutables. Aunque Python no es un lenguaje puramente funcional, ofrece muchas herramientas para trabajar de esta forma.

### <font color=#38bdf8>Principios clave</font>

+ <span style="color:yellow">Funciones puras</span>: siempre devuelven el mismo resultado para los mismos argumentos y no tienen efectos secundarios.
+ <span style="color:yellow">Inmutabilidad</span>: no modificar datos existentes, sino crear nuevos.
+ <span style="color:yellow">Funciones de orden superior</span>: reciben funciones como argumentos o devuelven funciones.
+ <span style="color:yellow">Expresiones en lugar de sentencias</span>: preferir map, filter, reduce y list comprehensions en lugar de bucles imperativos.

Python ofrece 3 funciones que encajan verdaderamente bien en este enfoque:

+ <span style="color:limegreen">map()</span>, 
+ <span style="color:limegreen">filter()</span> y 
+ <span style="color:limegreen">reduce()</span>.


### <font color=#38bdf8>La función map()</font>

+ Esta función <span style="color:magenta">aplica otra función</span> sobre cada elemento de un objeto-iterable.

+ La función <span style="color:limegreen">map()</span> devuelve un <span style="color:magenta">generador</span> no una lista.

![Función map()](./Figuras/map.png)

```python
# Ejemplo: función map

# Calcula la función x^2/2 para una serie de valores
def f(x):
    return x ** 2 / 2
```

```python
datos = range(1, 11)
print(datos)
```

```python
generados = map(f, datos)
print(type(generados))
```

```python
lista = list(generados)
print(lista)
```

```python
# Otra forma
g = map(lambda x: x**2/2, datos)
print(list(g))
```

### <font color=#38bdf8>La Función filter()</font>

+ Esta función <span style="color:magenta">selecciona</span> aquellos elementos de un objeto-iterable que cumplen una determinada condición.

+ La función <span style="color:limegreen">filter()</span> devuelve un <span style="color:magenta">generador</span>, no una lista.

![Función filter()](./Figuras/filter.png)

```python
# Ejemplo: función filter

# Selecciona solo aquellos valores impares dentro de un rango
def numero_impar(x):
    return x % 2 == 1
```

```python
datos = range(1, 21)

generados = filter(numero_impar, datos)
lista = list(generados)
```

```python
print(lista)
```

```python
# Otra forma
g = filter(lambda x: x%2 == 1, datos)
print(list(g))
```

### <font color=#38bdf8>La Función reduce()</font>

+ Para poder usar esta función debemos usar el módulo <span style="color:limegreen">functools</span>.

+ Nos permite aplicar una función dada sobre todos los elementos de un objeto-iterable de manera acumulativa.

+ formato de la función <span style="color:limegreen">reduce(función, iterable[, valor_inicial])</span>

```python
# Ejemplo: Función reduce()
# Calcula el producto de una serie de valores.
from functools import reduce

def multiplica_valores(a, b):
    return a * b
```

```python
datos = range(1, 5)

# Calcula ((1*2)*3)*4
print(reduce(multiplica_valores, datos))
```

```python
print(reduce(lambda a, b: a * b, datos))
```

```python
# Ejemplo práctico de programación Funcional
from functools import reduce

# Función pura: no depende de variables externas ni modifica nada
def cuadrado(x):
    return x * x

# Lista de números
numeros = [1, 2, 3, 4, 5]

# map: aplica una función a cada elemento
cuadrados = list(map(cuadrado, numeros))

# filter: filtra elementos según una condición
pares = list(filter(lambda x: x % 2 == 0, cuadrados))

# reduce: acumula valores (por ejemplo, suma)
suma_total = reduce(lambda a, b: a + b, pares, 0)

print("Números originales:", numeros)
print("Cuadrados:", cuadrados)
print("Pares:", pares)
print("Suma de pares:", suma_total)
```

### Problema:

Tenemos una lista de ventas (producto, cantidad, precio unitario). Queremos:

1. Filtrar solo los productos vendidos más de 2 veces.
1. Calcular el total por producto.
1. Ordenar de mayor a menor los totales.
1. Obtener el total global de ventas

```python
from functools import reduce
from itertools import groupby
from operator import itemgetter

# Datos de ejemplo: (producto, cantidad, precio_unitario)
ventas = [
    ("Laptop", 1, 1200),
    ("Mouse", 3, 25),
    ("Teclado", 2, 45),
    ("Laptop", 2, 1150),
    ("Mouse", 1, 30),
    ("Monitor", 4, 200),
    ("Teclado", 1, 50)
]

ventas
```

```python
# 1) Filtrar productos vendidos + de 2 veces
ventas_filtradas = list(filter(lambda v: v[1] > 2, ventas))
print("Ventas filtradas:", ventas_filtradas)
```

```python
# 2) Agrupar por producto y calcular total por producto
# Primero ordenamos por nombre para que groupby funcione
ventas_ordenadas = sorted(ventas, key=itemgetter(0))
ventas_ordenadas
```

```python
# groupby agrupa por producto
totales_por_producto = [
    (producto, reduce(lambda acc, x: acc + x[1] * x[2], grupo, 0))
    for producto, grupo in groupby(ventas_ordenadas, key=itemgetter(0))
]
totales_por_producto
```

```python
# 3) Ordenar de mayor a menor total
totales_ordenados = sorted(totales_por_producto, key=itemgetter(1), reverse=False)
totales_ordenados
```

```python
# 4) Calcular total global
total_global = reduce(lambda acc, x: acc + x[1], totales_ordenados, 0)
total_global
```

<h2 style="color:#38bdf8;color:#38bdf8;padding:25px;">5️⃣ Consejos para Buenas Prácticas de Programación</h2>
</div> 

1. Las funciones deberían hacer una tarea única.     

    + Por ejemplo, un mal diseño sería programar una única función que calcule el total de una compra, los impuestos y los gastos de envío. Sin embargo esto se debería hacer con tres funciones separadas. Así se consigue que el código sea más fácil de mantener, reutilizar y depurar.
    + 
    

2. Utilizar nombres descriptivos y con significado. Los nombres autoexplicativos de variables y funciones mejoran la legibilidad del código.     

    + Por ejemplo – se debería llamar *costo_total* a una variable que se usa para almacenar el total de un carrito de compras en lugar de *x*, ya que claramente explica su propósito.
    + 
    

3. No usar variables globales. Las variables globales pueden introducir muchos problemas, incluyendo efectos colaterales inesperados y errores de programación difíciles de detectar y seguir. 

    + Suponga que se tienen dos funciones que comparten una variable global. Si una función cambia su valor la otra función podría no funcionar como se espera.
    + 
    

4. Actualizar regularmente. El código inevitablemente cambia con el tiempo, lo que puede derivar en partes obsoletas, redundantes o desorganizadas. Tratar de mantener la calidad del código revisando y documentando nuevamente aquellas zonas que se hayan editado.


5. No utilizar *números mágicos* o valores *hard-coded*.

    + No es lo mismo escribir *(99 * 3)* que *(precio * cantidad)*. Esto último es más fácil de entender y usa variables con nombres descriptivos haciéndolo autoexplicativo.
    + 
     

6. Escribir el código para lo que se necesita ahora, no lo que se piensa que se podría necesitar en el futuro. Los programas simples y centrados en el problema son más flexibles y menos complejos.


7. Usar docstrings y comentarios para explicar el *por qué* y no el *qué*. El código limpio es autoexplicativo y por lo tanto los comentarios no deberían usarse para explicar lo que hace el código. En cambio, los comentarios deberían usarse para proporcionar contexto adicional, como por qué el código está diseñado de una cierta manera.
