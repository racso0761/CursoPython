Las variables definen nombres (<span style="color:yellow">identificadores</span>) para los datos.

Reglas de Nombres de Variables:

+ Solo pueden contener los siguientes caracteres:
    + Letras minúsculas
    + Letras mayúsculas
    + Dígitos
    + Guión Bajo (subrayado)
+ Nunca comenzar con un <span style="color:yellow">dígito</span>.
+ No se pueden usar las palabras reservadas del lenguaje (<span style="color:yellow">keywords</span>).
+ Distinguen entre mayúsculas y minúsculas.
+ Normalmente se recomienda usar <span style="color:yellow">sustantivos</span>.
+ Se usa el estilo <span style="color:yellow">snake_case</span>.

```python
# Ejemplo: Variables 1
cantidad_estudios = 4
print("Cantidad:", cantidad_estudios)
```

## <font color=#38bdf8>Constantes o Literales</font>

+ Una constante o literal es un objeto que no cambia de valor en ningún momento a lo largo del programa.
+ Normalmente se utilizan <span style="color:yellow">sustantivos</span> en mayúsculas para dar nombre a las constantes.

```python
VELOCIDAD_LUZ = 300_000
print(f'Velocidad de la Luz:', VELOCIDAD_LUZ)
```

## <font color=#38bdf8>Otra Regla Universal Recomendada</font>

+ Usar sustantivos en mayúsculas para las constantes. (<span style="color:yellow">PI</span>)
+ Usar sustantivos para las variables. (<span style="color:yellow">precio</span>)
+ Usar verbos para las funciones. (<span style="color:yellow">conseguir_precio()</span>)
+ Usar adjetivos para las variables booleanas. (<span style="color:yellow">disponible</span>)</li>

## <font color=#38bdf8>Asignación</font>

### <font color=#38bdf8>Asignación Simple</font>

En Python se usa el símbolo <span style="color:yellow">=</span> para asignar un valor a una variable.  

![Asignación](./Figuras/asignacion.png)

```python
# Ejemplo: Asignación de variables

poblacion_sj = 785_000
temperatura = 12.8
nombre_ciudad = 'San Juan'
contador1 = contador2 = contador3 = 0
temperatura_actual = temperatura

print('Población San Juan:', poblacion_sj)
print('Temperatura:', temperatura)
print('Ciudad:', nombre_ciudad)
print('Contadores:', contador1, contador2, contador3)
```

### <font color=#38bdf8>Asignación Mejorada o Aumentada</font>

Es una mezcla entre una asignación y el uso de un operador.

![Asignación aumentada](./Figuras/asignacion2.png)

<i>Ejemplo</i>: Supongase que se tienen 100 vehiculos para la venta y que el mes pasado se han vendido 20 de ellos, analicemos este ejemplo:

```python
# Ejemplo: Asignación de Variables
# Asignacion Tradicional
total_autos = 100
autos_vendidos = 20
total_autos = total_autos - autos_vendidos
print('Total Autos Disponibles:', total_autos)
```

```python
# Ejemplo:Asignación de Variables
# Asignación Mejorada o Aumentada
total_autos = 100
autos_vendidos = 20
total_autos -= autos_vendidos
print('Total Autos Diponibles:', total_autos)
```

## <font color=#38bdf8>Dirección de Memoria de un Objeto</font>

Se utiliza la función <span style="color:yellow">id()</span>

```python
# Ejemplo: Variables 3
a = 4
b = a
print('Memoria:', id(a))
print('Memoria:', id(b))
print('Memoria:', id(4))
```

## <font color=#38bdf8>Tipos de Datos Inmutables y Mutables</font>

+ Cuando la zona de memoria que ocupa un objeto se puede modificar se dice que el tipo de datos es <span style="color:yellow">mutable</span>. (<span style="color:limegreen">list, dict, set</span>)
+ Cuando la zona de memoria que ocupa un objeto no se puede modificar se dice que el tipo de datos es <span style="color:yellow">inmutable</span>. (<span style="color:limegreen">int, bool, float, str, tuple, frozenset</span>)
