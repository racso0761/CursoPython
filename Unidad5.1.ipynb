### <font color=#38bdf8>Creación de Matrices</font>

Crear una matriz consiste, pues, en crear una lista de listas. Si deseamos crear una matriz nula (todos sus elementos iguales a 0) de tamaño 2x2 lo haremos como ya conocemos:

```python
# Ejemplo: Matrices 1
m = [[0, 0], [0, 0]]
print(f'Matriz: {m}')
```

Pero ahora necesitamos crear una matriz de 3 x 6 elementos, o de 100 x 100 elementos. Para esto podríamos utilizar el operador <span style="color:limegreen">*</span> de repetición, es decir:

```python
# Ejemplo: Matrices 2
vec = [0]* 6
print(vec)
```

```python
# Ejemplo: Matrices 3
M = [vec] * 3
print(M)
```

Ya disponemos de la matriz de 3 x 6 elementos. Ahora asignamos valores, por ejemplo:

```python
# Ejemplo: Matrices 4
M[0][0] = 1
print(M)
# Funciona?
```

+ Haciendo que cada fila sea una lista distinta.


```python
# Ejemplo: Matrices 5
M = []
for i in range(3):
    vec = [0] * 6
    M.append(vec)
```

```python
print(M)
```

```python
M[0][0] = 1
print(M)
```
