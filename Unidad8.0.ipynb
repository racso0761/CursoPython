La Programación Orientada a Objetos (POO) es un nuevo modo de pensar (paradigma) en cómo resolver algunos aspectos de los sistemas de software. Básicamente, *se examina el problema a resolver como un conjunto de entidades independientes que se relacionan entre sí y con otras partes del problema para proporcionar una solución*. Estas entidades son denominadas <span style="color:magenta">objetos</span> en los programas, y así, en un lenguaje orientado a objetos, tal como el <span style="color:magenta">C++</span> o <span style="color:magenta">Python</span>, el centro del lenguaje no son las funciones ni los datos, sino más bien las entidades (objetos) que trabajan en forma conjunta para lograr la solución. Construir un programa orientado a objetos implica determinar aquellos objetos que son necesarios para resolver el problema (estos constituyen el dominio del problema), y luego, estos objetos serán capaces de interactuar entre sí y ser utilizados para construir los cálculos, operaciones y procesos que definen el comportamiento del sistema de software.

## Programación Orientada a Objetos

Superficialmente el término <span style="color:magenta">Orientado a Objetos</span> significa organizar un sistema de software como un conjunto de objetos del mundo real, que reúnen datos (variables o atributos) y funciones (comportamientos) en un única unidad lógica y operativa. Esto es completamente opuesto a lo que proclama la programación convencional, en la cual, los datos y el comportamiento están débilmente conectados.

### Los Objetos

La pregunta que se debe hacer es: Que es un objeto? La respuesta es simple: Un objeto es una representación simple o compleja, real o imaginaria de alguna cosa. Desde el punto de vista de un lenguaje orientado a objetos un <span style="color:magenta">objeto</span> es una forma de agrupar un <span style="color:magenta">conjunto de datos o atributos (estado)</span> y <span style="color:magenta">comportamientos (funcionalidad)</span> en un mismo bloque de código que se puede referenciar desde otras partes de un programa.

En un sistema, los datos están contenidos en entidades discretas perfectamente distinguibles llamadas objetos. Un objeto es la representación sistémica de la abstracción de un conjunto de cosas del mundo real, y que tiene sentido en el problema analizado. Por ejemplo, en un <span style="color:magenta">sistema de Punto de Venta</span> (como una caja de un supermercado) los <span style="color:magenta">objetos</span> participantes del problema podrían ser instancias específicas de <span style="color:magenta">Facturas, Productos, Tarjetas de Crédito, Efectivo, Almacén de Stock</span>, etc.

Todo objeto contiene una identidad, un estado interno (estructura de datos) y métodos (operaciones que definen el comportamiento), como se observa en la figura.

![](./Figuras/poo_f1.png)

+ <span style="color:magenta">Una identidad</span> que lo identifica del resto de los objetos del mismo tipo.

+ <span style="color:magenta">Una estructura de datos</span> que define su estado interno, denominados <span style="color:magenta">atributos</span>.

+ <span style="color:magenta">Un conjunto de operaciones</span> que definen su comportamiento, materializadas por funciones denominadas <span style="color:magenta">métodos</span>. Se suele expresar que estos comportamientos se manifiestan por medio de un <span style="color:magenta">conjunto de mensajes que el objeto es capaz de responder</span>.

De lo anterior, podemos concluir lo siguiente:

+ <span style="color:magenta">El objeto es una entidad que tiene un cierto comportamiento</span>: Cuando se piensa en un objeto, se tiene que preguntar: ¿Para que sirve? ¿Qué comportamientos tiene? ¿Qué se le puede pedir? ¿Que servicios proporciona? ¿Qué sabe hacer el objeto?. Esta visión de entender al objeto con cierta capacidad de hacer cosas es bastante compleja de asimilar pues se tiende a pensar que los procesos son activos, y que las cosas son pasivas, mientras que en el paradigma de objetos, los que saben hacer algo son los objetos. La idea de entender a los objetos como entidades con comportamientos es la idea central del paradigma de objetos. (Ejemplos de objetos pueden ser: una cuenta bancaria, los clientes del banco, un archivo de datos, una factura, un libro).

+ <span style="color:magenta">El comportamiento está expresado por el conjunto de mensajes que el objeto es capaz de responder</span>: Cuando un objeto recibe un mensaje (se le pide que haga algo), ejecuta una sección de código contenida en una función que se denomina genéricamente método. Lo importante es que este es un código muy procedural (se parece mucho a lo que es un método o función de la programación estructurada) que se activa cada vez que uno de los objetos recibe un mensaje que tiene su interfase (la interfase de un objeto esta definida por el conjunto de mensajes que puede recibir y ejecutar con sus métodos).

+ <span style="color:magenta">El estado interno</span>: los objetos tienen un estado interno que en términos de implementación está formado por variables (llamadas atributos) y que capturan lo que ese objeto conoce (el conocimiento del objeto), guardan las características individuales que diferencian a un objeto de otro y determinan su apariencia o estado.

+ <span style="color:magenta">Cada objeto tiene identidad propia</span>. En efecto, dos objetos son distintos aún si todos los valores de sus atributos son idénticos. Por ejemplo: dos alumnos pueden tener el mismo nombre y apellido pero son dos personas distintas. Dos autos pueden ser del mismo modelo y color, tener características idénticas respecto al tipo de motor pero aún así son autos distintos. En la vida real un objeto simplemente existe, pero en un lenguaje de programación, tiene un único identificador por medio del cual puede ser únicamente referenciado o accedido. El identificador puede ser implementado de varias formas: una dirección, un índice de arreglo o bien por medio del valor de un atributo.

Los objetos con la misma estructura de datos (atributos) y comportamientos (operaciones o métodos) disponen de un <span style="color:yellow">molde (plantilla)</span> que permite crearlos según ese patrón, y este molde se denomina <span style="color:yellow">clase</span>. Suele decirse que las clases agrupan los objetos del mismo tipo, pero esto no es cierto: "Las clases solo son plantillas para construir objetos que tienen una estructura de datos y comportamientos predeterminados". 

Cada clase describe, posiblemente, un conjunto infinito de objetos individuales que comparten la "forma" que la plantilla (clase) le provee. Cada objeto construído usando esa plantilla se dice que es una <span style="color:magenta">instancia de su clase</span>, y cada instancia de una clase tiene un valor propio para cada atributo, pero comparten los nombres de los mismos y los nombres de las operaciones con otras instancias de la misma clase.

Por ejemplo: 
En una empresa, el administrador y los obreros trabajan, ambos tienen datos personales (estructura de datos común) y también trabajan para la misma empresa, además cobran un sueldo, hacen aportes (comportamiento común). Cada uno de ellos son objetos del dominio del problema y como comparten la estructura de datos y comportamientos, se los puede modelar mediante una plantilla (clase) llamada ```Empleado```. Así, el administrador y cada uno de los obreros son instancias de la clase Empleado (han sido creados usando la plantilla Empleado).

## Pilares Fundamentales de la Programación Orientada a Objetos

Los pilares sobre los que se basa el paradigma de la Programación Orientada a Objetos (POO) son:  <span style="color:magenta">la abstracción y encapsulamiento</span>, <span style="color:magenta">la herencia</span> y el <span style="color:magenta">polimorfismo</span>.

### Abstracción y Encapsulamiento

La <span style="color:magenta">abstracción</span> es la especificación de ciertos aspectos esenciales de un sistema para un propósito particular, y su función es aislar los aspectos que son importantes dejando de lado aquellos que no lo son. La abstracción debe servir para algún propósito, porque ese propósito determina lo que es y lo que no es importante.

Un objeto tiene un estado interno y un comportamiento. Ese estado interno está <span style="color:magenta">encapsulado</span>, y esto significa que no hay forma de acceder a los atributos de un objeto desde el exterior del mismo, lo único que se puede hacer es enviarle mensajes al objeto para que en su respuesta entregue los valores de sus atributos. Cualquier cambio que ocurra en el estado interno de un objeto es producto de la ejecución de uno de sus métodos. 

La abstracción y el encapsulamiento son conceptos complementarios: 

+ La abstracción pone énfasis en la vista externa del objeto y 

+ El encapsulamiento (ocultamiento de la información) previene a los usuarios de acceder al estado interno del objeto directamente, donde se implementa el comportamiento de la abstracción.

#### Abstracción

La <span style="color:magenta">abstracción</span> es la propiedad que considera los aspectos más significativos o notables de un problema y expresa una solución en esos términos. La abstracción posee diversos grados o niveles de abstracción, los cuales ayudan a estructurar la complejidad intrínseca que poseen los sistemas del mundo real.

En el análisis hay que concentrarse en <span style="color:magenta">¿Qué hace un objeto?</span> y no en <span style="color:magenta">¿Cómo lo hace?</span>.

El principio de abstracción es más fácil de entender con una analogía del mundo real, por ejemplo, la televisión. Todos estamos familiarizados con sus características y su manejo manual o con el control remoto (encender, apagar, subir volumen, etc.) incluso sabemos cómo conectarlo con otros aparatos externos como altavoces o reproductores, o al internet. Sin embargo, no todos sabemos cómo funciona internamente, es decir, en como recibe la señal, en como la traduce y la visualiza en la pantalla, etc.

Normalmente sucederá que no sepamos cómo funciona el aparato de televisión pero si sabemos cómo utilizarlo. Esta característica se debe a que la televisión separa claramente su <span style="color:magenta">implementación interna</span> de su <span style="color:magenta">interfaz externa</span>. Interactuamos con la televisión a través de su <span style="color:magenta">interfaz</span>: los botones de encendido, cambio de canal, volumen, etc. no conocemos el tipo de tecnología que utiliza, el método de generar la imagen en la pantalla o como funciona internamente, es decir, su implementación, ya que ello no afecta a su interfaz.

#### Encapsulamiento

El <span style="color:magenta">encapsulamiento</span> hace referencia a reunir en una cierta estructura a todos los elementos que a un cierto nivel de abstracción se pueden considerar pertenecientes a una misma entidad, y es el proceso de agrupamiento de datos y operaciones relacionadas bajo una misma unidad de programación, lo que permite aumentar la cohesión de los componentes del sistema.

El encapsulamiento oculta lo que hace un objeto de lo que hacen otros objetos y también del mundo exterior por lo que se denomina también ocultación de datos. Un objeto tiene que presentar “una cara” (interfaz) al mundo exterior de modo que se puedan iniciar sus operaciones.

Por ejemplo, la televisión tiene un conjunto de botones para encender, apagar y/o cambiar canal. Una lavadora tiene un conjunto de mandos e indicadores que establecen la temperatura, el nivel de agua, etc. Los botones de la televisión y la lavadora constituyen la comunicación con el mundo interior, es decir son las interfaces.

La interfaz de una clase representa un <span style="color:yellow">“contrato”</span> de prestación de servicios entre ella y los demás componentes del sistema. De este modo, los clientes solo necesitan conocer los servicios que este ofrece y no como están implementados internamente. Por consiguiente, se puede modificar la implementación de una clase sin afectar a las restantes relacionadas con ella.

**Reglas de Visibilidad**

Las reglas de visibilidad complementan o refinan el concepto de encapsulamiento. Los diferentes niveles de visibilidad dependen del lenguaje de programación con el que se trabaje, pero en general siguen el modelo de <span style="color:magenta">C++</span>, estos niveles de visibilidad son:

+ El nivel más fuerte se denomina nivel <span style="color:magenta">“privado”</span>; la sección privada de una clase es totalmente invisible para otras clases, solo miembros de la misma clase pueden acceder a atributos localizados en la sección privada.

+ Es posible aliviar el nivel de ocultamiento situando algunos atributos en la sección <span style="color:magenta">“protegida”</span> de la clase. Estos atributos son visibles tanto para la misma clase como para las clases derivadas. Para las restantes clases permanecen invisibles.

+ El nivel más débil se obtiene situando los atributos en la sección <span style="color:magenta">“pública”</span> de la clase con lo cual se hacen visibles a todas las clases.

Los atributos privados están contenidos en el interior de la clase ocultos a cualquier otra clase. Ya que los atributos están encapsulados dentro de una clase, se necesitará definir cuáles son las clases que tienen acceso a visualizar y cambiar los atributos. Esta característica se conoce como visibilidad de los atributos. En general se recomienda visibilidad privada o protegida para los mismos.

**Modificadores de Acceso**

Los modificadores de acceso se utilizan para definir la visibilidad de los miembros de una clase (atributos y métodos) y de la propia clase. En los lenguajes de programación orientados a objetos existen 3 modificadores de acceso:

+ <span style="color:magenta">Visibilidad pública</span>
+ <span style="color:magenta">Visibilidad protegida </span>
+ <span style="color:magenta">Visibilidad privada</span>

Un principio fundamental en la programación orientada a objetos es la ocultación de la información, que significa que determinados datos del interior de una clase no pueden ser accedidos por funciones externas a la clase, esto sugiere que solamente la información sobre lo que puede hacer una clase debe estar visible desde el exterior, pero no cómo lo hace. 

Esto tiene una gran ventaja: si ninguna otra clase conoce cómo está almacenada la información entonces se puede cambiar fácilmente la forma de almacenarla sin afectar otras clases.

### Herencia

En la programación orientada a objetos, la <span style="color:magenta">herencia</span> está en todos lados, de hecho, se podría decir que es casi imposible escribir el más pequeño de los programas sin utilizar herencia. Todas las clases que se crean dentro de la mayoría de los lenguajes de programación orientados a objetos heredan implícitamente de la clase <span style="color:limegreen">Object</span> y, por ende, se pueden comportar como objetos (que es la base del paradigma).

En el mundo real existen entidades similares con propiedades comunes; y con el fin de reducir la redundancia al implementar estas diferentes entidades similares entre sí, el enfoque Orientado a Objetos introdujo un nuevo concepto: la herencia, que es compartir atributos y métodos entre clases, pero respetando una dependencia jerárquica. Esto permite que una clase, llamada <span style="color:magenta">superclase (clase base o clase padre)</span>, sea definida en forma general y luego ser refinada sucesivamente en otras clases – que dependen jerárquicamente de ellas - llamadas <span style="color:magenta">subclases (clases derivadas o clases hijas)</span>. Las subclases heredan la mayoría de las propiedades de sus superclases y a la vez agregan sus propiedades específicas. La herencia es una de las características principales de la POO.

La relación <span style="color:yellow">is-a (es un)</span> se basa en la herencia y permite afirmar que un objeto es de un tipo (clase) en específico.

Un ejemplo puede ser el de las figuras geométricas: ```triángulo```, ```rectángulo``` o ```círculo```, tal vez usado en un sistema de diseño asistido por computadora o juegos de simulación. Todas estas figuras tienen un ```tamaño```, una ```posición``` y un ```color```. Además, las figuras se ```dibujan```, se ```desplazan en el espacio```, y se ```rotan```, tienen un comportamiento similar. Podemos pensar que el tipo base es ```figura```, y cada figura tiene un ```tamaño```, un ```color```, una ```posición``` y así sucesivamente. Cada ```figura``` se puede ```dibujar```, ```borrar```, ```mover```, ```colorear```, etc. A partir de ahí, los tipos específicos de figuras derivan (heredan) de ella: ```círculo```, ```rectangulo```, ```triángulo```, y así sucesivamente, cada uno de ellos pueden tener características y comportamientos adicionales. Ciertas figuras pueden ser, por ejemplo, rotadas. Algunos comportamientos pueden ser diferentes, como cuando se quiere calcular el área de una figura. La jerarquía de tipos expresa las similitudes y las diferencias entre las figuras.

Cuando se hereda de una clase existente, se está creando una clase nueva. Esta nueva clase contiene no sólo todos los miembros de la clase base (aunque los datos privados están ocultos e inaccesibles), sino que además, y lo que es más importante, hereda la interfaz de la clase base. Es decir, todos los mensajes que se pueden enviar a los objetos de la clase base se pueden enviar también a los objetos de la clase derivada. Dado que se conoce el tipo de una clase por los mensajes que se le pueden enviar, eso significa que la clase derivada es del mismo tipo que la clase base. En el ejemplo anterior, "un círculo <span style="color:magenta">es una</span> figura". Esta equivalencia de tipos vía herencia es uno de las claves fundamentales para comprender la POO.

Se debe tener en cuenta que una clase se puede derivar de mas de una clase base, dando lugar a la <span style="color:magenta">herencia múltiple</span>. Esta es una característica que no todos los lenguajes de programación orientados a objetos disponen. 

### Polimorfismo

Las subclases también heredan de sus superclases la mayoría de sus operaciones, sin embargo, estas operaciones pueden ser redefinidas para que se ejecuten de manera diferente de acuerdo a la naturaleza de cada subclase. Es decir que una misma operación se puede comportar de forma diferente en clases distintas. Esta característica fundamental de la orientación a objetos, se denomina <span style="color:magenta">polimorfismo</span> (muchas formas), y permite a una misma interfaz ser usada para una clase general de acciones.

Por ejemplo, la función ```dibujar()``` se debe comportar de diferente manera para graficar un círculo, un rectángulo o un triángulo. Entonces, cada figura sobrescribirá esa función heredada de la superclase para ajustarla a sus propias necesidades, de manera tal que se podrá invocar el método ```dibujar()``` de cualquier objeto ```Figura``` y siempre se dibujará la "figura correcta" (ver Figura).

![](./Figuras/poo_f3.png)

El Polimorfismo, junto con la Herencia, son las características más importantes de la POO y permiten que objetos que tienen estructuras internas diferentes compartan la misma interfaz externa. Proveen una interfaz de software genérico de manera que un conjunto de tipos diferentes de objetos puedan ser manejados uniformemente. El lenguaje <span style="color:magenta">C++</span> por ejemplo, provee tres tipos de Polimorfismo: funciones virtuales, sobrecarga de nombre de funciones y sobrecarga de operadores, si bien el verdadero polimorfismo se manifiesta solo en el uso de la funciones virtuales.

## Otras Características de la POO

### Composición y Agregación

Una clase puede tener referencias a objetos de otras clases como miembros. A dicha capacidad se le conoce como composición ó agregación y algunas veces como relación <span style="color:yellow">has-a (tiene un)</span>.

El concepto de composición ó agregación no fue creado para la programación; suele usarse a menudo para objetos complejos en el mundo real. Toda criatura viviente y la mayor parte de los productos manufacturados están constituidos por partes. A menudo, cada parte es un subsistema que está integrado por su propio conjunto de subpartes. Junto, todo el sistema forma una jerarquía de composición ó agregación.

Por ejemplo, el cuerpo humano está compuesto por varios órganos: cerebro, corazón, estómago, huesos, músculos, etc. A su vez, cada uno de estos órganos está compuesto por muchas células, y cada una de estas células está compuesta por muchos orgánulos, como el núcleo y las mitocondrias. Cada orgánulo está compuesto por muchas moléculas. Y finalmente, cada molécula orgánica está compuesta por muchos átomos.

En una jerarquía de composición (así como en una jerarquía de agregación), la relación entre una clase contenedora y una de sus clases parte se denomina relación has-a (tiene un). Por ejemplo, cada cuerpo humano tiene un cerebro y tiene un corazón.

La composición y la agregación es una forma de reutilización de software, en donde una clase tiene como miembros referencias a objetos de otras clases.

La diferencia entre una composición y una agregación depende del contexto en el que se use la relación. Pero, en la literatura normalmente se hace referencia a que la diferencia entre agregación y composición es que en el caso de la agregación la parte tiene sentido sin el todo y en el caso de la composición no lo tiene.

Esta definición es un poco ambigua y se debe decir que el concepto que hace de parte debe considerarse dentro del contexto que se está tratando. Dentro de ese contexto se debe determinar si el objeto tiene sentido de forma independiente.
Por ejemplo, el objeto palabra en el contexto de un diccionario es una agregación pero en el contexto de una frase es una composición.

## Ventajas del Desarrollo Orientado a Objetos

El desarrollo orientado a objetos (DOO) no solo permite que la información sea compartida en una aplicación, sino que ofrece la posibilidad de reutilizar el diseño y código para proyectos futuros. El DOO provee las herramientas para facilitar la reusabilidad tales como la abstracción, el encapsulamiento y la herencia.

La orientación a objetos no es una fórmula mágica para asegurar la reusabilidad. La reusabilidad no sucede sino que debe ser planificada más allá de la aplicación inmediata e invirtiendo un esfuerzo extra en un diseño más general.

En el DOO, el principal beneficio no es reducir tiempo de desarrollo. El DOO puede tomar más tiempo que el desarrollo convencional, pues se intenta promover una futura reusabilidad y reducir errores y tiempo de mantenimiento. El tiempo hasta que el código se complete, probablemente es el mismo o quizás mayor, sin embargo, las iteraciones subsecuentes de un DOO son más rápidas y fáciles que con el desarrollo estructurado convencional porque las revisiones son más localizadas. 

Los sistemas orientados a objetos son más resistentes a los cambios y pueden desarrollarse y mantenerse mejor porque su diseño es estable.

La POO no es un lenguaje, sino una filosofía de análisis y diseño, y puede llevarse a cabo en varios lenguajes diferentes. No hay que confundir las limitaciones de cualquier lenguaje con la POO: esta es fundamentalmente una nueva forma de pensar y no una técnica de programación.

Construir un programa usando POO involucra determinar los objetos que son necesarios para resolver el problema. Luego, los objetos son usados para construir las operaciones que definen el comportamiento del sistema de software. El paso de mensajes es el mecanismo fundamental entre objetos; un mensaje (desde otro objeto) se envía a otro objeto para que ejecute una de sus operaciones. Los mensajes que se pueden enviar a un objeto están definidos en la interfaz de ese objeto. Solamente el objeto puede modificar su estado interno. Es decir, la construcción fundamental son los objetos, todos los cálculos se hacen dentro de los objetos, y no hay ningún tipo de código fuera de los mismos.

Muchos de los conceptos nombrados como <span style="color:magenta">mensaje</span>, <span style="color:magenta">método</span>, <span style="color:magenta">polimorfismo</span>, <span style="color:magenta">herencia</span> eran conocidos anteriormente, pero en el enfoque orientado a objetos son diferentes, un mensaje no es lo mismo que una llamada a una función, y herencia no es lo mismo que copiar y pegar código.

## Uso de la POO y Python

Luego de la breve revisión anterior de las generalidades referidas a la Programación Orientados a Objetos ( POO) es importante destacar que los contenidos referidos a estos temas y dictados en este curso están fuertemente limitados por motivos de tiempo y de formación previa de los estudiantes. Esto es así por cuanto la POO son nuevas formas de pensar un sistema de software, y su asimilación exige un tiempo y una dedicación que no está disponible en esta etapa. 

Sin embargo, en cursos futuros o en la vida profesional surgirá la necesidad de utilizar bibliotecas de componentes Orientados a Objetos en diferentes Lenguajes como C++ o Python, para construir sistemas de software con necesidades específicas, e integrarlos a los conocimientos y construcciones ya aprendidas.

Por este motivo, se dará la formación básica necesaria para aplicar, enlazar y construir sistemas de software que utilicen entidades Orientadas a Objetos que proporcionan las habilidades requeridas por la aplicación bajo desarrollo. 

En este contexto, nos limitaremos a conocer los mecanismos de instanciación de objetos de clases ya existentes y nos centraremos en la integración de estos componentes por medio del mecanismo de “intercambio de mensajes” (invocación de métodos), de manera muy similar al uso de bibliotecas de funciones que ya conocemos. Pero debe quedar claro que no se aprenderá a analizar, ni a diseñar ni a construir objetos de software orientado a objetos, dejando pendiente este proceso para etapas posteriores en la evolución de cada uno de ustedes.

## Presentación y Uso de los elementos de una Clase

Una clase es un nuevo tipo de datos que usted o alguna otra persona crea para resolver un problema particular. Una vez que se ha creado una clase, cualquiera puede utilizarla sin conocer los detalles de su funcionamiento, o incluso de la forma en que se han construido. Se va hacer uso de las clases como si sólo fueran otro tipo de datos predefinido disponible para su uso en programas. Las clases creadas por terceras personas se suelen empaquetar en bibliotecas.

Los objetos idénticos, exceptuando su estado durante la ejecución del programa, se agrupan en clases de objetos y de ahí viene la palabra reservada <span style="color:limegreen">class</span>. Crear tipos de datos abstractos (clases) es un concepto fundamental en la programación orientada a objetos. Los tipos de datos abstractos (clases) trabajan casi exactamente como tipos predefinidos: puede crear variables de un tipo (llamadas objetos o instancias) y manipular estas variables (llamado envío de mensajes o peticiones; envía un mensaje y el objeto decide qué hacer con él).

Por eso, lo que hace realmente un programa orientado a objetos es crear nuevos tipos de datos, prácticamente todos los lenguajes de programación orientados a objetos usan la palabra reservada <span style="color:limegreen">class</span>. Cuando vea la palabra <span style="color:limegreen">class</span>, piense en <span style="color:limegreen">type</span> y viceversa. Pero, ¿cómo se consigue que un objeto haga algo, como utilizar objetos? Debe haber una forma de hacer una petición al objeto para que haga algo, como completar una operación, dibujar algo en la pantalla, etc. Y cada objeto puede satisfacer sólo ciertas peticiones. Las peticiones a las que puede responder un objeto están definidas por su intefaz, y es el tipo lo que determina la interfaz.

La interfaz establece qué peticiones se pueden hacer a un objeto particular. Sin embargo, se debe codificar en algún sitio para satisfacer esta petición. Ésta, junto con los datos ocultos, constituyen la implementación. Desde el punto de vista de la programación procedural, no es complicado. Una clase tiene una función asociada para cada posible petición, y cuando se hace una petición particular a un objeto, se llama a esa función. Este proceso normalmente se resume diciendo que se ha <span style="color:magenta">enviado un mensaje</span> (se ha hecho una petición) a un objeto, y el objeto sabe qué hacer con este mensaje (ejecuta código). Cada clase se representa con una caja, con el nombre de la clase en la parte superior, los atributos que necesite describir en la parte central de la caja, y los métodos (las funciones o comportamientos que pertenecen a este objeto, que reciben cualquier mensaje que se envíe al objeto) en la parte inferior de la caja (ver Figura).

## Pensar en Objetos

+ Pensar en objetos, puede resultar una tarea difícil al principio. 

+ En Python todo es un “objeto” y debe ser entendido como tal. 

### ¿Qué es un objeto? ¿De qué se habla cuando se hace referencia a "orientación a objetos?"

+ Un objeto es *una cosa*. Y, si una cosa es un sustantivo, entonces un objeto es un **sustantivo**.

+ Si se mira alrededor se encontrarán decenas, cientos de objetos.

    + La computadora, es un objeto.

    + Usted es un objeto.

    + Una llave es un objeto.

    + El mate, es otro objeto.
    
    + Una mascota también es un objeto.

Entonces hay que pensar toda la vida en objetos.

### Cualidades de un Objeto

+ Describir un objeto, es simplemente mencionar sus cualidades, sus *atributos*. 

+ Las cualidades son **adjetivos**.

+ Para describir “la manera de ser” de un objeto, debemos preguntarnos ¿cómo es el objeto?

    + Toda respuesta que comience por *...el objeto es...*, seguida de un adjetivo, será una cualidad del objeto.

    + Por ejemplo:

        + El objeto es de color amarillo

        + El objeto es de tamaño grande
        
        + El objeto es de aspecto elegante

### Objetos formado por otros Objetos

+ Además de cualidades, los objetos *poseen otras cosas*.

+ Son aquellas *pseudo-cualidades* que en vez de responder a la pregunta ¿cómo es el objeto? responden a la pregunta “¿cómo está compuesto el objeto?” o “¿Qué tiene el objeto?”.

+ La respuesta a esta pregunta, estará dada por la frase “...el objeto tiene ...”, seguida de un **adverbio** de cantidad (uno, varios, muchos, algunos, unas cuantas) y un **sustantivo**.

    + El objeto tiene algunos *asientos*

    + El objeto tiene un *motor*

    + El objeto tiene unas cuantas *ventanas*

+ En el ejemplo, *asientos*, *motor*, *ventanas* son sustantivos que representan objetos contenidos dentro de un objeto, denominados *atributo-objeto* y responden a la pregunta ¿Como es/son ese/esas/esos objeto/s? seguido del atributo-objeto.

+ Por ejemplo:

    + El objeto tiene algunos asientos. ¿Como son esos asientos?

        + Los asientos son de color negro

        + Los asientos son de gran tamaño

    + El objeto tiene un motor. ¿Como es ese motor?

        + El motor es de gran potencia

        + El motor es de muchos cilindros

        + El motor es de combustión

    + El objeto tiene unas cuantas ventanas. ¿Como son esas ventanas?

        + Las ventanas son de color transparente

        + Las ventanas son medianas

+ Entonces se pueden resumir los *atributos* de la siguiente manera:

    + Los que responden a la pregunta ¿Como es el objeto? con la frase *"El objeto es..." + adjetivo* (atributos definidos por cualidades).

    + Los que responden a la pregunta ¿Que tiene el objeto? con la frase *"El objeto tiene..." + sustantivo* (atributos definidos por las cualidades de otro objeto).

### Objetos que comparten Características con Objetos

+ Suponga que un objeto, es prácticamente igual a un nuevo objeto. Es decir, que el nuevo objeto tiene absolutamente todas las características del primer objeto, es decir, tiene los mismos atributos.

+ Pero también, tiene algunos más. Por ejemplo, este nuevo objeto, tiene algunas *ruedas*. Es decir, que las características de nuestro nuevo objeto, serán todas las del objeto original, más una nueva: *ruedas*.

+ Ejemplo:

    + El nuevo objeto es de color amarillo.

    + El nuevo objeto es de tamaño grande.

    + El nuevo objeto es de aspecto elegante.

        + El motor es de gran potencia

        + El motor es de muchos cilindros

        + El motor es de combustión

    + El nuevo objeto tiene unas cuantas ventanas. ¿Como son esas ventanas?

        + Las ventanas son de color transparente

        + Las ventanas son medianas
    
    (nuevas características)
    + El nuevo objeto tiene algunas ruedas. ¿Cómo son esas ruedas?

        + Las ruedas son anchas

        + Las ruedas son con tornillos. ¿Cómo son esos tornillos?

            + Los tornillos son largos

Se observa que el nuevo objeto es una especie de “objeto original ampliado”. 

Es decir que el nuevo objeto, es exactamente igual al objeto original (comparte todos sus atributos) pero posee nuevas características.

Además, el objeto original y el nuevo objeto, son dos objetos diferentes. No obstante, el nuevo objeto es un sub-tipo del objeto original (se deriva de ...).

### Los Objetos pueden Hacer Cosas

+ Los objetos “pueden hacer cosas”, es decir los objetos tienen, *"capacidades"* o *comportamientos*.

+ Los objetos tiene la capacidad de realizar acciones.

    + Las acciones, son *verbos*. Es decir responden a la pregunta “¿Qué puede hacer el objeto?” y la respuesta a esta pregunta, comienza por la frase *“...el objeto puede...” + verbo* en infinitivo.

+ Por ejemplo:

    + El objeto original puede *arrancar*

    + El nuevo objeto se puede *mover*


> En el gráfico, se nota que el nuevo objeto, no solo tiene los mismos atributos que el objeto original, sino que además puede realizar las mismas acciones que éste.

#### Inicialización y Destrucción de Objetos

**Inicialización (<span style="color:limegreen">\_\_init\_\_</span>)**

+ Existe un método especial que se ejecuta en el momento de la creación del objeto denominado <span style="color:limegreen">\_\_init\_\_</span> (también llamado <span style="color:magenta">constructor</span> de la clase).

+ Este método puede recibir argumentos que le permiten a la clase darle valores iniciales a los atributos del objeto al momento de la instanciación.

+ Siempre se le debe pasar el argumento <span style="color:limegreen">self</span>.

**Destrucción (<span style="color:limegreen">\_\_del\_\_</span>)**

+ Existe un método especial que tiene por función la eliminación de un objeto denominado <span style="color:limegreen">\_\_del\_\_</span> (también llamado <span style="color:magenta">destructor</span> del objeto).

````python
# Ejemplo: POO 6
# Constructor y Destructor

class Saludo():
    def __init__(self):
        print('Me han instanciado')
    
    def __del__(self):
        print('He sido destruido')
````

#### Encapsulamiento

El <span style="color:magenta">encapsulamiento</span> en POO es un concepto que hace referencia al ocultamiento de los estados internos de una clase hacia el exterior. 

Dicho de otra manera:

+ Encapsular consiste en hacer que los atributos o métodos internos de una clase no se puedan acceder ni modificar desde fuera directamente, sino que se puedan hacer <span style="color:magenta">solo por los servicios propuestos por el objeto (su interfaz)</span>.

El encapsulamiento permite definir niveles de visibilidad para los elementos de la clase. Estos niveles de visibilidad definen los derechos de acceso a los datos en función de si se accede a ellos mediante un método de la propia clase, de una clase heredada o de cualquier otra clase. 

Hay cuatro niveles de visibilidad:

+ <span style="color:magenta">Visibilidad por defecto</span>: no se especifica ningún modificador de visibilidad.

+ <span style="color:magenta">Visibilidad pública</span>: los métodos de todas las clases pueden acceder a los atributos o métodos de una clase definida con el nivel de visibilidad «público». Este es el nivel más bajo de protección de datos.

+ <span style="color:magenta">Visibilidad protegida</span>: el acceso a los atributos está restringido a los métodos de las clases heredadas, es decir, de los métodos miembro de la clase y las clases derivadas. Así, un atributo o un método declarado «protegido» es accesible sólo por las clases de un paquete y a sus subclases, aunque estén definidas en un paquete diferente.

+ <span style="color:magenta">Visibilidad privada</span>: el acceso a los atributos se limita a los métodos de la propia clase. Este es el nivel más alto de protección de datos.

Python por defecto <span style="color:red">no oculta</span> los atributos y métodos de una clase.

#### Atributos de Instancia vs Atributos de Clase

La diferencia fundamental entre este tipo de atributos es que los <span style="color:magenta">atributos de clase</span> son compartidos por todas las instancias de esa clase, mientras que los <span style="color:magenta">atributos de instancia</span> son particulares para cada objeto creado con esa clase. Por tanto, las variables de instancia son para datos únicos y propios para cada objeto y las variables de clase son para atributos que deban ser compartidos por todas las instancias de esa clase.

+ Los <span style="color:magenta">atributos de instancia</span> son creados usando la sintaxis <span style="color:limegreen">self.attr</span>. Como se ha dicho son locales para esa instancia y por tanto solo accesibles desde una de ellas. Es decir debe existir un objeto instanciado para acceder al atributo.

    + En cuanto a su utilidad, genéricamente podemos decir que permiten definir un determinado "estado" de ese objeto particular y que lo diferencia de otros objetos de su clase. Como puede ser el atributo "matrícula" en una clase "Automóvil".

    + Son almacenados en un diccionario (diccionario de instancia u objeto) al cual se puede acceder mediante <span style="color:limegreen">NombreInstancia.\_\_dict\_\_</span>.

+ Los <span style="color:magenta">atributos de clase</span> se definen fuera de cualquier método de la clase, normalmente justo debajo del string de documentación.

    + Se pueden referenciar directamente desde la clase, es decir no necesitan de un objeto instanciado.

    + Se pueden referenciar desde cualquier instancia de la clase.
    
    + Se usan comunmente para establecer valores predeterminados para variables de instancia, definir constantes, etc.

    + Son almacenados en un diccionario distinto (diccionario de clase) e independiente al usado para los atributos de instancia, el cual se puede consultar mediante <span style="color:limegreen">NombreClase.\_\_dict\_\_</span> o desde una instancia con <span style="color:limegreen">NombreInstancia.\_\_class\_\_.\_\_dict\_\_</span>.

    + Al intentar asignar un nuevo objeto al atributo de clase usando *instancia.atributo* o desde un método de instancia vía <span style="color:limegreen">self.atributo</span> no se modifica el atributo de clase, se crea un atributo de instancia con el mismo nombre y se le asigna el nuevo valor.

#### Métodos de Instancia vs Métodos de Clase

+ Un <span style="color:magenta">método de instancia</span> es un método que modifica o accede al estado del objeto al que hace referencia. Recibe <span style="color:limegreen">self</span> como primer argumento, el cual se convierte en el propio objeto sobre el que se está trabajando.

+ Un <span style="color:magenta">método de clase</span> es un método que modifica o accede al estado de la clase a la que hace referencia. Recibe <span style="color:limegreen">cls</span> como primer argumento, el cual se convierte en la propia clase sobre la que se está trabajando. La identificación de estos métodos se completa aplicando el decorador <span style="color:limegreen">@classmethod</span>.

````python
# Ejemplo: POO 12
class Vehiculo():
    contador = 0
    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')

    @classmethod
    def cantidad_vehiculos(cls):
        print(f'{cls.contador} vehiculos creados hasta este momento!')
````

#### Métodos Estáticos

+ Un <span style="color:magenta">método estático</span> es un método que <span style="color:red">no debería</span> modificar el estado del objeto ni de la clase. No recibe ningún argumento especial (ni <span style="color:limegreen">self</span>, ni <span style="color:limegreen">cls</span>). 

+ La identificación de estos métodos se completa aplicando el decorador <span style="color:limegreen">@staticmethod</span> al método.

Los métodos estáticos son útiles cuando se quiere definir una funcionalidad que pertenece a una clase pero que no depende de ningún atributo de instancia o de clase.

A continuación, se presentan algunas situaciones en las que se pueden utilizar los métodos estáticos:

+ Cuando se quiere implementar una funcionalidad que no depende de ningún atributo de la clase ni de una instancia en particular de la clase. Por ejemplo, un método que realiza una operación matemática simple, como calcular el valor absoluto de un número.

+ Cuando se quiere proporcionar una función auxiliar que se utiliza en varias partes del programa y que no está específicamente relacionada con ninguna instancia de la clase. En este caso, se puede definir el método estático en la clase para tener una mejor organización del código.

+ Cuando se quiere crear un método de la clase que no requiere acceso a ningún atributo de instancia ni de clase, pero que se relaciona con la lógica de la clase. Por ejemplo, se puede utilizar un método estático para crear una función de ayuda que comprueba la validez de un string de entrada en relación con la clase.

````python
# Ejemplo: POO 13
class Vehiculo():
    contador = 0
    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')

    @staticmethod
    def get_vehiculos_posibles():
        return ['moto', 'automovil', 'tren', 'avión', 'bicicleta', 'barco']
````

````python
Vehiculo.get_vehiculos_posibles()
````

#### Métodos Decorados

+ Es posible que, según el escenario, se requiera <span style="color:magenta">decorar</span> ciertos métodos de una clase. 

+ Esto es posible siguiendo la misma estructura de <span style="color:magenta">decoradores</span> que ya se ha estudiado, pero con ciertos matices.

+ En el ejemplo siguiente se crea un decorador que audita las acciones de un objeto y saber que acción ha realizado.

````python
# Ejemplo: POO 14
class Vehiculo():
    contador = 0
    def __init__(self, nombre): # Constructor
        Vehiculo.contador += 1
        self.nombre = nombre # Atributo de instancia
        self.pos = [0, 0] # Posicion inicial

    @staticmethod
    def auditar(method):
        def wrapper(self, *args, **kwargs):
            print(f'Vehiculo {self.nombre} ejecutando {method.__name__}')
            return method(self, *args, **kwargs)
        return wrapper
    
    @auditar
    def mover(self, x, y):
        self.pos[0] += x
        self.pos[1] += y
    
    @auditar
    def reset(self):
        self.pos = [0, 0]
````

````python
moto = Vehiculo('Honda')
moto.mover(10,10)
print(moto.pos)
````

````python
moto.reset()
print(moto.pos)
````

#### Métodos Mágicos

+ Cuando se escribe <span style="color:limegreen">'hola mundo' * 3</span> ¿cómo sabe el objeto string <span style="color:limegreen">'hola mundo'</span> lo que debe hacer para multiplicarse con el objeto entero 3? O dicho de otra forma, ¿cuál es la implementación del operador <span style="color:limegreen">*</span> para <span style="color:limegreen">string</span> e <span style="color:limegreen">int</span>? En valores numéricos puede parecer evidente (siguiendo los operadores matemáticos), pero no es así para otros objetos. 

+ La solución que proporciona Python para estas (y otras) situaciones son los <span style="color:magenta">métodos mágicos</span> también llamados [métodos especiales](https://docs.python.org/es/3/reference/datamodel.html#special-method-names).

+ Los métodos mágicos comienzan y terminan por doble símbolo de subrayado (<span style="color:limegreen">\_\_</span>). Uno de los métodos mágicos más importantes es el constructor de una clase: <span style="color:limegreen">\_\_init\_\_()</span>.

+ Existe un método mágico asociado a cada operador (que se puede personalizar). Por ejemplo la comparación de dos objetos se realiza con el método <span style="color:limegreen">\_\_eq\_\_()</span>.

![Metodos Especiales](./Figuras/Clases_metodosmagicos.png)

**Caso particular: Sobrecarga de Operadores**

+ La redefinición de operadores utilizando los métodos mágicos se denomina <span style="color:magenta">sobrecarga de operadores</span>.

+ Esto sirve para lograr que los operadores conocidos de Python funcionen de determinada forma con los objetos definidos por el usuario.

+ Por ejemplo podemos sobrecargar el operador <span style="color:limegreen">*</span>, <span style="color:limegreen">+</span> , <span style="color:limegreen">/</span>, etc, para hacer que realice una operación específica y deseada con los objetos que se están definiendo.

````python
# Ejemplo: POO 15
# from __future__ import annotations
class Vehiculo():
    def __init__(self, nombre, color, cant_cilindros): # Constructor
        self.nombre = nombre # Atributo de instancia
        self.color = color
        self.cilindros = cant_cilindros

    # define un método de instancia
    def arrancar(self):
        print(f'Soy {self.nombre} y estoy funcionando')
    
    def __eq__(self, otro):
        return self.nombre == otro.nombre
    
    def __add__(self, otro):
        n_nombre = self.nombre + '-' + otro.nombre
        n_color = self.color + otro.color
        n_cilindros = (self.cilindros + otro.cilindros) / 2.
        return Vehiculo(n_nombre, n_color, n_cilindros)
    
    def __str__(self):
        return f'Vehiculo Marca: {self.nombre}, Color: {self.color}, Cilindros: {self.cilindros}'
````

````python
# Creación de objetos
b1 = Vehiculo('VW_Amarok', 'Gris', 4)
b2 = Vehiculo('VW_Amarok', 'Rojo', 6)
b3 = Vehiculo('Ford_Ranger', 'Azul', 4)
````

````python
# Invocación a __eq__
print(b1 == b2)
#print(b1*b3)
print(b1.__eq__(b3))
````

````python
# Invocación a __add__
b4 = b1 + b3
print(b4.nombre, b4.color, b4.cilindros)
````

**Gestores de Contexto**

+ Otra de las aplicaciones interesantes de los métodos mágicos es la de los <span style="color:magenta">gestores de contexto</span>. 

+ Un gestor de contexto permite aplicar una serie de acciones a la entrada y a la salida del contexto definido.

+ Hay dos métodos que son utilizados para implementar los gestores de contexto:

    + <span style="color:limegreen">\_\_enter\_\_()</span>: Acciones que se llevan a cabo al entrar al contexto.

    + <span style="color:limegreen">\_\_exit\_\_()</span>: Acciones que se llevan a cabo al salir del contexto.

````python
# Ejemplo: POO 16
# Medición de tiempos de ejecución
from time import time

class Timer():
    def __enter__(self):
        self.comienzo = time()
    
    def __exit__(self, tpo_ejec, valor_ejec, traza_ejec):
        # Omitir el gestor de la excepcion
        self.final = time()
        tpo_ejec = self.final - self.comienzo
        print(f'Tiempo de ejecución [seg]: {tpo_ejec:.5f}')
````

````python
# Definimos un contexto
with Timer():
    for _ in range(1_000_000):
        x = 2 ** 20
````

````python
# Definimos otro contexto
with Timer():
    x = 0
    for _ in range(1_000_000):
        x += 2 ** 20
````

### Herencia Múltiple

+ Aunque no está disponible en todos los lenguajes de programación, Python sí permite heredar de múltiples clases base.

+ Si en una clase se hace referencia a un método o atributo que no existe, Python lo buscará en todas sus clases base. Pero es posible que exista un conflicto en caso de que el método o el atributo buscado esté, a la vez, en varias clases base. En este caso, Python resuelve el conflicto a través del orden de resolución de métodos.

+ Todas las clases en Python disponen de un método especial llamado <span style="color:limegreen">mro() ("method resolution order")</span> que devuelve una lista de las clases que se visitarían en caso de acceder a un método o a un atributo:

    + También se puede acceder a la misma información usando el atributo <span style="color:limegreen">\_\_mro\_\_</span>.

    + Todos los objetos en Python heredan, en primera instancia, de <span style="color:limegreen">Object</span>. Esto se puede comprobar con el correspondiente <span style="color:limegreen">mro()</span> de cada objeto.

````python
# Ejemplo: POO 22
# Herencia Múltiple
class Nombre:
    def __init__(self, nombre):
        self._nombre = nombre

class Apellido:
    def __init__(self, apellido):
        self._apellido = apellido

class Saludo(Nombre, Apellido):
    def __init__(self, nombre: str, apellido: str, altura: float):
        Nombre.__init__(self, nombre)
        Apellido.__init__(self, apellido)
        self.altura = altura

    def saludar(self):
        print(f'Hola {self._nombre} {self._apellido} con altura: {self.altura}')
````

````python
saludo = Saludo('Alberto', 'Ramírez', 1.8)
````

````python
sal = Saludo('Carla', 'Gutierrez', 1.70)
````

````python
saludo.saludar()
````

````python
sal.saludar()
````

````python
Saludo.mro()
````

**Mixins**

Hay situaciones en las que nos interesa incorporar una clase base «independiente» de la jerarquía establecida, y sólo a efectos de tareas auxiliares o transversales. Esta aproximación podría ayudar a evitar conflictos en métodos o atributos reduciendo la ambigüedad que añade la herencia múltiple. A estas clases auxiliares se las conoce como <span style="color:magenta">mixins</span>.

**Agregación y Composición**

Aunque la herencia de clases nos permite modelar una gran cantidad de casos de uso en términos de <span style="color:yellow">is-a (es un)</span>, existen muchas otras situaciones en las que la agregación o la composición son una mejor opción. En este caso una clase se compone de otras clases: hablamos de una relación <span style="color:yellow">has a (tiene un)</span>.

Hay una sutil diferencia entre agregación y composición:
+ La <span style="color:magenta">composición</span> implica que el objeto utilizado no puede «funcionar» sin la presencia de su propietario.
+ La <span style="color:magenta">agregación</span> implica que el objeto utilizado puede funcionar por sí mismo.

![Agragacion vs Composición](./Figuras/agregacion_composicion.png)


````python
# Ejemplo: POO 24
# Agregación
class Herramienta:
    def __init__(self, nombre):
        self.nombre = nombre
        
    def __str__(self):
        return self.nombre.upper()

class Robot:
    def __init__(self, nombre, numero_serie, herramienta):
        self.nombre = nombre
        self.numero_serie = numero_serie
        self.herramienta = herramienta # agregación
        
    def __str__(self):
        return f'Robot {self.nombre} posee una {self.herramienta}'
````

### Iterables

+ Un objeto en Python se dice iterable si implementa el protocolo de iteración. Este
protocolo permite «entregar» de a un valor cada vez en forma de secuencia.

+ Hay muchos tipos de datos iterables en Python: <span style="color:magenta">strings, listas, tuplas, conjuntos, diccionarios</span>, etc.

+ Para ser un objeto iterable sólo es necesario implementar el método especial <span style="color:limegreen">\_\_iter\_\_()</span>. Este método debe proporcionar una referencia al objeto iterador que es quien se encargará de desarrollar el protocolo de iteración a través del método especial <span style="color:limegreen">\_\_next\_\_()</span>.

![Protocolo de Iteración](./Figuras/Protocolo_iteracion.png)

Cuando utilizamos un bucle <span style="color:limegreen">for</span> para recorrer los elementos de un iterable, ocurren varias cosas:

1. Se obtiene el objeto iterador del iterable.

2. Se realizan invocaciones sucesivas al método <span style="color:limegreen">\_\_next\_\_()</span> sobre dicho iterador para obtener cada elemento del iterable.

3. Se detiene la iteración cuando el iterador dispara la excepción <span style="color:limegreen">StopIteration</span>.

````python
# Ejemplo: POO 25
class Robot:
    def __init__(self, numero_serie):
        self.numero_serie = numero_serie * 5 # por que si
        
    def __repr__(self):
        return f'Robot: SN={self.numero_serie}.'

class Fabrica:
    def __init__(self, numero_robots):
        self.robots = [Robot(str(i)) for i in range(numero_robots)]
        self.numero_robots = numero_robots
        self.puntero = 0
    
    def __iter__(self):
        return self # El iterador es el propio objeto
    
    def __next__(self):
        # Protocolo de la iteración
        if self.puntero >= self.numero_robots:
            raise StopIteration
        robot = self.robots[self.puntero]
        self.puntero += 1
        return robot
````

````python
# Funcionamiento
for robot in Fabrica(10):
    print(robot)
````

### Otros Aspectos del Encapsulamiento

+ El <span style="color:magenta">encapsulamiento</span> permite agrupar datos y controlar su comportamiento en una clase. También permite controlar el acceso a los datos (atributos) y prevenir modificaciones no autorizadas.

+ Si, por ejemplo, se desea controlar el acceso a una variable o ejecutar una acción adicional al intentar acceder a un atributo, se debe hacer con los métodos *getters* y *setters* o usar propiedades.

#### Getters y Setters en Python

+ Los <span style="color:magenta">getters</span> son las funciones que nos permiten acceder a una variable privada. En Python se declaran creando una función con el decorador <span style="color:limegreen">@property</span>, es decir una propiedad.

+ Los <span style="color:magenta">setters</span> son las funciones que usamos para sobreescribir la información de una variable y se generan definiendo un método con el nombre de la variable sin guiones y utilizando como decorador el nombre de la variable sin guiones más <span style="color:limegreen">.setter</span>.

````python
# Ejemplo: POO 31
# Encapsulamiento

class ListadoBebidas():

    def __init__(self):
        self.__bebida = 'Gin Tonic'
        self.__bebidas_validas = ['Gin Tonic', 'Cerveza', 'Fernando']

    @property
    def bebida(self):
        return f'La bebida oficial es {self.__bebida}'

    @bebida.setter
    def bebida(self, bebida):
        if bebida in self.__bebidas_validas:
            self.__bebida = bebida
        else:
            raise ValueError(f'La bebida {bebida} no está en el listado de bebidas válidas')
````

````python
if __name__ == "__main__":
    bebidas = ListadoBebidas()
    print(bebidas.bebida)
    bebidas.bebida = 'Fernando'
    print(bebidas.bebida)
````

````python
# Ejemplo: POO 12
class Vehiculo():
    def __init__(self, nombre, color):
        self.__nombre = nombre
        self.color = color 
        self.tamanio = "grande"
        self.aspecto = "elegante"
        print(f'Soy {nombre} y he sido creado')
    
    #define un método
    def __arrancar(self):
        print(f'Soy {self.__nombre}, tengo color {self.color} y he Arrancado')

    @property
    def nombre(self):
        print('Dentro del getter')
        return f'El nombre es {self.__nombre}'

    @nombre.setter
    def nombre(self, nombre):
        print('dentro del setter')
        self.__nombre = nombre
````

````python
# Creación (instanciación) de un Objeto de tipo Vehiculo
automovil = Vehiculo('Honda', 'Azul') 
print(automovil.nombre)
````
