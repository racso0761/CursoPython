## <font color=#38bdf8>Enteros</font>

+ Especifican cantidades enteras (positivas y Negativas)

### <font color=#38bdf8>Operadores Relacionales</font>

En Python, los operadores relacionales (o de comparación) se usan para comparar dos valores y devuelven siempre un valor booleano (True o False).

*Operador* | *Descripción* | *Ejemplo* | *Resultado*
---|---|---|---
<span style="color:limegreen">==</span> | Igual a | 5 == 5 | True
<span style="color:limegreen">!=</span> | Distinto de | 5 != 3 | True
<span style="color:limegreen">></span> | Mayor que | 7 > 4 | True
<span style="color:limegreen">>=</span> | Mayor o igual a | 5 >= 5 | True
<span style="color:limegreen"><</span> | Menor que | 2 < 8 | True
<span style="color:limegreen"><=</span> | Menor o igual a | 3 <= 2 | False

### <font color=#38bdf8>Operadores Lógicos</font>

En Python, los operadores lógicos permiten combinar o modificar condiciones booleanas (True o False) para tomar decisiones en el flujo de un programa.

Los operadores lógicos son:

*Operador* | *Descripción* | *Ejemplo* | *Resultado*
---|---|---|---

### <font color=#38bdf8>Operadores de Identidad y Pertenencia</font>

Los operadores de identidad en Python se utilizan para comparar si dos variables hacen referencia al mismo objeto en la memoria. Estos operadores son útiles cuando deseas saber si dos objetos son literalmente el mismo, en lugar de solo tener el mismo valor.

Los operadores de pertenencia en Python son una herramienta útil para verificar si un elemento específico está presente en una secuencia, como una lista, un string o un conjunto. Estos operadores te permiten realizar comprobaciones rápidas de pertenencia sin necesidad de recorrer manualmente la secuencia.

*Operador* | *Descripción* | *Ejemplo* | *Resultado*
---|---|---|---
<span style="color:limegreen">is</span> | Identidad | a is b | ?
<span style="color:limegreen">is not</span> | Identidad | a is not b | ?
<span style="color:limegreen">in</span> | Pertenencia | 'a' in 'casa' | True
<span style="color:limegreen">not in</span> | Pertenencia | 'z' not in 'casa' | True

#### Comparación entre == e is

Es importante no confundir los operadores de identidad (<span style="color:limegreen">is</span>, <span style="color:limegreen">is not</span>) con el operador de igualdad (<span style="color:limegreen">==</span>). Mientras que <span style="color:limegreen">==</span> compara los valores de los objetos, <span style="color:limegreen">is</span> compara si los objetos ocupan el mismo espacio en la memoria.

### <font color=#38bdf8>Operadores Bit a Bit</font>

Los operadores bit a bit en Python son herramientas que permiten realizar operaciones a nivel de bits en números enteros. Estos operadores trabajan directamente sobre la representación binaria de los números, lo que es útil en aplicaciones que requieren manipulación de datos a bajo nivel, como en criptografía, programación de sistemas y redes.

En los ejemplos <span style="color:limegreen">a = 1</span> y <span style="color:limegreen">b = 0</span>

*Operador* | *Descripción* | *Ejemplo* | *Resultado*
---|---|---|---
<span style="color:limegreen">&</span> | AND bit a bit | a & b | 0
<span style="color:limegreen">\|</span> | OR bit a bit | a \| b | 1
<span style="color:limegreen">^</span> | XOR bit a bit | a ^ b | 1
<span style="color:limegreen">~</span> | NOT bit a bit (C1) | ~a | 0
<span style="color:limegreen"><<</span> | Desplazamiento a la Izquierda | 4 << 1 | 8
<span style="color:limegreen">>></span> | Desplazamiento a la Derecha | 4 >> 1 | 2

```python
a = 5
b = 3
print(f'ybb: {a & b}')
print(f'obb: {a | b}')
print(f'xbb: {a ^ b}')
print(f'binario:{~a}')
print(f'dibb: {a >> 1}')
print(f'ddbb:{a << 1}')
```

### <font color=#38bdf8>Información Detallada de Numeros de Coma Flotante</font>

En Python, <span style='color:limegreen'>sys.float_info</span> es una tupla con nombre (named tuple) que proporciona información detallada sobre la representación de números de punto flotante utilizada por el intérprete de Python.

Refleja el tipo `double` subyacente de C (generalmente precisión doble IEEE 754).

#### Significado de los atributos

+ <span style='color:yellow'>max</span>: Número flotante finito positivo representable de mayor valor.
+ <span style='color:yellow'>max_exp</span>: Exponente entero máximo para la base.
+ <span style='color:yellow'>max_10_exp</span>: Exponente entero máximo para la base 10.
+ <span style='color:yellow'>min</span>: Número flotante normalizado positivo más pequeño.
+ <span style='color:yellow'>min_exp</span>: Exponente entero mínimo para la base.
+ <span style='color:yellow'>min_10_exp</span>: Exponente entero mínimo para la base 10.
+ <span style='color:yellow'>dig</span>: Dígitos decimales aproximados de precisión.
+ <span style='color:yellow'>mant_dig</span>: Número de dígitos de la base en la mantisa.
+ <span style='color:yellow'>epsilon</span>: Diferencia entre 1.0 y el siguiente número flotante representable (épsilon de máquina).
+ <span style='color:yellow'>radix</span>: Base de la representación del exponente (normalmente 2).
+ <span style='color:yellow'>rounds</span>: Modo de redondeo de punto flotante (normalmente 1 = redondear al más cercano).

```python
import sys

# Máximo valor representable
max_float = sys.float_info.max
# Mínimo valor positivo normalizado
min_float = sys.float_info.min

print(f"Máximo float: {max_float}")
print(f"Mínimo float positivo normalizado: {min_float}")
```

```python
print(sys.float_info)
```

```python
print(f'Máximo: {sys.float_info.max}')
print(f'Exp máximo: {sys.float_info.max_exp}')
print(f'Exp 10 máximo: {sys.float_info.max_10_exp}')
print(f'Mínimo: {sys.float_info.min}')
print(f'Exp min: {sys.float_info.min_exp}')
print(f'Exp 10 min: {sys.float_info.min_10_exp}')
print(f'Dígitos Dec.: {sys.float_info.dig}')
print(f'Dig mantisa: {sys.float_info.mant_dig}')
print(f'Epsilon: {sys.float_info.epsilon}')
print(f'Radix: {sys.float_info.radix}')
print(f'Rounds: {sys.float_info.rounds}')
```

## <font color=#38bdf8>Introducción</font>

Este módulo proporciona acceso a las funciones matemáticas definidas en el estándar de C.

Estas funciones no pueden ser usadas con números complejos; para ello se usan las funciones del módulo <span style="color:limegreen">cmath</span> si requieres soporte para números complejos. La distinción entre las funciones que admiten números complejos y las que no se hace debido a que la mayoría de los usuarios no quieren aprender tantas matemáticas como se requiere para comprender los números complejos. Recibir una excepción en lugar de un resultado complejo permite la detección temprana del número complejo inesperado utilizado como parámetro, de modo que el programador pueda determinar cómo y porqué se generó en primer lugar.

Este módulo proporciona las funciones descritas a continuación. Excepto cuando se indique lo contrario explícitamente, todos los valores retornados son flotantes.

## <font color=#38bdf8>Teoría de Números y Funciones de Representación</font>

+ <span style="color:limegreen">math.ceil(x)</span>: Retorna el «techo» de x, el número entero más pequeño que es mayor o igual que x. Si x no es un flotante, converge a <span style="color:limegreen">x.__ceil__</span>.
+ <span style="color:limegreen">math.comb(n, k)</span>: Retorna el número de formas posibles de elegir <span style="color:limegreen">k</span> elementos de <span style="color:limegreen">n</span>, de forma ordenada y sin repetición.  
Se evalúa como: 
$$
\frac{n!}{(k! * (n - k)!)}
$$  
cuando $k \le n$ y como cero cuando $k \gt n$.

También llamado coeficiente binomial porque es equivalente al coeficiente del <span style="color:limegreen">k-ésimo</span> término en la expansión polinomial de $(1 + x)^n$.

Lanza una excepción del tipo <span style="color:limegreen">TypeError</span> si alguno de los argumentos no es un entero. Lanza una excepción del tipo <span style="color:limegreen">ValueError</span> si alguno de los argumentos es negativo.

+ <span style="color:limegreen">math.copysign(x, y)</span>: Retorna un flotante con la magnitud (valor absoluto) de <span style="color:limegreen">x</span> pero el signo de <span style="color:limegreen">y</span>. En plataformas que admiten ceros con signo, <span style="color:limegreen">copysign(1.0, -0.0)</span> retorna $-1.0$.

+ <span style="color:limegreen">math.fabs(x)</span>:  Retorna el valor absoluto de <span style="color:limegreen">x</span>.

+ <span style="color:limegreen">math.factorial(n)</span>: Retorna el factorial de <span style="color:limegreen">n</span> como un número entero. Lanza una excepción del tipo <span style="color:limegreen">ValueError</span> si <span style="color:limegreen">n</span> no es un entero o negativo. Obsoleto desde la versión 3.9: Aceptar flotantes con valores integrales (como 5.0) está obsoleto.

+ <span style="color:limegreen">math.floor(x)</span>: Retorna el «suelo» de <span style="color:limegreen">x</span>, el primer número entero menor o igual que <span style="color:limegreen">x</span>.

+ <span style="color:limegreen">math.fmod(x, y)</span>: Retorna <span style="color:limegreen">fmod(x, y)</span>, tal como se define en la biblioteca de C. Ten en cuenta que la expresión $x \% y$ de Python puede no retornar el mismo resultado. La intención del estándar de C es que <span style="color:limegreen">fmod(x, y)</span> sea exactamente (matemáticamente; con precisión infinita) igual a $x - n*y$ para algún número entero <span style="color:limegreen">n</span> tal que el resultado tenga el mismo signo que <span style="color:limegreen">x</span> y magnitud menor que <span style="color:limegreen">abs(y)</span>. La expresión $x \% y$ de Python retorna un resultado con el signo de <span style="color:limegreen">y</span> en su lugar, y es posible que no pueda calcularse con exactitud para argumentos flotantes. Por ejemplo, <span style="color:limegreen">fmod(-1e-100, 1e100)</span> es $-1e-100$, pero el resultado de $-1e-100 \% 1e100$ en Python es $1e100-1e-100$, que no se puede representar exactamente como un flotante, y se redondea sorprendentemente a $1e100$. Por esta razón, generalmente se prefiere la función <span style="color:limegreen">fmod()</span> cuando se trabaja con flotantes, mientras que se prefiere el uso de $x \% y$ de Python cuando se trabaja con enteros.

+ <span style="color:limegreen">math.frexp(x)</span>: Retorna la mantisa y el exponente de <span style="color:limegreen">x</span> como el par <span style="color:limegreen">(m, e)</span>. <span style="color:limegreen">m</span> es un flotante y <span style="color:limegreen">e</span> es un número entero tal que <span style="color:limegreen">x == m * 2**e</span> exactamente. Si <span style="color:limegreen">x</span> es cero, retorna $(0.0, 0)$, y retorna <span style="color:limegreen">0.5 <= abs(m) < 1</span> en caso contrario. Se utiliza como una forma portable de «extraer» la representación interna de un flotante.

+ <span style="color:limegreen">math.fsum(iterable)</span>: Retorna una suma precisa en coma flotante de los valores de un iterable. Evita la pérdida de precisión mediante el seguimiento de múltiples sumas parciales intermedias.

La precisión del algoritmo depende de las garantías aritméticas de IEEE-754 y del caso típico en el que se usa el redondeo a la mitad par (half-even) como método de redondeo. En algunas compilaciones que no son de Windows, la biblioteca de C subyacente utiliza la adición de precisión extendida y, ocasionalmente, puede realizar un doble redondeo en una suma intermedia, haciendo que el bit menos significativo tome el valor incorrecto.

+ <span style="color:limegreen">math.gcd(*integers)</span>: Retorna el máximo común divisor de los argumentos enteros. Si cualquiera de los argumentos no es cero, entonces el valor retornado es el entero positivo más grande que divide a todos los argumentos. Si todos los argumentos son cero, entonces el valor retornado es 0. <span style="color:limegreen">gcd()</span> sin argumentos retorna 0.

A partir de la versión 3.9 se ha agregado soporte para un número arbitrario de argumentos. Anteriormente sólo se soportaba dos argumentos.

+ <span style="color:limegreen">math.isclose(a, b, *, rel\_tol=1e-09, abs\_tol=0.0)</span>: Retorna <span style="color:limegreen">True</span> si los valores <span style="color:limegreen">a</span> y <span style="color:limegreen">b</span> están cerca el uno del otro y <span style="color:limegreen">False</span> en caso contrario.

Que dos valores se consideren cercanos o no, se determina de acuerdo con las tolerancias absolutas y relativas dadas.

   + <span style="color:limegreen">rel\_tol</span>: es la tolerancia relativa: esta es la diferencia máxima permitida entre <span style="color:limegreen">a</span> y <span style="color:limegreen">b</span>, en relación con el valor absoluto mayor de <span style="color:limegreen">a</span> o <span style="color:limegreen">b</span>. Por ejemplo, para establecer una tolerancia del $5\%$, pasa <span style="color:limegreen">rel_tol=0.05</span>. La tolerancia predeterminada es $1e-09$, lo que garantiza que los dos valores sean iguales considerando 9 dígitos decimales aproximadamente. <span style="color:limegreen">rel\_tol</span> debe ser mayor que cero.
   + <span style="color:limegreen">abs\_tol</span>: es la tolerancia absoluta mínima, útil para las comparaciones cercanas a cero. <span style="color:limegreen">abs\_tol</span> debe valer al menos cero.


Si no se encuentran errores, el resultado será: <span style="color:limegreen">abs(a-b) <= max(rel\_tol * max(abs(a), abs(b)), abs\_tol)</span>.

Los valores especiales de IEEE 754 <span style="color:limegreen">NaN</span>, <span style="color:limegreen">inf</span> e <span style="color:limegreen">-inf</span> se manejarán de acuerdo con las reglas del IEEE. Concretamente, <span style="color:limegreen">NaN</span> no se considera cercano a ningún otro valor, incluido <span style="color:limegreen">NaN</span>. Por su parte, \<span style="color:limegreen">inf</span> e <span style="color:limegreen">-inf</span> solo se consideran cercanos a sí mismos.

+ <span style="color:limegreen">math.isfinite(x)</span>: Retorna <span style="color:limegreen">True</span> si <span style="color:limegreen">x</span> no es infinito ni <span style="color:limegreen">NaN</span>, o <span style="color:limegreen">False</span> en caso contrario. (Ten en cuenta que 0.0 es considerado finito.)

+ <span style="color:limegreen">math.isinf(x)</span>: Retorna <span style="color:limegreen">True</span> si <span style="color:limegreen">x</span> es infinito positivo o negativo, o <span style="color:limegreen">False</span> en caso contrario.

+ <span style="color:limegreen">math.isnan(x)</span>: Retorna \<span style="color:limegreen">True</span> si <span style="color:limegreen">x</span> es <span style="color:limegreen">NaN</span>, o <span style="color:limegreen">False</span> en caso contrario.

+ <span style="color:limegreen">math.isqrt(n)</span>: Retorna la raíz cuadrada del número entero no negativo <span style="color:limegreen">n</span>. Es el resultado de aplicar la función <span style="color:limegreen">floor()</span> al valor exacto de la raíz cuadrada de <span style="color:limegreen">n</span>, o de forma equivalente, el mayor entero tal que $a^2 \le n$.

Para algunas aplicaciones, puede ser más conveniente tener el menor número entero a tal que $n \le a^2$, en otras palabras, el resultado de aplicar la función <span style="color:limegreen">ceil()</span> a la raíz cuadrada exacta de <span style="color:limegreen">n</span>. Para <span style="color:limegreen">n</span> positivo, esto se puede calcular usando <span style="color:limegreen">a = 1 + isqrt(n - 1)</span>.

+ <span style="color:limegreen">math.lcm(*integers)</span>: Retorna el mínimo común múltiplo de los argumentos enteros. Si todos los argumentos no son cero, entonces el valor retornado es el entero positivo más pequeño que es un múltiplo de todos los argumentos. Si cualquiera de los argumentos es cero, entonces el valor retornado es 0. <span style="color:limegreen">lcm()</span> sin argumentos retorna 1.

+ <span style="color:limegreen">math.ldexp(x, i)</span>: Retorna <span style="color:limegreen">x * (2**i)</span>. Esta es esencialmente la función inversa de <span style="color:limegreen">frexp()</span>.

+ <span style="color:limegreen">math.modf(x)</span>: Retorna la parte fraccionaria y entera de <span style="color:limegreen">x</span>. Ambos resultados son flotantes y tienen el mismo signo que <span style="color:limegreen">x</span>.

+ <span style="color:limegreen">math.nextafter(x, y, steps=1)</span>: Retorna el valor en coma flotante <span style="color:limegreen">steps</span> pasos después <span style="color:limegreen">x</span> hacia <span style="color:limegreen">y</span>.

Si <span style="color:limegreen">x</span> es igual a <span style="color:limegreen">y</span>, retorna <span style="color:limegreen">y</span>, a menos que <span style="color:limegreen">steps</span> sea cero.

Ejemplos:
<span style="color:limegreen">math.nextafter(x, math.inf)</span> va hacia el infinito positivo.
<span style="color:limegreen">math.nextafter(x, -math.inf)</span> va hacia el infinito negativo.
<span style="color:limegreen">math.nextafter(x, 0.0)</span> va hacia cero.
<span style="color:limegreen">math.nextafter(x, math.copysign(math.inf, x))</span> se aleja de cero.

+ <span style="color:limegreen">math.perm(n, k=None)</span>: Retorna el número de formas posibles de elegir <span style="color:limegreen">k</span> elementos de <span style="color:limegreen">n</span> elementos, sin repetición y en orden.

Se evalúa como 

$$
\frac{n!}{(n - k)!}
$$

cuando $k \le n$ y como cero cuando $k \gt n$.

If <span style="color:limegreen">k</span> no se especifica o es <span style="color:limegreen">None</span>, entonces <span style="color:limegreen">k</span> va hacia <span style="color:limegreen">n</span> y la función retorna $n!$.

Se lanza una excepción del tipo <span style="color:limegreen">TypeError</span> si alguno de los argumentos no es un entero. Se lanza una excepción del tipo <span style="color:limegreen">ValueError</span> si alguno de los argumentos es negativo.

+ <span style="color:limegreen">math.prod(iterable, *, start=1)</span>: Calcula el producto de todos los elementos en la entrada iterable. El valor <span style="color:limegreen">start</span> predeterminado para el producto es 1.

Cuando el iterable está vacío, retorna el valor inicial. Esta función está diseñada específicamente para su uso con valores numéricos y puede rechazar tipos no numéricos.

+ <span style="color:limegreen">math.remainder(x, y)</span>: Retorna el resto o residuo según la norma IEEE 754 de <span style="color:limegreen">x</span> con respecto a <span style="color:limegreen">y</span>. Para un valor <span style="color:limegreen">x</span> finito y un valor <span style="color:limegreen">y</span> finito distinto de cero, es la diferencia <span style="color:limegreen">x - n * y</span>, donde <span style="color:limegreen">n</span> es el número entero más cercano al valor exacto del cociente $x / y$. Si $x / y$ está exactamente en mitad de dos enteros consecutivos, el entero par más cercano se utiliza para <span style="color:limegreen">n</span>. Por lo tanto, el residuo <span style="color:limegreen">r = remainder(x, y)</span> siempre satisface <span style="color:limegreen">abs(r) <= 0.5 * abs(y)</span>.

Los casos especiales siguen el estándar IEEE 754: en particular, <span style="color:limegreen">remainder(x, math.inf)<7span> es <span style="color:limegreen">x</span> para todo <span style="color:limegreen">x</span> finito, y <span style="color:limegreen">remainder(x, 0)</span> junto a <span style="color:limegreen">remainder(math.inf, x)</span> lanzan una excepción del tipo <span style="color:limegreen">ValueError</span> para todo <span style="color:limegreen">x</span> que no sea <span style="color:limegreen">NaN</span>. Si el resultado de la operación residuo es cero, este cero tendrá el mismo signo que <span style="color:limegreen">x</span>.

En plataformas que utilizan la norma IEEE 754 para números en coma flotante binarios, el resultado de esta operación siempre es exactamente representable: no se introduce ningún error de redondeo.

+ <span style="color:limegreen">math.sumprod(p, q)</span>: Retorna la suma de productos de valores de dos iterables <span style="color:limegreen">p</span> y <span style="color:limegreen">q</span>.

Se lanza una excepción del tipo <span style="color:limegreen">ValueError</span> si las entradas no tienen la misma longitud.

Aproximadamente equivalente a:

<span style="color:limegreen">sum(itertools.starmap(operator.mul, zip(p, q, strict=True)))</span>

Para entradas flotantes y mixtas <span style="color:limegreen">int/float</span>, los productos intermedios y las sumas se calculan con precisión ampliada.

+ <span style="color:limegreen">math.trunc(x)</span>: Retorna <span style="color:limegreen">x</span> con la parte fraccionaria eliminada, dejando la parte entera. Esto redondea hacia 0: <span style="color:limegreen">trunc()</span> es equivalente a <span style="color:limegreen">floor()</span> para <span style="color:limegreen">x</span> positivo y equivalente a <span style="color:limegreen">ceil()</span> para <span style="color:limegreen">x</span> negativo. Si <span style="color:limegreen">x</span> no es un flotante, conduce a <span style="color:limegreen">x.\_\_trunc\_\_</span>, que debería retornar un valor Integral.

+ <span style="color:limegreen">math.ulp(x)</span>: Retorna el valor del bit menos significativo del flotante <span style="color:limegreen">x</span>:

    + <span style="color:limegreen">x</span> es un <span style="color:limegreen">NaN</span>  (not a number), retorna <span style="color:limegreen">x</span>.
    + <span style="color:limegreen">x</span> es negativo, retorna <span style="color:limegreen">ulp(-x)</span>.
    + Si <span style="color:limegreen">x</span> es un infinito positivo, retorna <span style="color:limegreen">x</span>.
    + Si <span style="color:limegreen">x</span> es igual a cero, retorna el flotante representable desnormalizado positivo más pequeño (menor que el flotante normalizado positivo mínimo, <span style="color:limegreen">sys.float\_info.min</span>).
    + Si <span style="color:limegreen">x</span> es igual al flotante representable positivo más pequeño, retorna el bit menos significativo de <span style="color:limegreen">x</span>, de tal manera que el primer flotante menor que <span style="color:limegreen">x</span> es <span style="color:limegreen">x - ulp(x)</span>.
    + De lo contrario (<span style="color:limegreen">x</span> es un número finito positivo), retorna el valor del bit menos significativo de <span style="color:limegreen">x</span>, de tal forma que el primer flotante mayor a <span style="color:limegreen">x</span> es <span style="color:limegreen">x + ulp(x)</span>.

<span style="color:magenta">ULP</span> significa <span style="color:magenta">Unit in the Last Place (unidad en el último lugar)</span>.

Ten en cuenta que <span style="color:limegreen">frexp()</span> y <span style="color:limegreen">modf()</span> tienen un patrón de llamada/retorno diferente al de sus equivalentes en C: toman un solo argumento y retornan un par de valores, en lugar de retornar su segundo valor de retorno a través de un parámetro de salida (no existe tal cosa en Python).

Para las funciones <span style="color:limegreen">ceil()</span>, <span style="color:limegreen">floor()</span> y <span style="color:limegreen">modf()<7span>, ten en cuenta que todos los números de coma flotante de magnitud suficientemente grande son enteros exactos. Los flotantes de Python normalmente no tienen más de 53 bits de precisión (lo mismo que el tipo double de C en la plataforma), en cuyo caso cualquier flotante <span style="color:limegreen">x</span> con <span style="color:limegreen">abs(x) >= 2**52</span> no necesariamente tiene bits fraccionarios.

## <font color=#38bdf8>Funciones de Potencia y Logarítmicas</font>

+ <span style="color:limegreen">math.cbrt(x)</span>: Retorna la raíz cúbica de <span style="color:limegreen">x</span>.

+ <span style="color:limegreen">math.exp(x)</span>: Devuelve <span style="color:limegreen">e</span> elevado a la potencia <span style="color:limegreen">x</span>, where $e = 2.718281\ldots$ es la base del logarítmo natural. Esto es usualmente mas exacto que <span style="color:limegreen">math.e**x</span> or <span style="color:limegreen">pow(math.e, x)</span>.

+ <span style="color:limegreen">math.exp2(x)</span>: Devuelve 2 elevado a la potencia <span style="color:limegreen">x</span>.

+ <span style="color:limegreen">math.expm1(x)</span>: Devuelve \<span style="color:limegreen">e</span> elevado a la potencia de <span style="color:limegreen">x</span>, menos 1. Aquí <span style="color:limegreen">e</span> es la base del logarítmo natural. Para flotantes pequeños <span style="color:limegreen">x</span>, la resta en <span style="color:limegreen">exp(x) - 1</span> puede resultar en una significativa pérdida de precisión; la función <span style="color:limegreen">expm1()</span> da una forma de calcular esta cantidad a precisión completa:

```python
from math import exp, expm1
exp(1e-5) - 1  # gives result accurate to 11 places
expm1(1e-5)    # result accurate to full precision
```

+ <span style="color:limegreen">math.log(x[, base])</span>: Con un argumento, devuelve el logaritmo natural de <span style="color:limegreen">x</span>.  
Con dos argumentos, devuelve el logaritmo de <span style="color:limegreen">x</span> en la <span style="color:limegreen">base</span> dada, calculado como $log(x)/log(base)$.

+ <span style="color:limegreen">math.log1p(x)</span>: Devuelve el logaritmo natural de <span style="color:limegreen">1+x</span> (base <span style="color:limegreen">e</span>). El resultado es calculado en una forma en la cual es exacto para valores de <span style="color:limegreen">x</span> cercanos a cero.

+ <span style="color:limegreen">math.log2(x)</span>: Retorna el logaritmo en base 2 de <span style="color:limegreen">x</span>. Esto suele ser más preciso que <span style="color:limegreen">log(x, 2)</span>.

<span style="color:orange">Nota</span>: Ver también <span style="color:limegreen">int.bit_length()</span> retorna el número de bits necesarios para representar un entero en binario, excluyendo el signo y los ceros iniciales.

+ <span style="color:limegreen">math.log10(x)</span>: Retorna el logaritmo en base 10 de <span style="color:limegreen">x</span>. Esto suele ser más preciso que <span style="color:limegreen">log(x, 10)</span>.

+ <span style="color:limegreen">math.pow(x, y)</span>: Retorna <span style="color:limegreen">x</span> elevado a la potencia <span style="color:limegreen">y</span>. Los casos excepcionales siguen el estándar IEEE 754 en la medida de lo posible. En particular, <span style="color:limegreen">pow(1.0, x)</span> y \<span style="color:limegreen">pow(x, 0.0)</span> siempre retornan $1.0$, incluso cuando <span style="color:limegreen">x</span> es cero o <span style="color:limegreen">NaN</span>. Si tanto <span style="color:limegreen">x</span> como <span style="color:limegreen">y</span> son finitos, <span style="color:limegreen">x</span> es negativo e <span style="color:limegreen">y</span> no es un número entero, entonces <span style="color:limegreen">pow(x, y)</span> no está definido y se lanza una excepción de tipo <span style="color:limegreen">ValueError</span>.
A diferencia del operador <span style="color:limegreen">** </span>, <span style="color:limegreen">math.pow()</span> convierte ambos argumentos al tipo float. Utiliza <span style="color:limegreen">**</span> o la función <span style="color:limegreen">pow()</span> para calcular potencias enteras exactas.
Los casos especiales <span style="color:limegreen">pow(0.0, -inf)</span> y <span style="color:limegreen">pow(-0.0, -inf)</span> se cambiaron para devolver <span style="color:limegreen">inf</span> en lugar de generar una excepción del tipo <span style="color:limegreen">ValueError</span>, por consistencia con IEEE 754.

+ <span style="color:limegreen">math.sqrt(x)</span>: Retorna la raíz cuadrada de <span style="color:limegreen">x</span>.

## <font color=#38bdf8>Introducción</font>

Si bien el conjunto de los números reales <span style="color:magenta">R</span>, parece contener todos los números que podríamos llegar a necesitar. Existe todavía una dificultad, el hecho de que sólo se pueden tomar raíces cuadradas de los números positivos (o cero) y no de los negativos. Desde el punto de vista matemático resultaría conveniente poder extraer raíces cuadradas de números negativos tanto como de números positivos. Por tal motivo, que tal si "inventamos" una raíz cuadrada para el número $-1$ y la expresamos con el símbolo <span style="color:magenta">j</span>, de modo que tenemos

$$
j^2=−1
$$

La cantidad <span style="color:magenta">j</span> no puede ser, por supuesto, un número real puesto que el producto de un número real por sí mismo es siempre positivo (o cero, si el propio número es el cero). Por esta razón se ha aplicado convencionalmente el término <span style="color:magenta">"imaginario"</span> a los números cuyos cuadrados son negativos. Sin embargo, es importante resaltar el hecho de que estos números imaginarios no son menos reales que los números reales a los que estamos acostumbrados.

Cuando estos números imaginarios se combinan con los números reales obtenemos lo que se conoce como números complejos; de esta forma, los números complejos vienen a completar a los números reales y nos permiten realizar todo tipo de operaciones algebraicas.

Los números complejos incluyen todas las raíces de los polinomios, a diferencia de lo que pasaba cuando solo teníamos a los números reales. Todo número complejo puede representarse como la suma de un número real y un número imaginario, el cual es un múltiplo real de la unidad imaginaria, que se indica con la letra \<span style="color:magenta">j</span>. O sea, que los números complejos tienen:

+ Una <span style="color:magenta">parte real</span>, como por ejemplo $3$ o $7.5$.
+ Y otra <span style="color:magenta">imaginaria</span>, como $5j$ o $-7.2j$.</li>

Veamos las distintas operaciones matemáticas que podemos hacer con estos números.

Como se puede ver, la parte imaginaria está acompañada de <span style="color:magenta">j</span>, aunque es también común usar la <span style="color:magenta">i (de imaginario)</span>. Un número complejo tiene la siguiente forma:

<span style="color:limegreen">parte\_real + parte\_imaginaria * j</span>

Por ejemplo:

$$
a = 5 + 5j
$$
$$
b = 1.3 - 7j
$$

## <font color=#38bdf8>Números Complejos en Python</font>

Los números complejos en Python pueden ser creados sin tener que importar ninguna librería debido a que la biblioteca <span style="color:limegreen">cmath</span> es del tipo built-in. Basta con hacer lo siguiente:

```python
c = 3 + 5j
print(c)       #(3+5j)
print(type(c)) #<class 'complex'>
```

Se puede ver como la clase que representa a los complejos en Python se llama <span style="color:limegreen">complex</span>. Una vez creado, es posible acceder a la parte real con <span style="color:limegreen">real</span> y a la imaginaria con <span style="color:limegreen">imag</span>.

```python
c = 3 + 5j
print(c.real) #3.0
print(c.imag) #5.0
```

También se puede crear un número complejo haciendo uso de <span style="color:limegreen">complex</span>, pero sin usar la <span style="color:limegreen">j</span>.

```python
c = complex(3,5)
print(c) #(3+5j)
```

## <font color=#38bdf8>Operaciones con Números Complejos</font>

Usando variables del tipo <span style="color:limegreen">complex</span>, podemos realizar las operaciones más comunes de los números complejos.

### <font color=#38bdf8>Suma de Complejos</font>

Para sumar números complejos, se suman las partes reales por un lado, y las imaginarias por otro. Es decir,

$$
(a+bi)+(c+di)=(a+c)+(b+d)i
$$

```python
a = 1 + 3j
b = 4 + 1j
print(a+b) #(5+4j)
```

### <font color=#38bdf8>Resta de Complejos</font>

Para restar, se restan las partes reales por un lado y las imaginarias por otro.

$$
(a+bi)-(c+di)=(a-c)+(b-d)i
$$

```python
a = 1 + 3j
b = 4 + 1j
print(a-b) #(-3+2j)
```

### <font color=#38bdf8>Producto por un Escalar</font>

Para calcular el producto escalar de un número complejo, multiplicamos al escalar por cada una de sus partes, la real y la imaginaria. Es decir que:

$$
r * (a+bi)=(r * a)+(r * b)i
$$

```python
a = 3 * (2 + 3j)
print(a)
```

### <font color=#38bdf8>Multiplicación de Complejos</font>

La multiplicación es algo más compleja. Si multiplicamos $a+bj$ por $c+dj$, se puede demostrar fácilmente que el resultado es $(ac-bd)$ para la parte real y $(ad+bc)$ para la imaginaria.

$$
(a+bi)(c+di)=ac+adi+bci+bdi^2
$$

En este punto, debemos recordad que $i^2$ es igual a $-1$; lo que nos facilita la solución del cálculo de la multiplicación. También existe una formula más simple para obtener el resultado de la multiplicación de dos números complejos, que es:

$$
(a+bi)(c+di)=(ac−bd)+(ad+bc)i
$$

```python
a = 1 + 3j
b = 4 + 1j
print(a*b) #(1+13j)
```

### <font color=#38bdf8>Igualdad de Números Complejos</font>

Dos números complejos van a ser iguales si y solo si sus partes reales e imaginarias son iguales. Es decir que:

$$
(a+bi)=(c+di)⟺a=c∧b=d
$$

```python
a = 3 + 2j
b = 3 + 2j
print(a == b)
```

### <font color=#38bdf8>Complejos Conjugados</font>

Se puede realizar el conjugado de un número complejo en Python con el método <span style="color:limegreen">conjugate()</span>. Calcular el conjugado consiste en negar la parte imaginaria, es decir, cambiar su signo de <span style="color:limegreen">+</span> a <span style="color:limegreen">-</span> y viceversa.

Tener en cuenta que la multiplicación de un número complejo por su conjugado, responde a la formula:

$$
(a+bi)(a−bi)=a^2+b^2
$$

```python
a = 1 + 1j
print(a.conjugate()) #(1-1j)
```

### <font color=#38bdf8>División de Complejos</font>

Para dividir dos números complejos, debemos utilizar el conjugado; ya que para realizar la división debemos multiplicar tanto el divisor como el dividendo por el conjugado del divisor. Así, por ejemplo si quisiéramos dividir:

$$
\frac{2+3i}{4−5i}
$$

Debemos realizar el siguiente cálculo:

$$
\frac{2+3i}{4−5i} \times \frac{4+5i}{4+5i}
$$

Entonces se puede resolver la división de la siguiente forma:

$$
\frac{2+3i}{4−5i} ×\frac{4+5i}{4+5i} = \frac{8+10i+12i+15i^2}{16+25}=\frac{−7+22i}{41} = −\frac{7}{41}+ \frac{22}{41}i
$$

Para simplificar el procedimiento, y no tener que realizar tantos cálculos, podríamos utilizar la siguiente formula:

$$
\frac{(a+bi)}{(c+di)}=\frac{(ac+bd)+(bc−ad)i}{c^2+d^2} = \left(\frac{ac+bd}{c^2+d^2}+\frac{(bc−ad)i}{c^2+d^2}\right)
$$

```python
a = 2 + 3j
b = 4 - 5j
print(a/b)
```

Algunas de las cosas que puedes hacer son las siguientes:

+ <span style="color:magenta">Calcular la forma polar</span>, es decir módulo y ángulo.

## <font color=#38bdf8>Conversión a y desde Coordenadas Polares</font>

Un número complejo <span style="color:limegreen">z</span> de Python se almacena internamente utilizando coordenadas rectangulares o cartesianas. Está completamente determinada por su parte real <span style="color:limegreen">z.real</span> y su parte imaginaria <span style="color:limegreen">z.imag</span>.

Las coordenadas polares dan una alternativa a la representación de números complejos. En las coordenadas polares, un número complejo <span style="color:limegreen">z</span> se define por los módulos <span style="color:magenta">r</span> y el ángulo de fase $\phi$. El módulo <span style="color:magenta">r</span> es la distancia desde <span style="color:limegreen">z</span> hasta el origen, mientras que la fase $\phi$ es el ángulo que va en contra de las agujas del reloj, medido en radianes, desde el eje positivo de las $x$ hasta el segmento de linea que une el origen con <span style="color:limegreen">z</span>.

Las siguientes funciones pueden ser usadas para convertir desde coordenadas rectangulares nativas hasta coordenadas polares y viceversa.

```python
import cmath
```

+ <span style="color:limegreen">cmath.phase(z)</span>: Devuelve la fase de <span style="color:limegreen">z</span> (también conocida como argumento de <span style="color:limegreen">z</span>), como un flotante. <span style="color:limegreen">phase(z)</span> es equivalente a <span style="color:limegreen">math.atan2(z.imag, z.real)</span>. El resultado se encuentra en el rango $[-\pi, \pi]$, y el branch cut para esta operación se encuentra a lo largo del eje real negativo. El signo del resultado es el mismo que el signo de <span style="color:limegreen">z.imag</span>, incluso cuando <span style="color:limegreen">z.imag</span> es cero:

```python
cmath.phase(complex(-1.0, 0.0))
cmath.phase(complex(-1.0, -0.0))
```

+ <span style="color:limegreen">cmath.polar()</span>: Retorna la representación de <span style="color:limegreen">z</span> en coordenadas polares. Retorna un par <span style="color:limegreen">(r, phi)</span> donde <span style="color:limegreen">r</span> es el módulo de <span style="color:limegreen">z</span> y <span style="color:limegreen">phi</span> ($\phi$) es la fase de <span style="color:limegreen">z</span>. <span style="color:limegreen">polar(z)</span> es equivalente a <span style="color:limegreen">(abs(x), phase(x))</span>.

+ <span style="color:limegreen">cmath.rect(r, phi)</span>:  Devuelve el número complejo <span style="color:limegreen">z</span> con coordenadas polares <span style="color:limegreen">r</span> y <span style="color:limegreen">phi</span> ($\phi$). Equivalente a <span style="color:limegreen">complex(r * math.cos(phi), r * math.sin(phi))</span>.

## <font color=#38bdf8>Funciones Logarítmicas y de Potencias</font>

+ <span style="color:limegreen">cmath.exp(z)</span>: Retorna <span style="color:limegreen">e</span> elevado a la potencia de <span style="color:limegreen">z</span>, donde <span style="color:limegreen">e</span> es la base de los logaritmos naturales.

+ <span style="color:limegreen">cmath.log(z[, base])</span>: Devuelve el logaritmo de <span style="color:limegreen">z</span> en la base dada. Si la base no se especifica, devuelve el logaritmo natural de <span style="color:limegreen">z</span>. Hay un branch cut, de 0 a lo largo del eje real negativo hasta $-\infty$.

+ <span style="color:limegreen">cmath.log10(z)</span>: Retorna el logaritmo en base de 10 de <span style="color:limegreen">z</span>. Tiene el mismo tramo que <span style="color:limegreen">log()</span>.

+ <span style="color:limegreen">cmath.sqrt(x)</span>: Retorna la raíz cuadrada de <span style="color:limegreen">z</span>. Tiene el mismo tramo que <span style="color:limegreen">log()</span>.

## <font color=#38bdf8>Funciones Trigonométricas</font>

+ <span style="color:limegreen">cmath.acos(z)</span>:  Devuelve el arcocoseno de <span style="color:limegreen">z</span>. Hay dos branch cuts: Uno se extiende desde la derecha desde 1 a lo largo del eje real hasta $\infty$. El otro se extiende hacia la izquierda desde $-1$ a lo largo del eje hasta $-\infty$.

+ <span style="color:limegreen">cmath.asin(z)</span>: Retorna el arcoseno de <span style="color:limegreen">z</span>. Este tiene los mismos tramos que <span style="color:limegreen">cmath.acos()</span>.

+ <span style="color:limegreen">cmath.atan(z)</span>: Devuelve el arcotangente de <span style="color:limegreen">z</span>. Hay 2 branch cuts: Uno se extiende desde $1j$  a lo largo del eje imaginario hasta $\infty j$ y el otro se extiende desde $-1j$ a lo largo del eje imaginario hasta $-\infty j$.

+ <span style="color:limegreen">cmath.cos(z)</span>: Retorna el coseno de <span style="color:limegreen">z</span>.

+ <span style="color:limegreen">cmath.sin(z)</span>: Retorna el seno de <span style="color:limegreen">z</span>.

+ <span style="color:limegreen">cmath.tan(z)</span>: Retorna la tangente de <span style="color:limegreen">z</span>.

## <font color=#38bdf8>Funciones de Clasificación</font>

+ <span style="color:limegreen">cmath.isfinite(z)</span>: Retorna <span style="color:limegreen">True</span> si tanto la parte imaginaria como real de <span style="color:limegreen">z</span> son finitas, y <span style="color:limegreen">False</span> en cualquier otro caso.

+ <span style="color:limegreen">cmath.isinf(z)</span>: Retorna <span style="color:limegreen">True</span> si la parte real o la imaginaria de <span style="color:limegreen">z</span> es infinito, y <span style="color:limegreen">False</span> en el caso contrario.

+ <span style="color:limegreen">cmath.isnan(z)</span>: Retorna <span style="color:limegreen">True</span> tanto si la parte real o imaginaria de <span style="color:limegreen">z</span> es <span style="color:limegreen">NaN</span>, y <span style="color:limegreen">False</span> en cualquier otro caso.

+ <span style="color:limegreen">cmath.isclose(a, b, *, rel\_tol=1e-09, abs\_tol=0.0)</span>: Retorna <span style="color:limegreen">True</span> si los valores <span style="color:limegreen">a</span> y <span style="color:limegreen">b</span> son cercanos el uno al otro y <span style="color:limegreen">False</span> de otro modo. Que dos valores sean o no considerados como cercanos es determinado de acuerdo al valor absoluto y las tolerancias relativas.

    + <span style="color:limegreen">rel\_tol</span> es la tolerancia relativa - es el máximo valor permitido de la resta entre <span style="color:limegreen">a</span> y <span style="color:limegreen">b</span>, relativo al valor absoluto más grande de <span style="color:limegreen">a</span> o <span style="color:limegreen">b</span>. Por ejemplo, para fijar una tolerancia del $5\%$, usar <span style="color:limegreen">rel\_tol=0.05</span>. El valor de tolerancia por defecto es $1e-09$, lo que asegura que los dos valores son los mismos en aproximadamente 9 dígitos decimales. <span style="color:limegreen">rel\_tol</span> debe ser mayor que cero.

    + <span style="color:limegreen">abs\_tol</span> es la tolerancia mínima absoluta - útil a la hora de hacer comparaciones cercanas al cero. <span style="color:limegreen">abs\_tol</span> debe ser al menos cero.

    Si no ocurren errores, el resultado será: <span style="color:limegreen">abs(a-b) <= max(rel\_tol * max(abs(a), abs(b)), abs\_tol)</span>.

    Los valores especiales IEEE 754 de <span style="color:limegreen">NaN</span>, <span style="color:limegreen">inf</span> y <span style="color:limegreen">-inf</span> serán manejados de acuerdo al estándar de IEEE. Especialmente, <span style="color:limegreen">NaN</span> no se considera cercano a ningún otro valor, incluido <span style="color:limegreen">NaN</span>. <span style="color:limegreen">inf</span> y <span style="color:limegreen">-inf</span> solo son considerados cercanos a sí mismos.
