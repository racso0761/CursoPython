# Calcular los Mínimos de una Función Escalar en un Intervalo

Suponga se desean calcular los mínimos de la función:

$$
y = -\cos \left(\frac{a\pi x}{b}\right) + c x^d
$$

Para esto se deben importar las bibliotecas y utilizar la función <span style="color:limegreen">fminbound()</span> que permite calcular los mínimos de una función. 

```python
# Ejemplo SciPy 1
import numpy as np # Importamos numpy con el alias np
import scipy as sp # Importamos scipy con el alias sp
import matplotlib.pyplot as plt
# Importamos fmindbound desde scipy.optimize
from scipy.optimize import fminbound
```

ver: [SciPy - Optimize - fminbound](https://docs.scipy.org/doc/scipy/reference/generated/scipy.optimize.fminbound.html#scipy.optimize.fminbound) Minimización Acotada para Funciones Escalares

```python
# Se define la funcion
def mi_funcion(x, a, b, c, d):
    y = -np.cos(a * np.pi * x / b) + c * x**d
    return y
 
# Definimos los coeficientes a, b, c, d
a = 2
b = 0.5
c = 0.05
d = 2
 
# Definimos el intervalo de busqueda del minimo
x1 = 0.2
x2 = 0.6
xt = np.arange(0, 1, .01)
yt = -np.cos(a * np.pi * xt / b) + c * xt**d
```

```python
# Calculamos el minimo local de la funcion entre x1 y x2 
x_minimo = fminbound(mi_funcion, x1, x2, args = (a,b,c,d))
ysol = mi_funcion(x_minimo, a, b, c, d)
```

```python
# Presentamos la grafica y en pantalla el resultado
print (f'El minimo esta en x = {x_minimo:.3f}, y = {ysol:.3f}')
 
plt.plot(xt, yt)
plt.plot(x_minimo, ysol, 'X')
plt.axvline(x = .2, color = 'r', linestyle=":")
plt.axvline(x = .6, color = 'g', linestyle=':')
plt.grid()
```

# Ajuste de una Función

Caso real: 

> Un supermercado tiene unas previsiones de venta de pescado durante las primeras 5 horas que está abierto al público y estas vienen determinadas por la siguiente función:

$$
ventas = a e^{\left(\frac{b x^2}{2 d^2}\right)} + c x
$$

El objetivo es mejorar las ventas a partir de los coeficientes desconocidos ``a``, ``b``, ``c``  y ``d``.

```python
# Ejemplo SciPy 2
import numpy as np # Importamos numpy como el alias np
import scipy as sp # Importamos scipy como el alias sp
from scipy.optimize import curve_fit # Importamos curve_fit de scipy.optimize
import scipy as sp # Importamos scipy como el alias sp
import matplotlib.pyplot as plt # Importamos matplotlib.pyplot como el alias plt.
 
def mi_funcion(x, a, b, c, d):
  return a * np.exp(-b * x**2 /(2 * d**2)) + c * x

x = np.linspace(0, 5,40)
y = mi_funcion(x, 2.5, 1.3, 0.5,1)
```

```python
# Añadimos ruido
def ruido(x,y,k):
  yn = y + k * np.random.normal(size = len(x))
  return yn

yn = ruido(x, y, .075)
```

```python
# Ajustamos los datos experimentales a nuestra funcion y los almacenamos
coeficientes_optimizados, covarianza_estimada = curve_fit(mi_funcion, x, yn)

# Mostramos los coeficientes calculados
print ("Coeficientes optimizados:", coeficientes_optimizados)
print ("Covarianza estimada:\n", covarianza_estimada)
```

```python
# Creamos la figura
plt.figure()

# Origen de datos
# plt.plot(x, y, 'gp', markersize=3, label='Origen')
# Dibujamos los datos ruido
plt.plot(x,yn,'ro', label = 'Experimental')
```

```python
# mantenemos la figura
# plt.hold(True)
results=mi_funcion(x,coeficientes_optimizados[0],coeficientes_optimizados[1],coeficientes_optimizados[2], coeficientes_optimizados[3])
 
plt.plot(x,results, label = 'Ajuste')
plt.legend()
plt.xlabel('Tiempo (h)')
plt.ylabel('Ventas del supermercado en pescado x100 ($)')

plt.show()
```

# Calculo de Integrales

Suponemos que se desea calcular la integral de:

$$
y = e^{-x}
$$

```python
# Ejemplo SciPy 3
import scipy as sp
from scipy import integrate
import numpy as np

def integral_1(limite_inferior, limite_superior):
    exponencial_decreciente = lambda x: np.exp(-x)
    # Los devuelvo
    return integrate.quad(exponencial_decreciente ,limite_inferior,limite_superior)

limite_inferior = 0
limite_superior = np.inf
resultado, error = integral_1(limite_inferior, limite_superior)
print (f'La integral entre {limite_inferior:.2f} e {limite_superior:.2f} es:\
    \nIntegral: {resultado}\nError Absoluto Estimado: {error}')
```

# Cálculo de Raices de un Polinomio

Dado el siguiente polinomio:

$$
p(x) = 4.3 x^3 + 9 x^2 + 0.6 x - 1
$$

Se desea calcular las raices.

```python
# Ejemplo SciPy 5
import numpy as np
import matplotlib.pyplot as plt

# Creamos un polinomio
polinomio = [4.3,9,.6,-1] # polinomio = 4.3 x^3 + 9 x^2 + 0.6 x - 1
# array
x = np.arange(-4,2,.05)
# Evaluamos el polinomio en x mediante polyval.
y = np.polyval(polinomio,x)
```

```python
# Calculamos las raices del polinomio 
raices = np.roots(polinomio)

# Evaluamos el polinomio en las raices
s = np.polyval(polinomio,raices)

# Las presentamos en pantalla
print (f'Las raices son {raices[0]:.2f}, {raices[1]:.2f}, {raices[2]:.2f}')
```

```python
# Creamos la figura
plt.figure

# Dibujamos
plt.plot(x, y, '-', label = 'y(x)')

# Dibujamos en la figura anterior
#plt.hold('on')

# Dibujamos
plt.plot(raices.real, s.real, 'ro', label = 'Raices')

# Etiquetas 
plt.xlabel('x')
plt.ylabel('y')
plt.title('Raices de un polinomio de x^3')

# Leyenda
plt.legend()

# Grilla
plt.grid()

# Mostramos la figura en pantalla
plt.show()
```

## Resolviendo un Sistema de Ecuaciones Lineales

Dado el sistema:

\begin{align*}
2 x + y &= 5\\
4 y - 5 x &= 7
\end{align*}

Su solución es se encuentra usando el método <span style="color:magenta">solve()</span>:

```python
# Ejemplo: SciPy 6
import numpy as np
from scipy import linalg

# Crea el arreglo de entrada
a = np.array([[2, 1], [4, -5]])

# Crea el arreglo solución
b = np.array([[5], [7]])

# Solución: encuentra x e y que satisface el sistema
x = linalg.solve(a, b)

# Muestra los resultados
print(x)

# Verifica los resultados
print("\n Verificando resultados.")
print(a.dot(x) - b)
```

## Determinante de una Matriz

Se crea una matriz *cuadrada* y luego se utiliza el método <span style="color:magenta">det()</span>.

```python
# Ejemplo: SciPy 7
from scipy import linalg
import numpy as np

# Crea la matriz cuadrada
mat = np.array([[1, 2, 3], [3, 4, 5], [7, 8, 6]])

# Calcula el determinante
x = linalg.det(mat)

# Muestra el resultado
print(f'Matriz\n{mat}')
print(f'\nDeterminante de la matriz: {x}')
```

## Inversa de una Matriz

Se utiliza el método <span style="color:magenta">inv()</span> para calcular la inversa de una matriz.

```python
# Ejemplo: SciPy 8
from scipy import linalg
import numpy as np

# Crea la matriz cuadrada
mat = np.array([[1, 2, 3], [3, 4, 5], [7, 8, 6]])

# Calculo de la Inversa
mat2 = linalg.inv(mat)

# Resultado
print(f'Matriz original: \n{mat}')
print(f'\nInversa: \n{mat2}')

# Verificación
i = np.dot(mat, mat2)

print('\nVerificación:\n',i)
```

## Descomposición de Valor Singular (SVD)

Se usa el método <span style="color:magenta">svd()</span> para descomponer una matriz en 2 matrices ortonormales y una matriz diagonal.

>La descomposición de valor singular (Singular Value Decomposition o SVD, en inglés) es una técnica fundamental en el análisis numérico y la reducción de dimensionalidad.
 
>Las aplicaciones de SVD son variadas y van desde la compresión de imágenes y reducción de ruido en señales hasta el análisis de componentes principales (PCA), recomendación de sistemas, procesamiento de lenguaje natural, y mucho más.

```python
# Ejemplo: SciPy 9
from scipy import linalg
import numpy as np

# Crea la matriz
mat = np.array([[1, 4], [9, 3]])

# Realiza la descomposición generando una tupla
mat2 = linalg.svd(mat)

# Muestra los resultados
print(mat2)
```

## Descomposición LU

Se puede descomponer una matriz en dos matrices, una triangular superior y la otra triangular inferior.

```python
# Ejemplo: SciPy 10
from scipy import linalg
import numpy as np

# Crea la matriz
mat = np.array([[1, 4], [9, 3]])

# Realiza la descomposición generando una tupla
P, L, U = linalg.lu(mat)

# Muestra los resultados de manera más clara
print("Matriz A:")
print(mat)
print("\nMatriz L:")
print(L)
print("\nMatriz U:")
print(U)
print("\nMatriz P:")
print(P)
```

## Autovalores y Autovectores

Se pueden encontrar los autovalores y los autovectores de una matriz con el método <span style="color:magenta">eig()</span>.

```python
# Ejemplo: SciPy 11
from scipy import linalg
import numpy as np

# Crea la matriz
mat = np.array([[3, 2, 7], [1, 3, 0], [4, 6, 8]])

# Realiza la descomposición generando una tupla
vals, vects = linalg.eig(mat)

# Muestra los resultados
print('Autovalores: ', vals)
print('Autovectores:\n', vects)
```

## Integral Simple

la función <span style="color:magenta">quad()</span> permite calcular la integral. 

en este caso la función <span style="color:magenta">f</span> es la que se necesita integrar.

```python
# Ejemplo: SciPy 12
from scipy import integrate
import numpy as np

# define la función a integrar
f = lambda x: x**3

# Calcula la integral en el intervalo 0, 1
integ = integrate.quad(f, 0, 1)

# Muestra el resultado (valor integral, error)
print('El valor de la integral es ', integ)
```

## Integral Doble

se utiliza <span style="color:magenta">dblquad(func, a, b, gfun, hfun)</span>.

```python
# Ejemplo: SciPy 13
from scipy import integrate
import numpy as np

# define las funciones
func = lambda x, y : x**y
gfunc = lambda x : x
hfunc = lambda y : y**2

# Calcula la integral en el intervalo 0, 1
integ = integrate.dblquad(func, 0, 0.5, gfunc, hfunc)

# Muestra el resultado (valor integral, error)
print('El valor de la integral doble es ', integ)
```

# Transformada de Fourier

La transformada de Fourier es la forma de representar una señal en forma de suma de sus componentes periódicos. Básicamente, convierte la señal del dominio temporal o espacial al dominio de la frecuencia. Es posible recuperar la señal original a partir de la señal transformada.

Podemos utilizar el método <span style="color:magenta">fft()</span> para convertir al dominio de la frecuencia e <span style="color:magenta">ifft()</span> para volver al dominio temporal o espacial.

El método <span style="color:magenta">fftfreq()</span> genera las muestras en el dominio de la frecuencia.

```python
# Ejemplo: SciPy 14
from matplotlib import pyplot as plt
import numpy as np 
from scipy.fftpack import fft, ifft,fftfreq

# Frecuencia en Hertz
f  = 10

# Frecuencia de Muestreo
sr = 100

# Senial Sinusoidal
t = np.linspace(0, 2, 2 * sr, endpoint = False )
senial= np.sin(f * 2 * np.pi * t)

# Calcula la fft
senial_fft = fft(senial)

# Calcula la inversa de la fft
senial_ifft = ifft(senial_fft)
```

```python
# Dibuja las seniales
fig=plt.figure()

plt.plot(t, senial)
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.title("Señal Original")

fig=plt.figure()

frecuencia = fftfreq(len(senial_fft)) * sr
plt.stem(frecuencia, np.abs(senial_fft))

plt.xlabel('Frecuencia')
plt.ylabel('Magnitud')
plt.title("Espectro de Frecuencia de la señal")

fig=plt.figure()
plt.plot(t, senial_ifft)
plt.xlabel('Tiempo')
plt.ylabel('Amplitud')
plt.title("Señal Sinusoidal Convertida")
```

## Procesamiento de una Imagen

El procesamiento de una imagen utiliza el paquete denominado *ndimage* para ejecutar diferentes operaciones sobre las imagenes.

Estas incluyen:

+ Manipulación básica de imagenes como recorter, rotar, etc.

+ Operaciones de filtrado de imagenes.

+ Segmentación de imagenes.

+ Clasificación de imagenes.

+ Extracción de Características.

+ Procesos de registración.

+ Etc.

```python
# Ejemplo: SciPy 15
from scipy import datasets, ndimage
import matplotlib.pyplot as plt

# Lee la imagen
img = plt.imread('./Figuras/con.jpg')

# Rota la imagen
rotate_img = ndimage.rotate(img, 90)

# Aplica un filtro Gausiano
imagen_filtro_gausiano = ndimage.gaussian_filter(img, sigma=5)
  
# Aplica un filtro de Mediana
imagen_filtro_mediana = ndimage.median_filter(img, size=9)
  
# Aplica un filtro de Sobel
imagen_filtro_sobel = ndimage.sobel(img)
```

```python
# Muestra la imagen original y la imagen rotada
fig=plt.figure()
plt.imshow(img)
fig=plt.figure()
plt.imshow(rotate_img)

figure, axis = plt.subplots(2, 2)
  
# original image
axis[0, 0].imshow(img)
axis[0, 0].set_title("Imagen Original")
axis[0, 0].axis('off')
  
# gaussian filter
axis[0, 1].imshow(imagen_filtro_gausiano)
axis[0, 1].set_title("Imagen Filtrada: Gaussiano")
axis[0, 1].axis('off')
  
# median filter
axis[1, 0].imshow(imagen_filtro_mediana)
axis[1, 0].set_title("Imagen Filtrada: Mediana")
axis[1, 0].axis('off')
  
# sobel filter
axis[1, 1].imshow(imagen_filtro_sobel)
axis[1, 1].set_title("Imagen Filtrada: Sobel")
axis[1, 1].axis('off')

plt.show()
```
