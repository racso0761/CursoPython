+ Un <span style="color:magenta">archivo</span> es un conjunto de bytes almacenados en algún dispositivo. El <span style="color:magenta">sistema de archivos</span> es la estructura lógica que alberga los archivos y está jerarquizado a través de <span style="color:magenta">directorios</span> (o carpetas). 

+ Cada archivo se identifica unívocamente a través de una <span style="color:yellow">ruta</span> que permite acceder a él y un <span style="color:yellow">nombre</span>.

## <font color=#38bdf8>Tipos de Archivos</font>

Python puede gestionar 2 tipos de archivos:

+ Archivos de <span style="color:yellow">texto</span>, normalmente texto Unicode, pero puede especificarse el formato que se desea. Modo: <span style="color:limegreen">'t'</span>.

+ Archivos de datos puros, también denominado <span style="color:yellow">binario</span>. Modo: <span style="color:limegreen">'b'</span>.

## <font color=#38bdf8>Operaciones Básicas con Archivos</font>

Al trabajar con archivos siempre se debe operar con las siguientes operaciones básicas:

+ <span style="color:yellow">Apertura</span> de un archivo.

+ <span style="color:yellow">Lectura / Escritura</span> de un archivo.

+ <span style="color:yellow">Cierre</span> del archivo.

## <font color=#38bdf8>Apertura de un Archivo</font>

Python posee la función <span style="color:limegreen">open()</span> para <span style="color:yellow">abrir</span> un archivo. Esta apertura se puede realizar en 3 modos básicos distintos:

+ Apertura para la lectura del contenido de un archivo existente.

+ Apertura para la escritura del contenido en un archivo nuevo.

+ Apertura para agregar información al contenido de un archivo existente.

Sintaxis básica:

<span style="color:limegreen">f = open(arg_1, arg_2)</span>

<span style="color:limegreen">f = open(arg_1, arg_2, encoding=arg_3)</span>

donde:
+ <span style="color:limegreen">f</span>: valor de retorno de la función <span style="color:limegreen">open()</span> y se le denomina <span style="color:yellow">gestor de archivo</span>.

+ <span style="color:limegreen">argumento1</span>: es un string con la <span style="color:yellow">ruta</span> y el <span style="color:yellow">nombre</span> del archivo que se desea abrir. Es conveniente siempre usar un string del tipo <span style="color:yellow">raw</span>. [Mas información sobre rutas relativas y absolutas](https://denovatoanovato.net/rutas-relativas-y-rutas-absolutas/).

+ <span style="color:limegreen">argumento2</span>: es un string que especifica el <span style="color:yellow">modo</span> de apertura del archivo:

*Modo* | *Descripción*
---|---
<span style="color:limegreen">'r'</span> | Modo de solo lectura (no se puede escribir en el archivo). Si el archivo no existe genera una excepción.
<span style="color:limegreen">'w'</span> | Modo de escritura, si el archivo existe se borra y se sobrescribe, sino es creado.
<span style="color:limegreen">'a'</span> | Modo de agregar: permite escribir información en un archivo existente, sino existe lo crea.
<span style="color:limegreen">'r+'</span> | Modo de lectura y de escritura.
<span style="color:limegreen">'w+'</span> | Modo de escritura y lectura. Al abrir el archivo se borra si existe, sino se crea.
<span style="color:limegreen">'a+'</span> | Modo de lectura y para agregar. Si el archivo no existe se crea
<span style="color:limegreen">'x'</span> | Se comporta como el modo <span style="color:limegreen">'w'</span>, pero si el archivo existe dispara una excepción. También se lo conoce como modo exclusivo.

+ A cada modo se le puede agregar la letra <span style="color:limegreen">'t'</span> o <span style="color:limegreen">'b'</span> (texto o binario) para especificar el tipo de archivo con el que se desea trabajar.

+ Si no se especifica Python asume que se desea trabajar con un archivo de <span style="color:magenta">texto</span>.

+ <span style="color:limegreen">encoding</span>: en este argumento se debe especificar la codificación que se pretende en el archivo de texto. Si no se especifica asume el valor dependiendo de la plataforma que se este utilizando que normalmente es <span style="color:magenta">"utf-8"</span>, pero en windows normalmente es <span style="color:magenta">ASCII estándar</span>.

```python
# Ejemplo: Archivos 1
# Ejemplo: Apertura de un archivo
f = open('./Archivos/mio.txt', 'r')
```

+ La variable <span style="color:limegreen">f</span> (el gestor del archivo) es un objeto-iterable que almacena información relativa al archivo con el que se está trabajando.

## <font color=#38bdf8>Cierre de un Archivo</font>

+ Toda vez que se abre un archivo, el programador tiene la responsabilidad de cerrarlo cuando deje de utilizarlo.

+ Si no se cierra un archivo se corre el riesgo de perder información, corromper e inclusive perder toda la información que contiene un archivo.

+ El cierre de un archivo de lleva a cabo con el método <span style="color:limegreen">close()</span>.

```python
# Ejemplo: Archivos 2
# Ejemplo: Apertura de un Archivo con gestion de excepción
try:
    f = open(r'./Archivos/mio.txt', 'r')
except:
    print('Archivo Inexistente')

print(type(f))

# Cierre del archivo
f.close()
```

## <font color=#38bdf8>Lectura de un Archivo de Texto</font>

+ Se debe usar el modo lectura (<span style="color:limegreen">'r'</span>) o lectura y escritura (<span style="color:limegreen">'r+'</span>).

+ Si el archivo no existe se producirá una <span style="color:yellow">Excepción</span> (un error).

+ Si no se producen errores entonces existen las siguientes formas posibles de leer información desde un archivo:

*Método* | *Descripción*
---|---
<span style="color:limegreen">read([n])</span> | Devuelve todo el contenido del archivo. En modo <span style="color:magenta">texto</span> como un <span style="color:limegreen">str</span>. Si se especifica <span style="color:limegreen">n</span> lee al menos n bytes.
<span style="color:limegreen">readlines([tamaño])</span> | Lee todas las líneas del archivo y las devuelve como una <span style="color:magenta">lista de strings</span>. El argumento <span style="color:yellow">tamaño</span> opcionalmente especifica el número aproximado de caracteres a leer antes de detenerse.
<span style="color:limegreen">readline([n])</span> | Lee una única línea de hasta <span style="color:limegreen">n</span> caracteres como un string. Si no se especifica <span style="color:limegreen">n</span> devuelve la línea completa. También desplaza el <span style="color:yellow">gestor del archivo</span> a la próxima línea.

+ Una vez completado el proceso de lectura de información se debe cerrar el archivo (<span style="color:limegreen">close()</span>).

+ Tener en cuenta que el modo de lectura puede ser sobre archivos de texto o binarios.

## <font color=#38bdf8>Escritura de Información en un Archivo de Texto</font>

+ Para poder escribir información en un archivo, se debe abrir en modo escritura (<span style="color:limegreen">'w'</span>) con la función <span style="color:limegreen">open()</span>.

    + Si el archivo no existe, se creará.

    + Si el archivo existe el contenido se borrará y se perderá.
    
+ Para escribir información en un archivo, Python posee dos métodos:

*Método* | *Descripción*
---|---
<span style="color:limegreen">write(s)</span> | Escribe el string <span style="color:limegreen">s</span>. 
<span style="color:limegreen">writelines(lineas)</span> | Escribe todos los strings en la secuencia <span style="color:limegreen">lineas</span>.

+ Luego de operar con el archivo este debe cerrarse.

## <font color=#38bdf8>Agregar Información a un Archivo de Texto</font>

+ Se debe abrir el archivo en modo agregar (modo <span style="color:limegreen">'a'</span>).
+ Una vez abierto el archivo se utilizan las funciones <span style="color:limegreen">write()</span> o <span style="color:limegreen">writelines()</span> para agregar la información al archivo.
+ Una vez finalizada la operación el archivo debe cerrarse. <span style="color:limegreen">close()</span>

+ El modo de apertura de <span style="color:magenta">lectura y escritura</span> (<span style="color:limegreen">'r+'</span>) permite modificar información existente en un archivo.

```python
# Ejemplo: Archivos 10
# Ejemplo: Agregado de Información en un Archivo
# Códigos Iata Aeropuertos Argentina
#iata_arg = ('AEP\n', 'EZE\n', 'UAQ\n', 'ROS\n', 'LUQ\n', 'SLA\n', 'MDZ\n', 'COR\n')

nuevo_aerp = 'CNQ\n' # Corrientes
```

```python
# Tener en cuenta los caracteres \n
print(f.readlines())
f.close()
```

## <font color=#38bdf8>Gestores de Contexto</font>

Los gestores de contexto o <span style="color:magenta">context managers</span> permiten asignar o liberar recursos del sistema de forma directa. El uso más común es el operador de Python <span style="color:limegreen">with</span>. Imagínese que se tienen dos operaciones relacionadas que se desean ejecutar con un determinado código de por medio. Los gestores de contexto permiten manejar esta situación de manera sencilla.

Python ofrece la posibilidad de utilizar los gestores de contexto como una solución para establecer reglas de entrada y salida a un determinado bloque de código.

Por ejemplo:

```python
# Ejemplo: Archivos 12
# Uso de Contexto
with open(r'./Archivos/prueba.txt', 'w') as f:
    f.write('Hola!')
```

En el ejemplo anterior se abre el archivo, se escriben datos y se cierra automáticamente. Si se produce un error al intentar abrir el archivo o al intentar escribir contenido en él, siempre el archivo se cierra al final. El siguiente código sería el equivalente con manejo de excepciones.

```python
# Ejemplo: Archivos 13
# Uso de Contexto
f = open(r'./Archivos/prueba.txt', 'w')
try:
    f.write('Hola!')
finally:
    f.close()
```

Al comparar los ejemplos anteriores se puede observar que gran cantidad de código repetido es eliminado al usar el gestor de contexto <span style="color:limegreen">with</span>. La principal ventaja del uso de <span style="color:limegreen">with</span> es que se asegura que el archivo se cierra, sin importar lo que hay en el bloque de código.

En general, los usos más comunes de los gestores de contexto son bloquear y liberar recursos, como en el ejemplo que anterior.

También es posible implementar un gestor de contexto propio, que se abordará cuando se estudie la Programación Orientada a Objetos.

Otros ejemplo útiles:

```python
# Ejemplo: Archivos 14
# Ejemplo: Uso de Contextos

with open(r'./Archivos/mio.txt', 'r') as f:
    for linea in f:
        temp_min, temp_max = linea.strip().split()
        print(temp_min, temp_max)

# El archivo se cierra automaticamente cuando se termina el bloque
```

## <font color=#38bdf8>Otros Métodos útilies para la Gestión de Archivos</font>

+ Python posee además un conjunto de métodos que permiten la gestion de archivos. Algunas de ellas son (<span style="color:limegreen">f</span> es el gestor de archivos):

*Método* | *Descripción*
---|---
<span style="color:limegreen">f.tell()</span> | Devuelve la posición actual del <span style="color:magenta">gestor del archivo</span>.
<span style="color:limegreen">f.seek(offset, [,whence])</span> | Coloca el gestor del archivo en una nueva posición.
<span style="color:limegreen">f.flush()</span> | Limpia el buffer de salida.
<span style="color:limegreen">f.fileno()</span> | Devuelve un descriptor de archivo (entero).
<span style="color:limegreen">f.\_\_next\_\_()</span> | Devuelve la próxima línea o genera una excepción del tipo <span style="color:limegreen">StopIteration</span>.

+ El argumento <span style="color:yellow">whence</span> del método <span style="color:limegreen">seek()</span> establece desde donde se mide el <span style="color:yellow">offset</span> (desplazamineto). Puede ser <span style="color:yellow">0</span>: desde el comienzo del archivo, <span style="color:yellow">1</span>: desde la posición actual, y <span style="color:yellow">2</span>: desde el final del archivo.

+ Además los archivos disponen de <span style="color:magenta">atributos</span> de solo lectura que en ocaciones pueden resultar de utilidad (<span style="color:limegreen">f</span> es el gestor de archivos):

*Atributo* | *Descripción*
---|---
<span style="color:limegreen">f.closed</span> | Valor booleano que indica el estado del archivo. <span style="color:limegreen">True</span>: archivo abierto. <span style="color:limegreen">False</span>: archivo cerrado.
<span style="color:limegreen">f.mode</span> | Indica el modo de Entrada/Salida del archivo.
<span style="color:limegreen">f.name</span> | Nombre del archivo.
<span style="color:limegreen">f.encoding</span> | Un string que indica la codificación del archivo (por ejemplo <span style="color:magenta">'utf-8'</span>).

```python
# Ejemplo: Archivos 17
f = open(r'./Archivos/mio.txt', 'r')
```

## <font color=#38bdf8>Archivos Binarios</font>

Python por defecto asume que los *archivos son de texto* y por lo tanto para trabajar con *archivos binarios* se debe procesar la información.

+ Se debe abrir el archivo para la operación que se desee realizar agregando una <span style="color:limegreen">'b'</span> indicando que el archivo es binario.

+ Como el archivo es binario, Python necesita que los datos estén estructurados como un <span style="color:magenta">stream de bytes</span>.

### <font color=#38bdf8>Escritura de Información en un Archivo Binario</font>

+ La forma de abrir un archivo para una operación (lectura, escritura o para agregar) en modo binario es igual que para el caso de los archivos de texto, simplemente agregando una letra <span style="color:limegreen">'b'</span> al modo de apertura.

+ Ahora para poder escribir datos en un archivo, en modo binario, se deben colocar los mismos en el formato adecuado, lo que significa transformarlos en un stream de bytes. Esto se puede realizar utilizando algunos métodos o funciones propios de Python:

    + <span style="color:limegreen">encode()</span>: (método) Devuelve una versión codificada del string dado.

    + <span style="color:limegreen">bytearray()</span>: (función) Devuelve un arreglo de bytes como un objeto <span style="color:limegreen">bytearray</span>.

+ Luego de preparar los datos se puede usar el método <span style="color:limegreen">write()</span> para escribirlos en el archivo.

### <font color=#38bdf8>Lectura de Información de un Archivo Binario</font>

+ La forma de abrir un archivo para una operación de lectura en modo binario es igual que para el caso de los archivos de texto, simplemente agregando una letra <span style="color:limegreen">'b'</span> al modo de apertura.

+ Ahora para poder leer datos desde un archivo, en modo binario, se debe utilizar el método <span style="color:limegreen">read()</span>. Luego se deben convertir los datos a un formato adecuado de acuerdo a las necesidades, por ejemplo utilizando el método <span style="color:limegreen">decode()</span>.

    + <span style="color:limegreen">decode()</span>: Devuelve una versión decodificada del objeto sobre el que se utiliza.

```python
# Ejemplo: Archivos 19
# Método encode()
titulo = 'Programación Python'
print(titulo.encode('utf-8'))
```

```python
# Ejemplo: Archivo 20
# Método bytearray()
valor = [2, 3, 5, 7, 8, 4, 6, 200]
valor_byte = bytearray(valor)
print(type(valor_byte))
```

```python
# Ejemplo: Archivos 22
# Escritura de Archivos Binarios
texto_contenido = ['python\n', 'java\n', 'c++\n', 'basic\n', 'pascal\n', 'ruby\n']
nombre_archivo = r'Archivos/utextbinario_1.bin'
archivo_binario = open(nombre_archivo, 'wb')
for item in texto_contenido:
    texto_bytes = item.encode('utf-8')
    archivo_binario.write(texto_bytes)
archivo_binario.close()
print('Finalizado...')
```

### <font color=#38bdf8>Paquete Struct</font>

+ El paquete <span style="color:yellow">struct</span> ejecuta una conversión entre valores Python y estructuras del lenguaje C representada como objetos <span style="color:magenta">bytes</span>.

+ El mecanismo utilizado para especificar la disposición de empaquetamiento o desempaquetamiento de los datos es a través de los string de formato.

Sintaxis:

<span style="color:limegreen">struct.pack(string_de_formato, v1, v2, ..., vn)</span>

+ Este método devuelve un string formateado en bytes con el formato especificado.

+ El string de formato que se debe utilizar posee los siguientes caracteres:

*Formato* | *Tipo C* | *Tipo Python* | *Tamaño*
---|---|---|---
<span style="color:yellow">c</span> | <span style="color:limegreen">char</span> | <span style="color:limegreen">byte</span> | 1
<span style="color:yellow">b</span> | <span style="color:limegreen">signed char</span> | <span style="color:limegreen">integer</span> | 1
<span style="color:yellow">B</span> | <span style="color:limegreen">unsigned char</span> | <span style="color:limegreen">integer</span> | 1
<span style="color:yellow">?</span> | <span style="color:limegreen">_Bool</span> | <span style="color:limegreen">bool</span> | 1
<span style="color:yellow">h</span> | <span style="color:limegreen">short</span> | <span style="color:limegreen">integer</span> | 2
<span style="color:yellow">H</span> | <span style="color:limegreen">unsigned short</span> | <span style="color:limegreen">integer</span> | 2
<span style="color:yellow">i</span> | <span style="color:limegreen">int</span> | <span style="color:limegreen">integer</span> | 4
<span style="color:yellow">I</span> | <span style="color:limegreen">unsigned int</span> | <span style="color:limegreen">integer</span> | 4
<span style="color:yellow">l</span> | <span style="color:limegreen">long</span> | <span style="color:limegreen">integer</span> | 4
<span style="color:yellow">L</span> | <span style="color:limegreen">unsigned long</span> | <span style="color:limegreen">integer</span> | 4
<span style="color:yellow">q</span> | <span style="color:limegreen">long long</span> | <span style="color:limegreen">integer</span> | 8
<span style="color:yellow">Q</span> | <span style="color:limegreen">unsigned long long</span> | <span style="color:limegreen">integer</span> | 8
<span style="color:yellow">n</span> | <span style="color:limegreen">ssize_t</span> | <span style="color:limegreen">integer</span>
<span style="color:yellow">N</span> | <span style="color:limegreen">ssize_t</span> | <span style="color:limegreen">integer</span>
<span style="color:yellow">e</span> | <span style="color:limegreen">---</span>| <span style="color:limegreen">float</span> | 2
<span style="color:yellow">f</span> | <span style="color:limegreen">float</span> | <span style="color:limegreen">float</span> | 4
<span style="color:yellow">d</span> | <span style="color:limegreen">double</span> | <span style="color:limegreen">float</span> | 8
<span style="color:yellow">s</span> | <span style="color:limegreen">char[ ]</span> | <span style="color:limegreen">bytes</span>
<span style="color:yellow">p</span> | <span style="color:limegreen">char[ ]</span> | <span style="color:limegreen">bytes</span>
<span style="color:yellow">P</span> | <span style="color:limegreen">void *</span> | <span style="color:limegreen">integer</span>

+ El método <span style="color:limegreen">struct.unpack()</span> siempre devuelve una <span style="color:yellow">tupla</span>.

Sintaxis:

<span style="color:limegreen">struct.unpack(string_de_formato, string)</span>

+ El método <span style="color:limegreen">struct.calcsize(fmt)</span> permite calcular el tamaño de la estructura.

[+info](https://docs.python.org/es/3/library/struct.html)

```python
# Ejemplo: Archivos 21
# Ejemplo de Archivos Binarios
import struct as st
texto = b'foo'
datos_binarios = st.pack('3sif', texto, 42, 3.14)
```

```python
f = open('./Archivos/b_arch.bdd', 'wb')
```

```python
f.write(datos_binarios)
f.close()
```

```python
with open('./Archivos/b_arch.bdd', 'rb') as ff:
    datos = ff.read()
```

```python
a, b, c = st.unpack('3s i f', datos)
print(f'Valores leidos: {a.decode("utf-8")}, {b}, {c:.2f}')
```
