<h2 style="color:#38bdf8;color:#38bdf8;text-decoration:none;padding:20px;">1️⃣ ¿Qué es Python y cuáles son sus características?</h2>
</div>
Python es un lenguaje de programación de los más importantes en la actualidad. Con múltiples usos, se trata de un lenguaje que tiene características que lo hace uno de los lenguajes preferidos por desarrolladores de distintos ámbitos.

Las principales características de Python son:

+ Lenguaje de programación interpretado, ejecuta el código línea a línea.
+ Fácil de usar gracias a una sintaxis sencilla.
+ Alto nivel, es lo más cercano que hay, en programación, al lenguaje natural.
+ Orientado a objetos, considerando todo como un objeto, permite y facilita la reutilización de segmentos de código en distintas aplicaciones.

<h2 style="color:#38bdf8;text-decoration:none;padding:20px;">2️⃣ ¿Por qué Python?</h2>
</div>
Hay motivos variados para decidir por qué aprender Python. Algunas de las cuales son:

## <font color=#38bdf8>Oportunidades laborales y carrera profesional</font>

De entre todos los motivos de por qué estudiar Python, el de las oportunidades laborales es, quizás, el más importante. Ahora mismo, Python es el lenguaje de programación con mayor crecimiento. Cada vez tiene más usos en distintos sectores, adquiriendo especial importancia en el campo de la Inteligencia Artificial, que ya tiene un gran impacto en todas las disciplinas y se mantendrá vigente los próximos años.

Por otra parte, son muchas las empresas que confían en Python como lenguaje de programación principal para el desarrollo de sus aplicaciones. De esta manera, aprender Python da una ventaja competitiva sobre otros candidatos que no tengan conocimientos en este Lenguaje de programación.

## <font color=#38bdf8>Campos de aplicación</font>

Python abre muchísimas puertas laborales porque se trata de un lenguaje de programación de referencia en distintos campos. Algunos de los ámbitos en los que se usa Python son:

+ Data Mining
+ Data Analytics
+ Big Data
+ Data Science
+ Visualización de Datos
+ Inteligencia Artificial, Machine learning y Deep Learning
+ Desarrollo de Aplicaciones
+ Diseño de Backends en Webs y Apps
+ Desarrollo de Videojuegos

Además en la actualidad existen módulos desarrollados para Python casi para cualquier disciplina lo que reduce considerablemente los tiempos de gestación de nuevas aplicaciones y la solución a problemas de alta comlpejidad.

## <font color=#38bdf8>Fácil de usar y aprender</font>

Una de las mejores razones de por qué aprender Python es que se trata de un lenguaje de programación relativamente fácil de usar y de aprender (sobre todo para los no programadores) ya que, Python emplea una sintaxis sencilla que usa elementos del inglés natural. Esto hace que la escritura, la lectura y el aprendizaje se simplifiquen. Esto se conoce como un Lenguaje cercano al Lenguaje Natural.

Por otra parte, al eliminar toda la sintaxis y las líneas de código innecesarias, Python se ha convertido en un lenguaje muy accesible para aprender a programar y lograr cosas en poco tiempo. Además, se trata de un lenguaje de programación interpretado, lo que quiere decir que podemos ejecutar cada línea de código una vez escrita, sin tener que compilar todo como sucede con lenguajes como C o C++.

## <font color=#38bdf8>Comunidad activa</font>

La programación puede ser, en ocasiones, un trabajo solitario. No obstante, son muchos los lenguajes que cuentan con una comunidad activa y que brindan soporte a los que recién se inician. En este sentido, la comunidad de Python es una de las que más destaca.

Así, en Internet se puede encontrar una comunidad tremendamente activa, de todos los niveles, que comparten recursos y resuelven dudas. Además, en **Github**, se pueden encontrar secciones específicas dedicadas a Python.

Tener en cuenta que, Python cuenta con más de 56 millones de usuarios y 2 millones de repositorios alojados en la plataforma **Github**.

## <font color=#38bdf8>Python es un Lenguaje Versátil</font>

Python es uno de los lenguajes de programación más versátiles que existen. Pues se trata de un código que cuenta con varias funcionalidades distintas. Así, son muchas las capacidades que tiene Python, y, esto le lleva a tener multitud de aplicaciones y una gran variedad de campos profesionales en los que se destaca.

Además, Python cuenta con una amplia selección de bibliotecas que contienen código prescrito. Este tipo de recursos se puede aprovechar para crear código propio y desarrollar con él aplicaciones de todo tipo.

Al mismo tiempo, la versatilidad de Python permite ejecutar código de otros lenguajes de programación como Java o C. Esto hace que tenga funcionalidades cruzadas, lo que amplía todavía más sus posibles usos.

## <font color=#38bdf8>Otras razones por las que se debe aprender Python</font>

Python presenta una serie de ventajas que lo hacen muy atractivo, tanto para un uso profesional como para el aprendizaje de la programación. Entre las más interesantes desde el punto de vista didáctico se pueden mencionar: 

+ Python es un lenguaje expresivo, es decir, los programas Python son muy compactos: un programa Python suele ser bastante más corto que su equivalente en lenguajes como C. (Python llega a ser considerado por muchos un lenguaje de programación de muy alto nivel).
+ Python es legible. La sintaxis de Python es muy elegante y permite la escritura de programas cuya lectura resulta más sencilla que si se utilizáran otros lenguajes de programación.
+ Python ofrece un entorno interactivo que facilita la realización de pruebas y ayuda a despejar dudas acerca de ciertas características del lenguaje.
+ El intérprete de Python detecta muchos de los errores de programación que escapan al control de los compiladores y proporciona información muy rica para su detección y corrección.
+ Python puede usarse como lenguaje imperativo procedimental o como lenguaje orientado a objetos.
+ Posee un rico juego de estructuras de datos que se pueden manipular de modo sencillo.
