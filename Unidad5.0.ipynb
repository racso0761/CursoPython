### <font color=#38bdf8>Operaciones con Listas</font>

<span style="color:yellow">Indexación</span>: Al igual que con los strings se puede utilizar el operador de indexación <span style="color:limegreen">[ ]</span> para acceder a un elemento de una lista.

```python
# Ejemplo: Listas 2
lenguajes = ['C++', 'Pascal', 'Java', 'Python']

print(f'Elemento: 2, valor: {lenguajes[2]}')
print(f'Elemento: -1, valor: {lenguajes[-1]}')
print(f'Elemento: 20, valor: {lenguajes[20]}')
```

<span style="color:yellow">Slicing</span>: el operador de slicing también se puede utilizar con listas, de igual manera que lo visto en strings.

```python
# Ejemplo: Listas 3
lenguajes = ['C++', 'Pascal', 'Java', 'Python', 'Julia']

print(f'Sublista [1:4]: {lenguajes[1:4]}')
print(f'Sublista [0:4]: {lenguajes[0:4]}')
print(f'Sublista [4]: {lenguajes[4]}')
```

```python
print(f'Sublista [0:4:2]: {lenguajes[0:4:2]}')
print(f'Sublista [-1:-3:-1]: {lenguajes[-1:-3:-1]}')
print(f'Sublista [::-1]: {lenguajes[::-1]}') # Equivale a invertir la lista
print(f'Sublista [10:]: {lenguajes[10:]}')
```

<span style="color:yellow">Inversión de una Lista</span>:

+ Conservando la lista Original: 

    + Con <span style="color:yellow">slicing</span> y <span style="color:yellow">paso</span> negativo.

    + Con la función <span style="color:limegreen">reversed()</span>    

+ Modificando la lista Original:

    + Con el método <span style="color:limegreen">reverse()</span>.


```python
# Ejemplo: Listas 4
lenguajes = ['C++', 'Pascal', 'Java', 'Python', 'Julia']
# Alternativa 1
leng_1 = lenguajes[::-1]
print(f'Lista Original: {lenguajes}\nLista invertida: {leng_1}')
```

```python
# Alternativa 2 - reversed devuelve un iterable que se debe convertir en lista
leng_2 = list(reversed(lenguajes))
print(f'Lista Original: {lenguajes}\nLista invertida: {leng_2}')
```

```python
# Alternativa 3 - Se modifica la lista original
lenguajes.reverse()
print(f'Lista Original: {lenguajes}')
```

### <font color=#38bdf8>Ordenar una Lista</font>

Existen al menos dos formas para ordenar una lista:

+ *Conservando la lista original*: usando la función <span style="color:limegreen">sorted(lista, reverse=False)</span>.

+ *Modificando la lista original*: usando el método <span style="color:limegreen">sort(key, reverse=False)</span>.

+ Ambas formas admiten un parámetro booleano <span style="color:yellow">reverse</span> para lograr un orden de ordenación inverso de la lista.

<span style="color:yellow">reverse = False</span>: valor por defecto ordena de manera ascendente.  
<span style="color:yellow">key</span>: en este argumento se debe pasar una función con el criterio de la ordenación. Por defecto es <span style="color:limegreen">None</span>.

[+info método sort()](https://www.freecodecamp.org/espanol/news/python-ordenar-lista-con-sort-ascendente-y-descendente-explicado-con-ejemplos/)

```python
# Ejemplo: Listas 13
lenguajes = ['C++', 'Java', 'Pascal', 'Java', 'Python', 'Julia', 'Java']
print(f'Lista Original: {lenguajes}')
```

```python
orden = sorted(lenguajes)
print(f'Lista Original: {lenguajes}\nLista Ordenada: {orden}')
```

```python
lenguajes = ['C++', 'Java', 'Pascal', 'C', 'Java', 'Python', 'Julia', 'Java']
lenguajes.sort()
print(f'Lista Original: {lenguajes}')
```

```python
lenguajes = ['C++', 'Java', 'Pascal', 'T', 'Java', 'Python', 'Julia', 'Java']
lenguajes.sort(reverse = True)
print(f'Lista inversa: {lenguajes}')
```

```python
lenguajes.sort(key = lambda x: len(x))
print(f'Lista ordenada por longitud: {lenguajes}')
```

```python
lenguajes = ['C++', 'Java', 'Pascal', 'T', 'Java', 'Python', 'Julia', 'Java']
otra = sorted(lenguajes, key = lambda x: len(x))
print(f'otra: {otra}')
```

### <font color=#38bdf8>Iteración sobre una Lista</font>

+ Se utiliza la sentencia de control <span style="color:limegreen">for</span> al igual que se usa con strings.

+ Si se desea también conocer el índice de cada elemento se utiliza la función <span style="color:limegreen">enumerate()</span>.

```python
# Ejemplo: Listas 14
lista_compras = ['Azucar', 'Aceite', 'Arroz', 'Fideos']
detalles = ['Primera Calidad', 'de Oliva Virgen', 'Yamani', 'de Harina Integral']

print(f'Cantidad en lista de Compras: {len(lista_compras)}')
```

```python

for i, producto in enumerate(lista_compras):
    print(f'Indice: {i}, Producto: {producto}')
print('\n')
```

```python
for i, producto in enumerate(lista_compras, start = 8):
    print(f'Indice: {i}, Producto: {producto}')
print('\n')
```

```python
print(lista_compras[0])
print(list(enumerate(lista_compras)))
print('\n')
```

```python
for producto, detalle in zip(lista_compras, detalles):
    print(f'Producto: {producto}  \tDetalle: {detalle}')
```

### <font color=#38bdf8>Veracidad Múltiple</font>

+ La función <span style="color:limegreen">all()</span> permite evaluar si se cumplen todas las condiciones sobre un <span style="color:yellow">objeto-iterable</span>.

+ La función <span style="color:limegreen">any()</span> permite evaluar si se cumple alguna condición sobre un <span style="color:yellow">objeto-iterable</span>.

```python
# Ejemplo con all()
condiciones = [5 > 2, 10 == 10, "hola" != ""]
resultado = all(condiciones)
print(resultado)  # True, porque todas las condiciones son verdaderas
```

```python
# Ejemplo: Listas 20 (Tradicional)
palabra = 'Python'

if len(palabra) > 4 and palabra.startswith('P') and palabra.count('y') >= 1:
    print('Buena Palabra!')
else:
    print('No Tan Buena')
```

```python
# Ejemplo: Listas 21
# Uso de veracidad multiple - all()
palabra = 'Java'
es_buena_palabra = [len(palabra) > 4, palabra.startswith("P"), palabra.count('y') >= 1]

if all(es_buena_palabra):
    print('Buena Palabra')
else:
    print('No Tan Buena')
```

```python
# Ejemplo con any()
condiciones = [5 < 2, 10 == 10, ""]
resultado = any(condiciones)
print(resultado)  # True, porque al menos una condición es verdadera
```

```python
# Ejemplo: Listas 22
# Uso de veracidad múltiple - any()
palabra = 'yeah'

es_palabra_correcta = [len(palabra) > 4, palabra.startswith("P"), palabra.count('y') >= 1]

if any(es_palabra_correcta):
    print('Buena Palabra')
else:
    print('No Tan Buena')
```

```python
# Veracidad múltiple combinada
def validar_usuario(usuario, contraseña):
    # Verificar que no haya campos vacíos
    if not all([usuario, contraseña]):
        return "Faltan datos"

    # Verificar si cumple alguna condición especial
    if any([usuario == "admin", contraseña == "1234"]):
        return "Acceso especial"

    return "Acceso normal"

print(validar_usuario("admin", "pass"))   # Acceso especial
print(validar_usuario("", "pass"))        # Faltan datos
print(validar_usuario("user", "pass"))    # Acceso normal
```

### <font color=#38bdf8>List Comprehension (Listas por Comprensión)</font>

Teoría de Conjuntos: La definición de conjuntos se puede realizar de dos maneras diferentes:

+ *Definición por Extensión*: se enumeran todos sus elementos.

+ *Definición por Comprensión*: cuando se da una propiedad, que la cumplan todos los elementos del conjunto.

Por ejemplo:

+ Definición por Extensión: $A = \{a, e, i, o, u\}$.

+ Definición por Comprensión: $A = \{x / x \text{ es una vocal}\}$.

#### <font color=#38bdf8>Que sucede con las Listas?</font>

Con este mismo principio, las <span style="color:limegreen">Listas</span> también se pueden crear por <span style="color:yellow">comprensión</span>.

+ Permiten crear listas de manera compacta.

+ El buen uso permite crear código compacto de alta calidad.

![Listas por Comprensión](./Figuras/lista_comprension.png)

```python
# Ejemplo: Listas 23
# Alternativa clásica
valores = '32, 45, 11, 87, 20, 48' # String

valores_enteros = []
for valor in valores.split(','):
    valor_entero = int(valor)
    valores_enteros.append(valor_entero)

print(f'Lista de valores enteros: {valores_enteros}')
```

```python
# Ejemplo: Listas 24
# Alternativa con list comprehension
valores = '32, 45, 11, 87, 20, 48'

valores_enteros = [int(valor) for valor in valores.split(',')]
print(f'Lista de valores enteros: {valores_enteros}')
```

![Funcionamiento de la list comprehension](./Figuras/list_comprension2.png)

#### <font color=#38bdf8>Condicionales en Listas por Comprension</font>

+ Tambien es posible incluir una estructura de selección en la lista por comprension.

#### <font color=#38bdf8>Anidamiento en Listas por Comprensión</font>

+ También es posible usar anidamiento de itraciones en list comprehension.

```python
# Ejemplo: Listas 25
# Ejemplo: Condiciones en las list comprehension
valores = [32, 45, 11, 87, 20, 48]
valores_sup40 = [valor for valor in valores if valor > 40]
print(f'Nueva lista: {valores_sup40}')
```

```python
# Ejemplo: Listas 25.1
val = '32, 45, 11, 87, 20, 48'
val_s40 = [int(valor) for valor in val.split(',') if int(valor)> 40]
print(f'Nuevalista: {val_s40}')
```

```python
# Ejemplo: Listas 26
# Ejemplo: listas por comprension y strings
valores = '32,45,11,87,20,48'
valores_c4 = [int(v) for v in valores.split(',') if v.startswith('4')]
print(f'Nueva lista: {valores_c4}')
```

```python
# Ejemplo: Listas 27
# Ejemplo: Anidamiento en listas por comprension
valores = '32,45,11,87,20,48'
svalores = valores.split(',') # Genera una lista de strings
print(f'Lista de strings: {svalores}\n')

combinaciones = [f'{v1} x {v2}' for v1 in svalores for v2 in svalores]
print(f'Lista por comprension: {combinaciones}\n')

for v1 in range(len(combinaciones)):
    print(combinaciones[v1])
```

## Listas por Comprensión y Veracidad Múltiple

```python
# Verificar si una contraseña cumple todos los requisitos
password = "alMendra8"

requisitos = [
    len(password) >= 8,
    any(c.isdigit() for c in password),
    any(c.isupper() for c in password),
    any(c.islower() for c in password)
]

print(all(requisitos))  # True si cumple todo
```

```python
# Verifica si al menos un número es par
numeros = [1, 3, 7, 11, 13]

if any(n % 2 == 0 for n in numeros):
    print("Hay al menos un número par")
else:
    print("Todos son impares")
```

```python
# Validar que todos los números sean positivos
numeros = [3, 7, 2, 9]

if all(n > 0 for n in numeros):
    print("Todos son positivos")
else:
    print("Hay números no positivos")
```

```python
# Comprobar si una cadena contiene al menos una vocal
texto = "Python"

vocales = "aeiouAEIOU"

if any(v in texto for v in vocales):
    print("Tiene vocales")
else:
    print("No tiene vocales")
```

```python
# Verificar si todos los elementos son del mismo tipo
items = [5, 7.66, 12, 3]

if all(isinstance(x, int) for x in items):
    print("Todos son enteros")
else:
    print("Hay tipos distintos")
```

```python
# Validar que una matriz no tenga ceros
matriz = [
    [1, 2, 3],
    [4, 0, 6]
]

if all(all(n != 0 for n in fila) for fila in matriz):
    print("No hay ceros en la matriz")
else:
    print("Hay ceros en la matriz")
```

```python
# Saber si alguna palabra comienza con mayúscula
palabras = ["hola", "mundo", "Python"]

if any(p[0].isupper() for p in palabras):
    print("Hay palabras con mayúscula")
else:
    print("Todas comienzan en minúscula")
```

```python
# Detectar si alguna palabra contiene números
palabras = ["hola", "mundo2", "python"]

if any(any(c.isdigit() for c in p) for p in palabras):
    print("Alguna palabra tiene números")
else:
    print("Ninguna palabra contiene números")
```

```python
# Uso combinado: validar formularios y detectar errores
form = {
    "nombre": "Ana",
    "email": "ana@example.com",
    "direccion": ""
}

if all(form.values()):
    print("Formulario completo")
elif any(not v for v in form.values()):
    print("Hay campos vacíos")
```

### <span style="color:yellow">sys.argv</span>

Cuando se ejecuta un programa desde la línea de comandos, se puede acceder a la lista de argumentos a través de la lista <span style="color:limegreen">argv</span> que se encuentra en el módulo <span style="color:limegreen">sys</span>.

![Uso de la lista sys.argv](./Figuras/sysargv.png)

```python
# Ejemplo: Listas 28
# Ejemplo de sys.argv
# Ejecutar desde consola decimal2base.py 1024 2
# Tambien ejecutar desde consola decimal2base_v2.py
import sys

numero = int(sys.argv[1])
base = int(sys.argv[2])

match base:
    case 2:
        resultado = f'{numero:b}'
    case 8:
        resultado = f'{numero:o}'
    case 16:
        resultado = f'{numero:x}'
    case _:
        resultado = None
if resultado == None:
    print(f'Base: {base} no implementado!')
else:
    print(f'Resultado: {resultado}')
```

### <font color=#38bdf8>Prestando Atención a las Copias</font>

+ Las Listas son objetos <span style="color:yellow">mutables</span>.

+ Cuando se copian las Listas la modificación de una copia puede afectar a la lista original. 

```python
# Ejemplo: Listas 16
# Copia directa
original = [4, 3, 7, 1]

copia = original
print(copia)
```

```python
copia[0] = 100
print(f'Lista Original: {original}\nCopia: {copia}')
```

```python
# Ejemplo: Listas 17a
# Uso de copy(). Copia Soft
original = [4, 3, 7, 1]

copia = original.copy()
print(f'Lista Original: {original}\tCopia: {copia}')
```

```python
copia[0] = 100
print(f'Lista Original: {original}\nCopia: {copia}')
```

```python
# Ejemplo: Listas 17b
# Copia Soft con Otra lista
original = [4, 3, 7, [1, 2]]
print(f'Lista Original: {original}')
```

```python
copia = original.copy()
print(f'Copia: {copia}')
```

```python
copia[0] = 300

print(f'Copia: {copia}')
print(f'Lista Original: {original}')
```

```python
copia[3][0] = 100

print(f'Copia: {copia}')
print(f'Lista Original: {original}')
```

```python
# Ejemplo: Listas 18
# Uso de deepcopy()
import copy

original = [4, 3, 7, [1, 2]]
copia = copy.deepcopy(original)
```

```python
copia[3][0] = 100

print(f'Copia: {copia}')
print(f'Lista Original: {original}')
```

```python
# Ejemplo: Listas 19
# Uso de Slicing

original = [4, 3, 7, [1, 2]]
copia = original[:]
```

```python
copia[3][0] = 100
copia[0] = 300
```

```python
print(f'Copia: {copia}')
print(f'Lista Original: {original}')
```
