## Axes (Marcos)

+ Para poder graficar debe existir por lo menos un marco (axes). Se usa la función <span style="color:limegreen">add_subplot()</span> que requiere como argumentos el numero de filas, de columnas y el marco activo.

![Creación de marcos dentro de una figura](./Figuras/matplotlib_marcos.png)

La escala por defecto de cada eje de cada marco es de 0 a 1 con marcas cada 0.2 unidades.

```python
# Ejemplo: Matplotlib 5
import matplotlib.pyplot as plt # type: ignore

# Se crea la figura
plt.rcParams["figure.figsize"] = (6.4, 4.8)#(19.2, 10.8)
plt.rcParams["figure.dpi"] = 100

# Se crea la Figura
fig = plt.figure()
# Se crea el Marco
ax = fig.add_subplot(1, 1, 1)
```

```python
print(fig)
```

```python
# Ejemplo: Matplotlib 6
import matplotlib.pyplot as plt # type: ignore

# Se crea la Figura
fig = plt.figure()

# Creación del marco
for i in range(1, 5):
    ax = fig.add_subplot(2, 2, i)
    ax.set_title(f'Gráfico {i}')
    
fig.tight_layout(pad = 1.08)  # para que los titulos no se solapen
# plt.plot()
plt.show()
```

### Marcas

+ Por defecto, los ejes del marco tienen marcas (<span style="color:limegreen">ticks</span>) equiespaciadas y escaladas, denominadas `major ticks`.

+ También se pueden colocar marcas menores (<span style="color:limegreen">minor ticks</span>) que, a priori, no están activas.

+ Ambos elementos se pueden modificar de acuerdo a las necesidades, por ejemplo se pueden colocar las marcas menores con distinto espaciado en cada eje y además darle estilo diferente a cada rejilla.

Por ejemplo:

```python
# Ejemplo: Matplotlib 10
from matplotlib.ticker import MultipleLocator # type: ignore
import matplotlib.pyplot as plt # type: ignore
```

```python
# Forma Compacta
fig, ax = plt.subplots(1, 1)
ax.xaxis.set_minor_locator(MultipleLocator(0.1))  # X: separación cada 0.1 unidades
ax.yaxis.set_minor_locator(MultipleLocator(0.05))  # Y: separación cada 0.05 unidades
ax.set_title("Gráfico en Blanco")
ax.set_xlabel("Etiqueta para el eje X - $\\tau$")
ax.set_ylabel("Etiqueta para el eje Y - $\\psi$")
ax.xaxis.grid(True, color="r", linestyle="solid")
ax.yaxis.grid(True, color="g", linestyle="solid")
ax.xaxis.grid(which="minor", color="gray", linestyle="dashed")
ax.yaxis.grid(which="minor", color="lightskyblue", linestyle="dotted")
```

## Múltiples Funciones

```python
# Ejemplo: Matplotlib 13
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

## Leyendas

En el caso de que se tengan múltiples gráficos en el mismo marco puede ser deseable mostrar una leyenda identificativa. Para usarla se necesita asignar etiquetas a cada función.

```python
# Ejemplo: Matplotlib 14
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

```python
# En el mismo marco
fig, ax = plt.subplots()
ax.plot(x, y, label="seno")
ax.plot(x, z, label="coseno")
ax.legend()
ax.set_title('Funciones Trigonométricas');
```

```python
# Ejemplo: Matplotlib 15
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

```python
# En el mismo marco usando ecuaciones en Latex
fig, ax = plt.subplots()
ax.plot(x, y, label="$f_1(x) = \sin(x)$")
ax.plot(x, z, label="$f_2(x) = \cos(x)$")
ax.legend()
ax.set_title('Funciones Trigonométricas');
```

## Aplicación de Estilos

Para cada función que se incluye en un marco es posible establecer un [estilo personalizado](https://matplotlib.org/stable/api/_as_gen/matplotlib.pyplot.plot.html) de graficación con una gran cantidad de posibles argumentos.

+ Por ejemplo: Analizar las opciones de [matplotlib.legend](https://matplotlib.org/stable/api/legend_api.html).

```python
# Ejemplo: Matplotlib 16
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

```python
# En el mismo marco usando ecuaciones en Latex
fig, ax = plt.subplots()
ax.plot(x, y, 'ro', label = "$f_1(x) = \sin(x)$")
ax.plot(x, z, label = "$f_2(x) = \cos(x)$", linestyle='dotted')
# Leyenda en la parte superior al centro
ax.legend(loc = "lower left")
```

```python
# Ejemplo: Matplotlib 17
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

```python
# En el mismo marco usando ecuaciones en Latex
fig, ax = plt.subplots()

# definimos estilos
estilo_seno = dict(linewidth = 2, color = "darkorange")
estilo_coseno = dict(marker = "o", markerfacecolor = "limegreen", color = "darkgreen")

ax.plot(x, y, label="$f_1(x) = \sin(x)$", **estilo_seno)
ax.plot(x, z, label="$f_2(x) = \cos(x)$", **estilo_coseno)
ax.legend()
```

## Anotaciones

En ocasiones se necesita añadir ciertas [anotaciones](https://matplotlib.org/stable/tutorials/text/annotations.html) al gráfico que se está diseñando. 

> <span style="color:magenta">Por ejemplo</span>: Partiendo de las funciones <span style="color:limegreen">seno()</span> y <span style="color:limegreen">coseno()</span> con las que se está trabajando. Suponga que se desean obtener los puntos en el cual las dos curvas coinciden, es decir, resolver la siguiente ecuación:

$$
\begin{align}
\sin(x) &= \cos(x)\\
&\Downarrow \\
x &= \frac{\pi}{4} + \pi n,\,\, n \in Z
\end{align}
$$

Para $n = 0$ se obtiene una solución.

En el ejemplo, se insertarán algunas anotaciones en el gráfico para mayor claridad:

1. Una flecha señalando el punto de cruce con la etiqueta de la ecuación.

2. Las coordenadas de la solución en el punto de cruce.

3. Proyección del punto de cruce en ambos ejes.

```python
# Ejemplo: Matplotlib 19
import numpy as np
import matplotlib.pyplot as plt

x = np.linspace(0, 2 * np.pi)
y = np.sin(x)
z = np.cos(x)
```

```python
# Puntos de corte
xsol = np.pi / 4
ysol = np.sin(xsol)
```

```python
# En el mismo marco usando ecuaciones en Latex
fig, ax = plt.subplots()

# definimos estilos
estilo_seno = dict(linewidth = 3, color = "darkorange")
estilo_coseno = dict(marker = "o", markerfacecolor = "limegreen", color = "darkgreen")

# Modificando límites
ax.set_xlim(0, np.pi / 2)
ax.set_ylim(0, 1)
ax.grid()

# Anotaciones ax.annotate(text, xy, xytext=None, xycoords='data', \
#    textcoords=None, arrowprops=None, annotation_clip=None, **kwargs)
ax.annotate(
    "$\sin(x) = \cos(x)$", # texto
    xy = (xsol, ysol), # posicion del punto donde se colocará el comentario
    xytext = (1.2, 0.8), # posicion del texto del comentario
    arrowprops = dict(color='lightblue', facecolor = "lightblue", shrink = 0.05), # propiedades de la flecha
)
# Escribe la tupla solucion
ax.text(0.47, 0.72, f"({xsol:.2f}, {ysol:.2f})")

ax.plot(x, y, label = "$f_1(x) = \sin(x)$", **estilo_seno)
ax.plot(x, z, label = "$f_2(x) = \cos(x)$", **estilo_coseno)

# Grafica las proyecciones
ax.plot([xsol, xsol], [0, ysol], color = "gray", linestyle = "--")
ax.plot([0, xsol], [ysol, ysol], color = "blue", linestyle = "dotted")
ax.legend()
plt.show()
```

Este medallero se encuentra en un archivo Excel denominado <span style="color:magenta">Medallas.xlsx</span> con la información de Ranking por medalllas de oro, Equipo o País, Cantidad de medallas de oro, cantidad de medallas de plata, cantidad de medallas de bronce, total de medallas y Ranking por total de medallas. En total se cuenta en el archivo con 92 registros. Se solicita realizar un diagrama de codigo de barras por ranking de cada País por cantidad de medallas.

```python
# Ejemplo: Matplotlib 20 Graficos de Barras
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("Archivos/Medallas.xlsx", engine = "openpyxl")
df.head()
#print(f'DataFrame Original: \n{df.head()}')
```

```python
# Creación de un Gráfico de Barras
# Con medallas de oro, plata y bronce ordenados por ranking
df_mejor = df.nsmallest(10, "Rank")
df_mejor
#print(f'Ranking: \n{df_mejor}\n')
```

```python
df_mejor.index
```

```python
np.arange(df_mejor.index.size)
```

```python
ancho_barra = 0.30
x = np.arange(df_mejor.index.size)

fig, ax = plt.subplots(figsize = (8, 5), dpi = 100)  # 800 x 500 px
medallas_oro = ax.bar(
    x - ancho_barra, df_mejor["Gold"], ancho_barra, label = "Oro", color = "#ffd700"
)
medallas_plata = ax.bar(
    x, df_mejor["Silver"], ancho_barra, label = "Plata", color = "#aaa9ad"
)
medallas_bronce = ax.bar(
    x + ancho_barra, df_mejor["Bronze"], ancho_barra, label = "Bronce", color = "#cd7f32"
)
ax.set_xticks(x, df_mejor['Rank'], rotation = 90) # df_mejor.index
ax.legend()

# Etiquetas de las barras
ax.bar_label(medallas_oro, padding = 3)
ax.bar_label(medallas_plata, padding = 3)
ax.bar_label(medallas_bronce, padding = 3)

# Oculta borde derecho y superior
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)
fig.tight_layout()  # Ajusta elementos al tamaño de la figura
```

### Ejemplo 2

Para ejemplificar el uso de un diagrama de dispersión se cuenta con un dataset histórico de los jugadores de la NBA (<span style="color:magenta">nba_data.csv</span>). Este archivo de texto posee variada información de cada jugador entre las que podemos citar, nombre del jugador, equipo en el que juega, cantidad de puntos, rebotes y asistencias, etc.

Se solicita realizar un gráfico de dispersión tomando como referencia las variables cantidad de puntos realizados, cantidad de rebotes ganados y cantidad de asistencias realizadas.

```python
# Ejemplo: Matplotlib 21 Grafico de Dispersión
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("Archivos/nba_data.csv", usecols = ["pts", "reb", "ast"])
print(f'Dataframe NBA: \n{df.head(10)}')
```

```python
# Creacion Grafico de Dispersión
# Relación puntos anotados con rebotes capturados y asistencias dadas.
# Crear Variables auxiliares
x = df["pts"]
y = df["reb"]
colores = df["ast"]
```

```python
fig, ax = plt.subplots(figsize = (8, 6), dpi = 100)
p = ax.scatter(
    x,
    y,
    s = 10,  # tamaño de los puntos
    c = colores,
    cmap = "RdBu_r",  # Colores
    vmin = colores.min(),
    vmax = colores.max(),  # normalización de colores
    alpha = 0.7,
    edgecolors = "none",
)

# Barras de Colores
cb = fig.colorbar(p, ax = ax, label = "Asistencias", extend = "max")
cb.outline.set_visible(False)

# Rotulos de los ejes
ax.set_xlabel("Puntos")
ax.set_ylabel("Rebotes")

# Saca los spines
ax.spines["right"].set_visible(False)
ax.spines["top"].set_visible(False)
fig.tight_layout()
```

### Ejemplo 3

### Ejemplo 4

Se dispone de un <span style="color:yellow">DataFrame</span> con información de las <span style="color:yellow">temperaturas globales</span> por mes y año desde 1750 hasta el 2015 (3193 registros). Esta información se dispone en un archivo de texto denominado <span style="color:magenta">GlobalTemperatures.csv</span>. Lamentablemente la información de este <span style="color:yellow">dataset</span> esta fragmentada e incompleta. De toda la información disponible se utilizará un diagrama para Series temporales y la linea de tendencia sobre la <span style="color:yellow">temperatura global promedio</span> y la <span style="color:yellow">fecha</span> en la que se produjo. 

```python
# Ejemplo: Matplotlib 23 Grafico para Series Temporales
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Archivos/GlobalTemperatures.csv', 
                 parse_dates = ['dt'], # Conversión a tipo datetime
                 usecols = ['dt', 'LandAverageTemperature'])
```

```python
df.rename(columns = {'dt': 'Cuando', 'LandAverageTemperature': 'Temp'}, inplace = True)
```

```python
# Remueve los valores faltantes
df.dropna(inplace = True) 
print(f'Dataframe Leído: \n{df.head(10)}\n')
print(f'Formato: {df.shape}')
```

```python
# Gráfico de las mediciones Históricas de la Temperatura media global
# del Planeta y se agrega línea de tendencia
from matplotlib.dates import YearLocator, DateFormatter, date2num
from matplotlib.ticker import MultipleLocator
```

```python
# Definimos alias
x = df.Cuando
y = df.Temp
```

```python
fig, ax = plt.subplots(figsize = (8, 4), dpi = 100)
ax.plot(x, y,
        linestyle = 'None', marker = '.', color = 'tomato', # estilo de línea
        zorder = 2) # orden para colocar sobre rejilla

# Construcción de la línea de tendencia
x = date2num(x)
z = np.polyfit(x, y, 2) # ajuste por mínimos cuadrados
p = np.poly1d(z)
plt.plot(x, p(x), linewidth = 4, alpha = 0.8, color = 'royalblue')

# Formato de ejes
ax.xaxis.set_minor_locator(YearLocator(10))
ax.xaxis.set_minor_formatter(DateFormatter('%Y'))
ax.tick_params(axis = 'x', which = 'minor',
               labelsize = 8, labelcolor = 'lightgray', rotation = 90)

ax.xaxis.grid(which = 'minor', color = 'lightgray', linestyle = 'dashed')
ax.yaxis.set_major_formatter('{x:.0f}º')
ax.yaxis.set_minor_locator(MultipleLocator(1))
ax.tick_params(axis = 'y', which = 'minor',
               labelsize = 8, labelcolor = 'lightgray')

ax.yaxis.grid(which = 'minor', color = 'lightgray', linestyle = 'dashed')
ax.yaxis.set_minor_formatter('{x:.0f}')
ax.tick_params(axis = 'y', which = 'minor',
               labelsize = 8, labelcolor = 'lightgray')

ax.spines['right'].set_visible(False)
ax.spines['top'].set_visible(False)
fig.tight_layout()
```

### Ejemplo 5

Para mostrar el uso de los diagramas de calor se utilizará un <span style="color:yellow">dataset</span> de las 1000 películas mas valoradas en IMDB. Este <span style="color:yellow">dataset</span> se encuentra en el archivo <span style="color:magenta">imdb_top_1000.csv</span>. Para este propósito se utilizarán las columnas Certificate: clasificación según edades, Genre: género y Rating: valoración en IMDB. 

```python
# Ejemplo: Matplotlib 24 Grafico para Mapas de Calor
# Data set las 1000 películas más valoradas en IMDB
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv('Archivos/imdb_top_1000.csv', 
                 usecols=['Certificate', 'Genre', 'IMDB_Rating'])
print(f'Dataframe leído: \n{df.head(10)}\n')
```

```python
# Se crea una nueva columna con el genero principal de cada pelicula
df['Genero Principal'] = df['Genre'].str.split(',', expand = True)[0]
print(f'Dataframe modificado: \n{df.head(10)}\n')
```

```python
# Se agrupan y se obtienen los valores medios de las valoraciones
# unstack() permite disponer de la agrupación en forma de tabla
calificacion = df.groupby(['Certificate', 'Genero Principal'])['IMDB_Rating'].mean().unstack()
print(f'Clasificación: \n{calificacion}\n')
```

```python
# Toma solo un subconjunto de certificados y generos.
revision_certif = ['U', 'UA', 'PG-13', 'R', 'A']
revision_generos = ['Animation', 'Action', 'Adventure', 'Biography',
                    'Comedy', 'Crime', 'Drama']
calificacion = calificacion.loc[revision_certif, revision_generos]
```

```python
# Se recodifican los certificados con códigos entendibles
descripcion_certif = {'U': 'ALL', 'UA': '>12', 'PG-13': '>13', 'R': '>17', 'A': '>18'}
calificacion.index = calificacion.reset_index()['Certificate'].replace(descripcion_certif)
print(f'Clasificación: \n{calificacion}\n')
```

```python
# Construcción de la figura
fig, ax = plt.subplots(figsize = (8, 4), dpi = 100)

colores_texto = ('black', 'white')
im = ax.imshow(calificacion, cmap='Reds') # Mapa de calor
cbar = fig.colorbar(im, ax = ax, label = 'Clasificación IMDB') #leyenda

x = calificacion.columns
y = calificacion.index

# Muestra las etiquetas. El color del texto cambia segun su mormalizacion
for i in range(len(y)):
    for j in range(len(x)):
        valor = calificacion.iloc[i, j]
        color_texto = colores_texto[int(im.norm(valor) > 0.5)] # color etiqueta
        ax.text(j, i, f'{valor:.2f}', color = color_texto, va = 'center', ha = 'center')

# Formato de los ejes
ax.set_xticks(range(len(x)))
ax.set_xticklabels(x, rotation=90)
ax.set_yticks(range(len(y)))
ax.set_yticklabels(y)
ax.invert_yaxis()
ax.spines[:].set_visible(False)
fig.tight_layout()
```

### Ejemplo 6

Para ejemplificar un gráfico de evolución se utiliza un <span style="color:yellow">dataset</span> de la evolución del precio de una cripto moneda. El objetivo será crear un gráfico que represente el valor de la criptomoneda (a lo largo del tiempo) en contraposición al volumen de unidades disponibles (<span style="color:magenta">eth_usd.csv</span>).

```python
# Ejemplo: Matplotlib 25 Grafico de Evolución
# Precios de Criptomonedas Ethereum
# Crear gráfico que represente el valor de la criptomoneda 
# en contraposicion al volúmen de unidades

import pandas as pd
import matplotlib.pyplot as plt
import datetime

df = pd.read_csv('Archivos/eth_usd.csv', parse_dates = ['Date'],
                 usecols = ['Date', 'Open', 'Volume'],
                 index_col = 'Date')
```

```python
fecha_min = datetime.datetime(year = 2017, month = 1, day = 1)
```

```python
df = df.loc[df.index > fecha_min]
df['Volume'] /= 1e7
df_suave = df.rolling(20).mean().dropna() # proporciona calculos de ventanas moviles
print(f'Dataframe Suave: \n{df_suave.head(20)}\n')
```

```python
# Crea el Gráfico
fig, ax = plt.subplots(figsize = (8, 4), dpi = 100)

# Alias para facilitar el acceso
x = df_suave.index
y_abierto = df_suave['Open']
y_vol = df_suave['Volume']

# Lineas de Evolución
ax.plot(x, y_abierto, label = 'Valor ($)', color = 'skyblue', linewidth = 1.5)
ax.plot(x, -y_vol, label = 'Volumen (10M ud.)', color = 'pink', linewidth = 1.5)

# Relleno del área
plt.fill_between(x, y_abierto, alpha = 0.5, color = 'skyblue', zorder = 3)
plt.fill_between(x, -y_vol, alpha = 0.5, color = 'pink', zorder = 3)

# Formato de los ejes
ax.xaxis.set_ticks_position('none')
ax.yaxis.set_ticks_position('none')
y_ticks = [-4000, -2000, 0, 2000, 4000]
y_tick_labels = ['4000', '2000', '0', '2000', '4000']
ax.set_yticks(y_ticks)
ax.set_yticklabels(y_tick_labels)
ax.set_ylim(-6000, 6000)

# Grilla
ax.xaxis.grid(color='lightgray', linewidth=.5)
for y_tick in y_ticks:
    if y_tick != 0:
        ax.axhline(y_tick, color='lightgray', linewidth=.5)
ax.legend()
ax.spines[:].set_visible(False)
fig.tight_layout()
```
