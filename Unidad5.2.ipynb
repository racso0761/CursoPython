+ Son objetos similares a las listas.

+ son <span style="color:yellow">inmutables</span>, <span style="color:magenta">NO</span> se pueden modificar.

+ Pueden contener cualquier tipo de objetos.


### <font color=#38bdf8>Creando Tuplas</font>

+ Se crean las tuplas de igual manera que las listas pero en lugar de corchetes se usan paréntesis.

+ También es posible no utilizar los paréntesis para crear una tupla, aunque no es lo recomendado.

+ Las tuplas de un solo elemento tienen una sintaxis especial.

```python
# Ejemplo: Tuplas 1
# Ejemplo: Creacion de tuplas
tupla_vacia = ()
otra_forma = tuple()

geolocalizacion_sanjuan = (-31.545395, -68.513058)
reyes_magos = ('Melchor', 'Gaspar', 'Baltazar')
temperatura = 23, 26, 29, 12
```

```python
print(f'GeoLoc: {geolocalizacion_sanjuan} \
      \nReyes Magos: {reyes_magos}\nTemperatura: {temperatura}')
```

```python
a = (25)
print("Tupla?: ", type(a))
```

```python
a = (25,)
print("Tupla?: ", type(a))
```

### <font color=#38bdf8>Conversión a Tuplas</font>

Se utiliza la función <span style="color:limegreen">tuple()</span> para convertir otros tipos de datos en una tupla, siempre y cuando sean tipos de datos iterables.

```python
# Ejemplo: Tuplas 2
# Ejemplo de conversión a tuplas
lista_compras = ['Agua', 'Aceite', 'Arroz']
```

```python
tupla_compras = tuple(lista_compras)
print("Tipo de datos: ", type(tupla_compras))
print("\nTupla: ", tupla_compras)
```

```python
a = 20,
tupla_entero = tuple(a) # no es iterable
print(tupla_entero)
```

### <font color=#38bdf8>Operaciones con Tuplas</font>

+ Las tuplas admiten todas las operaciones de las listas, excepto aquellas que conllevan una modificación.

+ Recordar que la tuplas son inmutables.

+ La comparación de tuplas se hace de igual manera que la comparación de listas.

#### <font color=#38bdf8>Desempaquetado de Tuplas</font>

Esta potente característica permite asignar una tupla a variables independientes.

![Desempaquetado de un Tupla](./Figuras/desempaquetado.png)

```python
# Ejemplo: Tuplas 3
# Ejempo: Desempaquetado de Tuplas

reyes_magos = ('Melchor', 'Gaspar', 'Baltazar')

otro = reyes_magos
print(f'Tupla: {otro}')
```

```python
var1, var2, var3 = reyes_magos
print(f'Rey 1: {var1}')
print(f'Rey 2: {var2}')
print(f'Rey 3: {var3}')
```

```python
# Funcion divmod devuelve una tupla
cociente, resto = divmod(27,4)
print(f'27/4 --> Q: {cociente}, R: {resto}')
```

```python
Q = divmod(27,4)
type(Q)
```

#### <font color=#38bdf8>Intercambio de Valores</font>

Las tuplas se pueden usar para intercambiar dos valores.

#### <font color=#38bdf8>Desempaquetado Extendido</font>

Con el operador <span style="color:limegreen">*</span> podemos indicar grupos de desempaquetado.


```python
# Ejemplo: Tuplas 4
valor1 = 40
valor2 = 30
print(f'valor1: {valor1}, valor2: {valor2}')
```

```python
# Intercambio de valores
valor1, valor2 = valor2, valor1
print(f'valor1: {valor1}, valor2: {valor2}')
```

```python
# Ejemplo: Tuplas 5
# Ejemplo: Desempaquetado Extendido
ranking = ('primero', 'segundo', 'tercero', 'cuarto', 'quinto')

cabeza, *cuerpo, cola = ranking
print(cabeza)
print(cuerpo) # Notar que lo convierte en una lista
print(cola)
```

```python
temperatura = (23, 29, 35, 42, 12, 18, -1)
primero, segundo, *otros, ultimo = temperatura
print(primero)
print(segundo)
print(otros)
print(ultimo)
```

```python
frase = 'Somos solo humanos'
cabeza, *cuerpo, cola = frase
print(type(cabeza))
print(cabeza)
print(cuerpo)
print(cola)
```

#### <font color=#38bdf8>Tuplas vs Listas</font>

Aunque son estructuras similares tienen diferencias significativas.

Por que usar tuplas?:

+ Ocupan menos especio en memoria.

+ Poseen mecanismos de protección frente a cambios indeseados.

+ Se pueden utilizar como <span style="color:yellow">claves</span> en los diccionarios (son <span style="color:yellow">hasheables</span>).

+ Las [namedtuples](https://docs.python.org/es/3/library/collections.html#collections.namedtuple) son una alternativa sencilla a los objetos.

+ No admiten la característica de tuplas por comprensión. Si se intenta hacer el intérprete no da error, pero se obtiene un objeto denominado <span style="color:yellow">generador</span>.


```python
# Ejemplo: Tuplas 6
mi_rango = (numero for numero in range(1, 6))

print(type(mi_rango))
```

+ Se basan en el concepto de un diccionario real. (palabras: significado).

+ Un diccionario contiene un conjunto de elementos de la forma <span style="color:yellow">(claves : valor)</span>. Es un objeto indexado por <span style="color:yellow">claves</span> que tienen asociados unos <span style="color:yellow">valores</span>.

+ Mantienen el orden en el que se insertan las claves.

+ Las <span style="color:yellow">claves</span> deben ser únicas y objetos inmutables. (normalmente se usan strings).

+ Acceso muy rápido a sus elementos.

+ Son objetos-iterables.

### <font color=#38bdf8>Iteración sobre un Diccionario</font>

Para iterar sobre un diccionario se ofrecen tres alternativas:

+ Iterar sobre las <span style="color:yellow">claves</span>.

+ Iterar sobre los <span style="color:yellow">valores</span>.

+ Iterar sobre los pares <span style="color:yellow">clave : valor</span>.

```python
# Ejemplo: Diccionarios 8
# Ejemplo: Longitud e iteración sobre diccionarios
rae = {
    'estudiante': 'Persona que cursa estudios en un establecimiento de enseñanza.',
    'profesor': 'Persona que ejerce o enseña una ciencia o arte.',
    'administrativo': 'Persona empleada en la administración de alguna entidad.'
    }

print("La longitud del 'diccionario' rae es: ", len(rae))
```

```python
# iteración sobre claves
for palabra in rae.keys():
    print("Claves: ", palabra)
```

```python
# iteración sobre valores
for significado in rae.values():
    print("Significado: ", significado)
```

```python
#iteración sobre pares clave-valor
for palabra, significado in rae.items():
    print(f'{palabra}: {significado}')
```

### <font color=#38bdf8>Combinar Diccionarios</font>

Dados dos o mas diccionarios es posible mezclarlos para obtener una combinación de ellos.

Esta combinación se basa en dos premisas:

1. Si la <span style="color:yellow">clave</span> no existe se añade con su valor.

2. Si la <span style="color:yellow">clave</span> ya existe, se añade el valor del <span style="color:magenta">último</span> diccionario en la combinación.


Python ofrece tres maneras de combinar diccionarios:

+ Sin modificar los diccionarios originales: mediante el operador <span style="color:limegreen">**</span>.

+ Sin modificación de los diccionarios originales mediante el operador <span style="color:limegreen">|</span> ( a partir de la version 3.9).

+ Modificando los diccionariso originales: mediante el método <span style="color:limegreen">update()</span>.

```python
# Ejemplo: Diccionarios 9
# Ejemplo: Combinación de diccionarios
rae1 = {
    'estudiante': 'Persona que cursa estudios en un establecimiento de enseñanza.',
    'profesor': 'Persona que ejerce o enseña una ciencia o arte.',
    'administrativo': 'Persona empleada en la administración de alguna entidad.'
    }

rae2 = {
    'ayudante':'En algunos cuerpos y oficinas, oficial subalterno.',
    'jefe': 'Superior o cabeza de una corporación, partido u oficio.'
}
```

```python
# Sin modificación de los diccionarios originales 1
nuevo_rae = {**rae1, **rae2}
for clave, significado in nuevo_rae.items():
    print(f'{clave}: {significado}')
```

```python
# Sin modificación de los diccionarios originarles 2
# A partir de Python v3.9 se puede utilizar el operador "|" para combinar diccionarios
otro_rae = rae1 | rae2
for clave, significado in otro_rae.items():
    print(f'{clave}: {significado}')
```

```python
# Con modificación del diccionario rae1
rae1.update(rae2)
for clave, significado in rae1.items():
    print(f'{clave}: {significado}')
```

### <font color=#38bdf8>Copias de Diccionarios</font>

Al igual que con las listas se debe tener cuidado cuando se copian diccionarios.

+ Copia del nombre.

+ Copia profunda (real).

```python
# Ejemplo: Diccionarios 11
# Ejemplo: Copias de diccionarios
original_rae = {
    'estudiante': 'Persona que cursa estudios en un establecimiento de enseñanza.',
    'profesor': 'Persona que ejerce o enseña una ciencia o arte.',
    'administrativo': 'Persona empleada en la administración de alguna entidad.',
    'ayudante':'En algunos cuerpos y oficinas, oficial subalterno.',
    'jefe': 'Superior o cabeza de una corporación, partido u oficio.'
}
```

```python
copia_rae = original_rae
for clave, significado in copia_rae.items():
    print(f'{clave}: {significado}')
```

```python
original_rae['estudiante'] = 'Estudiante de Bioingeniería'
for clave, significado in copia_rae.items():
    print(f'{clave}: {significado}')
```

```python
# Solución copias profundas
import copy

copia_rae = copy.deepcopy(original_rae)
copia_rae['ayudante'] = 'Alguna cosa interesante'
for clave, significado in copia_rae.items():
    print(f'{clave}: {significado}')
```

```python
for clave, significado in original_rae.items():
    print(f'{clave}: {significado}')
```

### <font color=#38bdf8>Objetos "Hashables"</font>

+ La única restricción que deben cumplir las *claves* de un diccionario es que sean <span style="color:yellow">hashables</span>.

> > <span style="color:magenta">Un objeto es hashable si se le puede asignar un valor hash que no cambia durante toda la vida de ese objeto</span>.

+ Para encontrar el valor hash de un objeto se usa la función <span style="color:limegreen">hash()</span>.

+ Para nosotros en este curso un objeto es <span style="color:yellow">hashable</span> cuando es inmutable.

### <font color=#38bdf8>Algo mas sobre Listas y Diccionarios</font>

```python
#Ejemplo
# Listas
lista1 = [1, 2]
lista2 = [3, 4]
combinada = [*lista1, *lista2]
print(f'Lista con desempaquetado: {combinada}')
combinada2 = [lista1, lista2]
print(f'Lista sin desempaquetado: {combinada2}')

# Diccionarios
dict1 = {"x": 1, "y": 2}
dict2 = {"y": 99, "z": 3}
combinado = {**dict1, **dict2}
print(f'Diccionario con desempaquetado: {combinado}')
# Ver Combinar Diccionarios
combinado2 = {dict1, dict2}
print(f'Diccionario sin desempaquetado: {combinado2}')
```

### <font color=#38bdf8>Creación de Conjuntos</font>

+ Para crear un conjunto se deben separar sus elementos con comas y encerrarlos entre llaves.

+ Si se desea crear un conjunto vacío, no se puede realizar a través de las llaves, pues eso crea un diccionario.

#### <font color=#38bdf8>Teoría de Conjuntos</font>

En Python se puede operar con conjuntos de la misma manera que se opera en la teoría de conjuntos. Las operaciones definidas son:

+ <span style="color:yellow">Intersección</span>: Elementos que están a la vez en A y en B ($A \cap B$). Se usa el operador <span style="color:limegreen">&</span>.

+ <span style="color:yellow">Union</span>: Elementos que están tanto en A como en B ($A \cup B$). Se usa el operador <span style="color:limegreen">|</span>.

+ <span style="color:yellow">Diferencia</span>: Elementos que están en A y no están en B ($A - B$). Se usa el método <span style="color:limegreen">difference()</span>.

+ <span style="color:yellow">Diferencia simétrica</span>: Elementos que están en A o en B pero no en ambos ($\overline{A \cap B}$). Se usa el operador <span style="color:limegreen">^</span>.

+ <span style="color:yellow">Inclusión</span>: 

    + Un conjunto B es un subconjunto de otro A si todos los elementos de B están incluidos en A. ($B \subset A$), ($B \subseteq A$). Se usa el operador <span style="color:limegreen"><</span>, <span style="color:limegreen"><=</span>.
    
    + Un subconjunto A es un superconjunto de otro conjunto B si todos los elementos de B están incluidos en A. ($A \supset B$), ($A \supseteq B$). Se usa el operador <span style="color:limegreen">></span>, <span style="color:limegreen">>=</span>.


```python
# Ejemplo: Diccionarios 17
# Ejemplo: Teoría de Conjuntos
A = {1, 2}
B = {2, 3}
```

```python
C = A & B
print(f'Conjunto Intersección: {C}\n')
```

```python
D = A | B
print(f'Conjunto Union: {D}\n')
```

```python
E = A - B
print(f'Diferencia: {E}\n')
# E = A.difference(B)
```

```python
F = A ^ B
print(f'Diferencia Simétrica: {F}\n')
```

```python
A = {1, 2, 3, 4}
B = {2, 3}
G = B < A
print(f'Es B subconjunto de A: {G}\n')
```

```python
H = B > A
print(f'Es B superconjunto de A: {H}')
```

Python ofrece la oportunidad de crear conjuntos inmutables. Para esto se utiliza la función <span style="color:limegreen">frozenset()</span>.
