## <font color=#38bdf8>Mas sobre Strings</font>

Python posee dos funciones importantes que se deben conocer en el trabajo habitual con strings.

En Python, las funciones <span style="color:limegreen">str()</span> y <span style="color:limegreen">repr()</span> se utilizan para obtener representaciones de strings. Aunque a primera vista pueden parecer similares, existen algunas diferencias en su comportamiento. Ambas funciones pueden ser útiles para depurar o imprimir información útil sobre un objeto.

Las funciones son:

+ <span style="color:limegreen">str()</span>.

+ <span style="color:limegreen">repr()</span>.

```python
# Ejemplo: String+ 1
s = 'Hola, Cheee.'
print (str(s))
print (str(2.0/11.0))
```

```python
print (repr(s))
print (repr(2.0/11.0))
```

### <font color=#38bdf8>str() vs repr()</font>

De la salida del ejemplo anterior se podría pensar que las funciones son iguales o que funcionan de la misma manera. En realidad son distintas y se analizarán sus diferencias:

+ La función <span style="color:limegreen">str()</span> se usa para crear los mensajes que serán presentados al usuario final, por lo que son fácilmente legibles. Mientras que la función <span style="color:limegreen">repr()</span> se usa con propósitos de depuración y desarrollo, por lo que sus mensajes han de ser inequívocos.

+ Además, <span style="color:limegreen">repr()</span> calcula la representación <span style="color:yellow">"oficial"</span> de un string y <span style="color:limegreen">str()</span> se usa para calcular la representación <span style="color:yellow">"informal"</span> de un string. Por lo que la primera ha de contener más información que la segunda.

+ La función <span style="color:limegreen">print()</span> usa <span style="color:limegreen">__str__</span>, al igual que la función <span style="color:limegreen">str()</span> para generar el string que representa al objeto. Solo si no se ha definido una implementación de <span style="color:limegreen">__str__</span>, las dos funciones usarán la implementación que exista de <span style="color:limegreen">__repr__</span>.

```python
# Ejemplo: String+ 2
s = 'Hola\tsoy\tun\tstring\ncon\tformato'
print(str(s))
print(repr(s))
```

```python
# Ejemplo: String+ 3
s = r'Hola\tsoy\tun\tstring\ncon\tformato'
print(str(s))
print(repr(s))
```

```python
# Ejemplo: String+ 4
s = '''Hola soy
un string
multilinea y siempre
uso 3 apostrofes'''

print(str(s))
print(repr(s))
```

```python
# Ejemplo: String+ 5
import datetime
today = datetime.datetime.now()
print(str(today))
print(repr(today))
```
