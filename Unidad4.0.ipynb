### <font color=#38bdf8>Evaluación de Cortocircuito</font>

+ Los predicados lógicos no siempre se evalúan por completo, dependen de determinadas circunstancias.

+ Cuando Python tiene la certeza del resultado deja de evaluar. A esto se le denomina escenario de cortocircuito.

+ se utiliza <span style="color:limegreen">and</span> cuando se desea que se cumplan ambas condiciones simples al mismo tiempo.

+ se utiliza <span style="color:limegreen">or</span> cuando se desea que una u otra se cumpla.

````python
# Ejemplo: Control de Flujo 9
edad = 15
altura = 1.8
````

````python
# Evaluacion de cortocircuito
print(edad > 20 and altura > 1.6)
````

````python
print(edad < 20 or altura > 1.6) 
````

### <font color=#38bdf8><span style="color:yellow">if</span> Alternativo</font>

En situaciones de sentencias anidadas (una dentro de la otra) Python posee la sentencia <span style="color:limegreen">elif</span> que sustituye un <span style="color:limegreen">else</span> y un <span style="color:limegreen">if</span> juntos.

Sintaxis:

![](./Figuras/if_alternativo.png)

````python
# Ejemplo: Control de Flujo 5
# Uso de elif
temp_ambiente = 28

if temp_ambiente < 20:
    if temp_ambiente < 10:
        print(f'Nivel Azul')
    else:
        print(f'Nivel Verde')
elif temp_ambiente < 30:
        print(f'Nivel Naranja')
else:
        print(f'Nivel Rojo')
````

### <font color=#38bdf8>Asignaciones Condicionales</font>

Suponga que se desea asignar un nivel de riesgo de incendio en función de la temperatura. Normalmente se escribiría el código de la siguiente manera:

````python
# Ejemplo: Control de Flujo 6
temperatura = 35

if temperatura < 30:
    riesgo_incendio = 'BAJO'
else:
    riesgo_incendio = 'ALTO'

print('El riesgo de incendio es: ', riesgo_incendio)
````

La forma condicional de escribir el código anterior es:

 <p>&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp&nbsp <span style="color:yellow">accion_verdadera if condicion else accion_falsa</span>

````python
# Ejemplo: Control de Flujo 6
temperatura = 35
riego_incendio = 'BAJO' if temperatura < 30 else 'ALTO'
print('El riesgo de incendio es: ', riego_incendio)
````

## <font color=#38bdf8>Sentencia de Selección:  <span style="color:limegreen">match-case</span></font>

La versión 3.10 de Python incorpora la sentencia <span style="color:limegreen">match-case</span> que es parecida a la sentencia <span style="color:yellow">switch</span> del lenguaje C.

+ En su versión mas simple permite comparar un valor de entrada con una serie de constantes. Como una serie de sentencias <span style="color:limegreen">if</span> anidadas.

Sintaxis:

![](./Figuras/match_case.png)

````python
# Ejemplo: Control de Flujo 10
letra_g = '3b3'
lg = int(letra_g, 16)
print('Valor: ', lg, 'en hexa: ', hex(lg))
match letra_g:
    case '3b1':
        print('Caracter: ', chr(lg))
    case '3b2':
        print('Caracter: ', chr(lg))
    case '3b3':
        print('Caracter: ', chr(lg))
    case _:
        print("Codigo no reconocido") 
````

La sentencia <span style="color:limegreen">match-case</span> no solo permite comparar un valor con constantes, también permite deconstruir estructuras de datos, capturar elementos y mapear distintos valores.

[+info match-case](https://docs.python.org/es/dev/tutorial/controlflow.html)

````python
# Ejemplo: Control de Flujo 11
punto = (2, 3, 4, 5)

match punto:
    case (x, y):
        print(f'({x}, {y}): está en el plano.')
    case (x, y, z):
        print(f'({x}, {y}, {z}) está en el espacio.')
    case _:
        print ("Punto en otra dimensión")
````

````python
x = 10 #25
match x:
    case 10 | 20 | 30:  # Matches 10, 20, or 30
        print(f"Coincide: {x}")
    case _:
        print("No se encontró ninguna coincidencia")
````

````python
mes = 4
match mes:
    case 12 | 1 | 2: print("Invierno")
    case 3 | 4 | 5: print("Primavera")
    case 6 | 7 | 8: print("Verano")
    case 9 | 10 | 11: print("Otoño")
    case _: print("Error")
````

````python
coordenada = (0, 1)
match coordenada:
    case (0, 0):
        print("Coordenada en origen")
    case (x, 0):
        print(f"Coordenada en eje x: {x}")
    case (0, y):
        print(f"Coordenada en eje y: {y}")
    case (x, y):
        print(f"Coordenada en: {x}, {y}")
    case _:
        print("Error")
````

````python
# Ejemplo: Control de Flujo 12
# Calcular la distancia al origen.
# Sin saber si el punto está en el plano o el espacio
punto = (2, 5)

match punto:
    case (int(x), int(y)):
        dist_al_origen = (x ** 2 + y ** 2) ** (1 / 2)
    case (int(x), int(y), int(z)):
        dist_al_origen = (x ** 2 + y ** 2 + z ** 2) ** (1 / 2)
    case _: 
        dist_al_origen = 0

if not dist_al_origen:
    print(f'Distancia desconocida.')
else:
    print(f'{punto}, distancia al origen {dist_al_origen}')
````

## <font color=#38bdf8>Iteraciones</font>

### <font color=#38bdf8>La sentencia <span style="color:limegreen">while</span></font>

Esta sentencia permite repetir un grupo de acciones en función de una condición.

Sintaxis:

![](./Figuras/while.png)

+ Mientras la <span style="color:magenta">condición</span> evalúe a <span style="color:limegreen">True</span> -> se ejecutan acciones

+ Cuando la <span style="color:magenta">condición</span> evalúa a <span style="color:limegreen">False</span> -> termina la repetición

````python
# Ejemplo: Control de Flujo 14
salida = 'N'

while salida == 'N':
    print("Hola, Como estas?")
    salida = input('Desea Salir [no N]: ')
````

### <font color=#38bdf8>Finalizar una Iteración</font>

Python permite finalizar la iteración antes de que se cumpla la condición de detención.
Para esto se utiliza la sentencia <span style="color:limegreen">break</span>.

````python
# Ejemplo: Control de Flujo 15
salida = 'N'
respuestas = 0

while salida == 'N' or salida == 'n':
    print("Hola, Como estas?")
    salida = input('Desea Salir [no N]: ')
    respuestas += 1
    if respuestas == 4:
        print('Numero máximo de respuestas alcanzado')
        break
print('Chauuuu')
````

````python
# Otra forma mas elegante sería:

while (salida == 'N' or salida == 'n') and (respuestas < 4):
    pass
````

Python permite evaluar si la finalización de la iteración se produjo normalmente o por la ejecución de la sentencia <span style="color:limegreen">break</span>.

Para esto la sentencia <span style="color:limegreen">while</span> cuenta con una sentencia <span style="color:limegreen">else</span> que se ejecuta siempre y cuando la iteración haya finalizado normalmente.

<font color=yellow>while condicion:</font>  
&nbsp;&nbsp;&nbsp;&nbsp;<font color=yellow>acciones_1</font>

<font color=yellow>else:</font>  
&nbsp;&nbsp;&nbsp;&nbsp;<font color=yellow>acciones_2</font>

````python
# Ejemplo: Control de Flujo 16
salida = 'N'
respuestas = 0

while salida == 'N' or salida == 'n':
    print("Hola, Como estas?")
    salida = input('Desea Salir [no N]: ')
    respuestas += 1
    if respuestas == 4:
        print('Numero máximo de respuestas alcanzado')
        break
else:
    print('Usted decidió Salir')
print('Chauuuu')
````

### <font color=#38bdf8>Continuar con una Iteración</font>

De la misma manera que se permite finalizar prematuramente una iteración también se permite continuar una iteración. Saltando hacia adelante a la siguiente iteración. Con este objetivo se utiliza la sentencia <span style="color:limegreen">continue</span>.

````python
# Ejemplo: Control de Flujo 17
salida = 'N'
opciones_validas = 0

while salida == 'N' or salida == 'n':
    print("Hola, Como estas?")
    salida = input('Desea Salir [S/N]: ')
    if salida not in 'SNsn':
        salida = 'N'
        continue
    opciones_validas += 1
print(f'{opciones_validas} respuestas validas')
print('Chauuuu')
````

## <font color=#38bdf8>La sentencia</font> <font color=yellow>for</font>

La sentencia <span style="color:limegreen">for</span> se utiliza principalmente para recorrer aquellos tipos de datos que son *iterables*, generalmente secuencias y mapeos, por ejemplo: strings, listas, tuplas, diccionarios, archivos y otros.

Sintaxis:

<font color=yellow>for indice in objeto_iterable:</font>  
&nbsp;&nbsp;&nbsp;&nbsp;<font color=yellow>acciones</font>

En cada iteracion <span style="color:yellow">indice</span> va tomando cada uno de los elementos de <span style="color:yellow">objeto_iterable</span> y ejecuta acciones. Cuando se terminan los elementos de <span style="color:yellow">objeto_iterable</span> finaliza la iteración.

````python
# Ejemplo: Control de Flujo 18
frase = 'Curso de Python'

for letra in frase:
    print('Letra: ', letra)
````

Al igual que con la iteración <span style="color:limegreen">while</span> en la sentencia <span style="color:limegreen">for</span> también se puede interrumpir la iteración con el uso de la sentencia <span style="color:limegreen">break</span>. 

````python
# Ejemplo: Control de Flujo 19
frase = 'Curso de Python'

for letra in frase:
    if letra == 'd':
        break
    print('Letra: ', letra)
````

Extendiendo lo aprendido en la sentencia <span style="color:limegreen">while</span> también se puede utilizar <span style="color:limegreen">continue</span> y <span style="color:limegreen">else</span> dentro de una iteración <span style="color:limegreen">for</span>.

### <font color=#38bdf8>Iteraciones: Secuencia de números</font>

+ Es muy común utilizar secuencias de números en iteraciones.

+ La función <span style="color:limegreen">range()</span> sirve a este propósito, y devuelve un flujo de números enteros en el rango que se especifica.

    + <span style="color:limegreen">range(comienzo, final, paso)</span>

        + <span style="color:yellow">comienzo</span>: es el valor inicial de la secuencia, es opcional.

        + <span style="color:yellow">final</span>: es obligatorio y la secuencia llega a uno menos de este valor.

        + <span style="color:yellow">paso</span>: es el incremento en la secuencia de números, es opcional y por defecto vale 1.
        
+ La función <span style="color:limegreen">range()</span> devuelve lo que se conoco como un <span style="color:yellow">objeto-iterable</span>.

````python
# Ejemplo: Control de Flujo 20
for i in range(0, 5):
    print(i)
````

````python
# Ejemplo: Control de Flujo 21
for i in range(5):
    print(i)
````

````python
# Ejemplo: Control de Flujo 22
for i in range(1, 8, 2):
    print(i)
````

````python
# Ejemplo: Control de Flujo 23
for i in range(2, -1, -1):
    print(i)
````

````python
# Ejemplo: Control de Flujo 24
for _ in range(15):
    print('Repetir 15 veces.')
````

````python
# Ejemplo: Control de Flujo 25
# Anidamiento
for i in range(1, 5):
    for j in range(1, 5):
        resultado = i * j
        print(f'{i} x {j} = {resultado}')
````

````python
# Ejemplo: Control de Flujo 26
for i in range(5):
    for j in range(5):
        if i == j:
            print('X ', end='')
        if i > j:
            print('D ', end='')
        if i < j:
            print('U ', end='')
    print('')
````
