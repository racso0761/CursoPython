### <font color=#38bdf8>Módulos</font>

+ La <span style="color:yellow">modularidad</span> del código es muy importante.

+ A medida que los proyectos crecen el código también lo hace.

+ Es importante <span style="color:yellow">dividir el proyecto en porciones de código</span> que resuelvan determinadas tareas. Normalmente se dividen de acuerdo a su funcionalidad.

Los <span style="color:magenta">módulos</span> son archivos de texto que contienen código Python (con extensión <span style="color:limegreen">.py</span>) y representan unidades que favorecen la reutilización del código.

### <font color=#38bdf8>Importación de Módulos</font>

+ Para hacer uso de un módulo desarrollado por usted o por alguien mas, se utiliza la sentencia <span style="color:limegreen">import</span> seguida del nombre del módulo sin la extensión <span style="color:limegreen">.py</span>.

Sintaxis:

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen">import nombre_módulo</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen">import nombre_módulo as alias</span>

```python
# Modulos y Paquetes 1
# Importacion de un módulo completo
import math
```

```python
ang = float(input('Ingrese el valor del ángulo [grados]: '))
rad_ang = ang * math.pi / 180
```

```python
print(rad_ang)
```

```python
# Cálculos
seno = math.sin(rad_ang)
coseno = math.cos(rad_ang)
```

```python
print(f'Seno = {seno:.2f}, Coseno = {coseno:.2f}')
```

### <font color=#38bdf8>Buenas Prácticas</font>

+ Cuando se desarrolla código en Python se usan normalmente varios archivos <span style="color:limegreen">.py</span>.

+ Algunos de estos archivos conforman <span style="color:magenta">módulos</span> o <span style="color:magenta">paquetes</span>.

+ Siempre hay un archivo que será el punto de entrada (por donde comienza la ejecución), normalmente se denomina programa principal y posee el nombre <span style="color:limegreen">main.py</span>.

La estructura del programa principal suele ser:

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># Importar las bibliotecas (módulos/paquetes) estándar</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># Importar las bibliotecas (módulos/paquetes) específicas</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># Importar los módulos o paquetes propios</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># CODIGO PROPIO DEL PROGRAMA PRINCIPAL</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen">...</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># FIN CODIGO PROPIO</span>

&nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen">if __name__ == '__main__':</span>

&nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; &nbsp; <span style="color:limegreen"># punto de entrada real</span>

La sentencia de selección del final del programa principal permite diferenciar el código que se ejecutará cuando el archivo se ejecute directamente o cuando se importe como parte de un módulo o paquete.

+ Una <span style="color:magenta">excepción</span> es el bloque de código que se ejecuta cuando se produce un <span style="color:magenta">error</span> en la ejecución de un programa Python.

### <font color=#38bdf8>Gestion de Errores</font>

+ Si se produce un error en la ejecución de un programa y no se captura, se muestra un mensaje de error y se detiene la ejecución.

```python
# Excepciones 1
# Excepcion no capturada
def div_entera(a, b):
    return a // b

print(div_entera(4, 0))
```

+ Lo que se debe hacer es <span style="color:yellow">capturar las excepciones</span> que se produzcan y tomar algún tipo de acción con ellas.

+ Para capturar una excepción se utiliza un bloque <span style="color:limegreen">try-except</span>.

```python
# Excepciones 2
# Excepción Capturada
def div_entera(a, b):
    try:
        return a // b
    except:
        print('Por Favor no dividir por cero')
```

```python
print(div_entera(4, 0))
print(f'Sigo la ejecución del script de Python')
```

+ <span style="color:red">NO</span> es buena práctica utilizar un <span style="color:limegreen">except</span> sin indicar el <span style="color:magenta">tipo de excepción</span>.

### <font color=#38bdf8>Especificando el Tipo de Excepción</font>

+ Existen muchos [tipos de excepciones predefinidas](https://docs.python.org/es/3/tutorial/errors.html) en Python. 

Por ejemplo:

+ <span style="color:limegreen">TypeError</span>: error de tipo de datos

+ <span style="color:limegreen">ZeroDivisionError</span>: División por cero
    
+ <span style="color:limegreen">NameError</span>: errores de nombres.

```python
# Excepciones 3
# Captura de Excepción especifica

def div_entera(x, b):
    try:
        a = x
        #print(h)
        return a // b
    except TypeError:
        print('Verifique el tipo de datos')
    except ZeroDivisionError:
        print('Por Favor, No dividir por cero')
    except NameError:
        print('Ups, algo no funciono')
```

```python
print(div_entera(4, 3))
```

```python
print(div_entera(4, 0))
```

```python
print(div_entera(4, '3'))
```

```python
print(div_entera(28, (2, 3)))
```

```python
print(div_entera(4, 2))
```

+ Consultar [Tipos de Excepciones en Python](https://docs.python.org/es/3/library/exceptions.html) para más infomación.

### <font color=#38bdf8>Otros Casos</font>

+ Para la gestion de Excepciones existe también una clausula <span style="color:limegreen">else</span> para saber que todo ha funcionado correctamente.

+ De igual manera la clausula <span style="color:limegreen">finally</span> se ejecuta siempre independientemente se haya producido o no una excepción.

```python
# Excepciones 4
# Gestion completa de excepciones

valores = [4, 2, 7]
indice = 3

try:
    r = valores[indice]
    print(r)
except IndexError:
    print("Error: indice fuera de rango")
else:
    print('Sus deseos son ordenes para mi')
finally:
    print('Que tenga un buen día')
```

### <font color=#38bdf8>Excepciones Propias</font>

+ Hay ocaciones en que es interesante crear excepciones propias.

+ Para lograr esto se debe crear una clase.

+ Se utiliza la sentencia <span style="color:limegreen">raise</span> para disparar una excepción que podría ser capturada por un bloque <span style="color:limegreen">try-except</span> en un nivel superior.

```python
# Excepciones 5
# Excepciones Propias
class NotIntError(Exception):
    pass

valores = (4, 7, 2.11, 9)
for valor in valores:
    if not isinstance(valor, int):
        raise NotIntError(valor)
```

```python
# Excepciones 6
# Excepciones Propias Mensaje Personalizado
class NotIntError(Exception):
    def __init__(self, message = 'Este módulo solo trabaja con enteros!'):
        self.valor = valor
        self.message = message
        super().__init__(self.message)

    def __str__(self):
        return f'{self.valor} -> {self.message}'
```

```python
raise NotIntError()
```

### <font color=#38bdf8>Aserciones</font>

La sentencia <span style="color:limegreen">assert</span> de Python permite comprobar si se están cumpliendo ciertas condiciones en un programa, y disparar una excepción en caso de que no sea de esa manera.

El formato de esta sentencia es:

<span style="color:limegreen">assert condicion, mensaje_opcional</span>

donde:
+ <span style="color:limegreen">condicion</span>: expresión booleana que se espera que sea <span style="color:limegreen">True</span>.
+ <span style="color:limegreen">mensaje_opcional</span>: texto que se muestra si la condición falla.

<span style="color:yellow">Usos comunes:</span>

+ Verificar supuestos en el código.
+ Detectar errores lógicos durante el desarrollo.
+ No se recomienda usar <span style="color:limegreen">assert</span> para validaciones en producción, ya que puede desactivarse si Python se ejecuta con la opción -O (optimize).

```python
# Ejemplo: Aserciones

resultado = -10
assert resultado > 0, 'El resultado debe ser positivo'
print(resultado)
```

#### En tests unitarios (unittest)

En el módulo estándar <span style="color:limegreen">unittest</span>, se suelen usar métodos como <span style="color:limegreen">assertEqual</span>, <span style="color:limegreen">assertTrue</span>, etc., en lugar de la sentencia <span style="color:limegreen">assert</span> directa.

+ <span style="color:limegreen">assertEqual</span> compara valores esperados.

+ <span style="color:limegreen">assertTrue</sapn> y <span style="color:limegreen">assertFalse</span> verifican condiciones booleanas.

+ <span style="color:limegreen">assertRaises</span> asegura que se lance una excepción esperada. 

```python
import unittest

class TestStringMethods(unittest.TestCase):

    def test_upper(self):
        self.assertEqual('foo'.upper(), 'FOO')

    def test_isupper(self):
        self.assertTrue('FOO'.isupper())
        self.assertFalse('Foo'.isupper())

    def test_split(self):
        s = 'hello world'
        self.assertEqual(s.split(), ['hello', 'world'])
        # Verificar que split falle si el separador no es string
        with self.assertRaises(TypeError):
            s.split(2)

if __name__ == '__main__':
    unittest.main()
```

#### En validación de datos (pytest o depuración)

Con <span style="color:limegreen">pytest</span>, se usa directamente la sentencia <span style="color:limegreen">assert</span>:

```python
def calcular_promedio(valores):
    if not valores:
        return 0
    return sum(valores) / len(valores)

def test_calcular_promedio():
    resultado = calcular_promedio([8, 6, 10])
    assert resultado == 8  # pasa

def test_promedio_lista_vacia():
    resultado = calcular_promedio([])
    assert resultado == 0  # pasa
```

#### Otros Ejemplos

```python
# Comparar cadenas
def test_normalizar_etiqueta():
    resultado = " Curso Python ".strip().lower().replace(" ", "-")
    assert resultado == "curso-python"

# Verificar booleanos
estudiante = {"nombre": "Carla", "nota": 6, "aprobado": True}
assert estudiante["aprobado"] is True

# Verificar pertenencia
resultado = "curso-python"
assert "-" in resultado

# Verificar excepciones
import pytest
def test_nota_invalida():
    with pytest.raises(ValueError):
        int("abc")  # lanza ValueError
```
