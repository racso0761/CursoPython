+ Los Strings son Tipos de datos que se clasifican como secuencias.
Tal como su nombre lo indica son secuencias de caracteres [Unicode](https://es.wikipedia.org/wiki/Unicode).

+ Para crear un string se encierra la secuencia de caracteres en comillas simples (<span style="color:limegreen">'</span>) (apóstrofes), comillas dobles (<span style="color:limegreen">"</span>) o comillas triples (<span style="color:limegreen">'''</span>) (secuencias de apóstrofes).

+ El uso de los limitadores debe ser consistente, es decir si se abre con apóstrofe se cierra con apóstrofe.

<a href="https://home.unicode.org/" style="color:#38bdf8;text-decoration:none;">+info Unicode</a>

```python
# Ejemplo: Strings 1
a = 'Mi primer string' # String de una sola linea
b = "Mi primer string" # String de una sola linea
c = '''Mi primer


    string''' # String multilinea

print(f'a: {a}, b: {b}, \nc: {c}')
print('Tipo de a:', type(a))
```

+ Los strings encerrados en comillas triples están pensados para cadenas de texto multilínea.

+ Los strings encerrados en comillas simples y dobles solo pueden ocupar una línea.

## <font color=#38bdf8>Conversión de Datos</font>

## <font color=#38bdf8>Secuencias de Escape</font>

+ Las secuencias de escape se utilizan para incorporar en un string caracteres especiales.

+ No siempre son necesarias, pero a veces son inevitables.

+ Se especifican las secuencias de escape con una barra invertida (<span style="color:limegreen">\ </span>).

Las secuencias de escape mas comunes utilizadas son:

*Secuencia* | *Descripción*
---|---
<span style="color:limegreen">\n</span> | Caracter nueva línea o salto de línea
<span style="color:limegreen">\t</span> | Caracter tabulador horizontal
<span style="color:limegreen">\ \ </span> | Caracter de la barra invertida
<span style="color:limegreen">\ '</span> | Caracter de comillas simples
<span style="color:limegreen">\ "</span> | Carcater de comillas dobles

+ Cuando se desea que el texto dentro de un string aparezca tal y como se ha escrito, con todos los caracteres especiales se utiliza un <span style="color:yellow">"raw-data"</span> y se especifica anteponiendo una <span style="color:limegreen">r</span> al string.

```python
# Ejemplo: Strings 4
print("Hoy\t debemos\t aprender\t Python\nMañana también\n")
```

```python
print(r'Este    es un \t ejemplo de un \t raw-data\n')
```

```python
print('Este es un ejemplo de como "evitar" las secuencias de escape')
```

## <font color=#38bdf8>Operador de Slicing</font>

+ Este operador permite extraer una porción de un string (es decir, un <span style="color:yellow">substring</span>).

+ El formato general de este operador es:

&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;&nbsp;<span style="color:yellow">nombre_string[[comienzo ]:final[: paso ]]</span>

donde:
+ <span style="color:yellow">comienzo</span>: especifica el indice inicial (opcional)

+ <span style="color:yellow">final</span>: especifica el indice final, no incluído (opcional)

+ <span style="color:yellow">paso</span>: especifica el salto entre elementos (opcional)

```python
# Ejemplo: Strings 8
a = 'Agua pasada no mueve molino' # Un String
```

```python
b = a[5:11]
print('Substring 1:', b)
```

```python
b = a[0:20:2]
print('Substring 2:', b)
```

```python
b = a[:]
print('Substring 3:', b)
```

```python
b = a[:8]
print('Substring 4:', b)
```

```python
b = a[8:]
print('Substring 5:', b)
```

```python
b = a[::-1]
print('Substring 6:', b)
```

```python
b = a[5:11] + " por Jose"
print('Substring:', b)
```

## <font color=#38bdf8>Algunas Funciones, Métodos y Operadores útiles con Strings</font>

+ <span style="color:yellow">len()</span>: Permite obtener la longitud de un string

+ <span style="color:yellow">in</span>: Permite comprobar que un string sea substring de otro. Siempre entrega por resultado un tipo <span style="color:limegreen">bool</span>.

+ <span style="color:yellow">split()</span>: Permite dividir un string por medio de algún caracter separador.

+ <span style="color:yellow">strip()</span>: Permite limpiar un string de los caracteres de relleno al comienzo y final.

    + <span style="color:yellow">lstrip()</span>: limpieza por izquierda.

    + <span style="color:yellow">rstrip()</span>: limpieza por derecha.


+ Búsquedas:

    + <span style="color:yellow">startswith()</span>: Evalúa si un string comienza con un substring determinado.

    + <span style="color:yellow">endswith()</span>: Evalúa si un string finaliza con un substring determinado.

    + <span style="color:yellow">find()</span>: Encuentra la primera ocurrencia de un substring dentro de un string.

    + <span style="color:yellow">index()</span>: Idem a <span style="color:yellow">find()</span>.

    + <span style="color:yellow">count()</span>: Cuenta la cantidad de ocurrencias de un substring.

    + <span style="color:yellow">replace()</span>: Reemplaza elementos en un string.

    + <span style="color:yellow">capitalize()</span>: Coloca en mayúsculas la primer letra de la primer palabra.

    + <span style="color:yellow">title()</span>: Coloca en mayúsculas la primer letra de cada palabra dentro del string.

    + <span style="color:yellow">upper()</span>: Coloca todo el string en mayúsculas.

    + <span style="color:yellow">lower()</span>: Coloca todo el string en minúsculas.

    + <span style="color:yellow">swapcase()</span>: intercambia mayúsculas y minúsculas.

    + <span style="color:yellow">ord()</span>: permite obtener el código (decimal) de un caracter a partir de su representación.
    
    + <span style="color:yellow">char()</span>: permite representar un caracter a partir de su código <span style="color:magenta">Unicode</span>.

## <font color=#38bdf8>Strings de Formato</font>

En Python, existen varias opciones para dar formato a los strings y de esta manera lograr que la información se pueda presentar adecuadamente.

Estilos:
*Nombre* | *Símbolo* | *Soporte*
---|---|---
Estilo Antiguo | <span style="color:limegreen">%</span> | >= Python2
Estilo con "format" | <span style="color:limegreen">.format()</span> | >= Python 2.6
"f-strings" | <span style="color:limegreen">f' '</span> | >= Python3.6

[Documentación de Strings de Formato](https://docs.python.org/es/3/tutorial/inputoutput.html)

### <font color=#38bdf8>Estilo Antiguo (<span style="color:limegreen">%</span>)</font>

Sintaxis básica: <span style="color:yellow">"Este es un string %s" % nombre_de_la_variable_con_el_string_a_mostrar</span>

**Ejemplo 14**

```python
# Ejemplo: Strings 14
a = 'Carla Schmadke'
print("El nombre del estudiante es: %20s" % a)
```

```python
b = 12
print("El valor entero es: %+020d" % b)
```

```python
c = 3.14159265358 # Numero PI
print("El numero pi es %5.2f" % c)
```

### <font color=#38bdf8>El método <b>str.format()</b></font>

Sintáxis básica: <span style="color:orange">"String de formato { }" . format(argumentos)</span>

+ donde <span style="color:limegreen">{ }</span>: es el campo de formato

#### <font color=#38bdf8>Sintáxis del campo de formato:</font>

[{campo ! marca de conversión: formato}](https://docs.python.org/es/3/library/string.html#formatstrings)

+ <span style="color:yellow">campo</span> es un valor que identifica el número de argumento del método format.

+ <span style="color:yellow">:formato</span>	->	<span style="color:yellow">[[ relleno ] alineamiento ][ signo ][ # ][ 0 ][ ancho ][ , precisión ][ código de tipo ]</span>

	+ <span style="color:yellow">[ ]</span>: representa un elemento opcional

	+ <span style="color:yellow">relleno</span>: caracter con el que se rellenan los espacios que requieren un alineamiento (por defecto espacios en blanco).

	+ <span style="color:yellow">alineamiento</span>: caracter <span style="color:yellow"><</span> para alinear a la izquierda, <span style="color:yellow">></span> para alinear a la derecha (valor por defecto), <span style="color:yellow">^</span> para centrar en el espacio disponible.

	+ <span style="color:yellow">signo</span>: <span style="color:yellow">+</span> para forzar la aparición de un signo, <span style="color:yellow">-</span> para indicar que solo aparece el signo en los negativos (valor por defecto), espacio en blanco para indicar que los numeros positivos deben ir precedidos por un espacio en blanco.

	+ <span style="color:yellow">#</span>: Si aparece, los enteros que se muestran en binario, octal o hexadecimal irán precedidos de <span style="color:yellow">0b</span>, <span style="color:yellow">0o</span> y <span style="color:yellow">0x</span>.

	+ <span style="color:yellow">0</span>: Si aparece, se usa el caracter <span style="color:yellow">0</span> para sustituir los espacio en blanco.
	
	+ <span style="color:yellow">ancho</span>: es un numero entero que indica cuantos caracteres se desea que ocupe el valor representado (como mínimo).

	+ <span style="color:yellow">.precisión</span>: numero de decimales con que se desean representar un numero real.

	+ <span style="color:yellow">código de tipo</span>: caracter que indica el tipo de representación que se desea: <span style="color:yellow">d</span>, <span style="color:yellow">f</span>, <span style="color:yellow">s</span>.

```python
# Ejemplo: Strings 15
# Ejemplos de uso de string de formato con .format()

anio = 23
peso = 65.4976
nombre = 'Juan Sinnombre'
```

```python
print("El estudiante: {0:30s} tiene {1:2d} años y pesa {2:5.2f}"\
      .format(nombre, anio, peso))
```

```python
lista = ['Juan', 23, 34567865, 'Av. Libertador 1234']
print(lista)
```

```python
print('El joven {0} de {1} años y documento {2}, vive en {3}'.format(lista[0], lista[1], lista[2], lista[3]))
```

```python
print('El joven {0[0]} de {0[1]} años y documento {0[2]}, vive en {0[3]}'.format(lista))
```

### <font color=#38bdf8>f-strings</font>

+ Es el estandar por defecto que se utiliza en la actualidad para los strings de formato.

+ para indicar que se trata de un <span style="color:magenta">f-string</span> se debe anteponer al mismo la letra <span style="color:yellow">f</span> e incluir las variables a mostrar en un campo <span style="color:yellow">{}</span> al igual que con el estilo format.

[+ info](https://docs.python.org/3/reference/lexical_analysis.html#f-strings)

```python
# Ejemplo: Strings 16
# Ejemplos de uso de string de formato con .format()

anio = 23
peso = 65.4976
nombre = 'Juan Sinnombre'
```

```python
print(f'El estudiante: {nombre:>30s} tiene {anio:03d} años y pesa {peso:5.2f}')
```

```python
legajo = f'El estudiante: {nombre=:30s} tiene {anio:2d} años y pesa {peso:5.2f}'
print(legajo)
```

## <font color=#38bdf8>Comparación de Strings</font>

Se utilizan con este propósito los operadores relacionales.

Operador | Descripción
---|---
<span style="color:yellow">==</span> | igual a
<span style="color:yellow">!=</span> | distinto a
<span style="color:yellow"><</span> | menor que

Siempre se compara en términos Lexicográficos.

## <font color=#38bdf8>Lectura de Datos desde Teclado</font>

Para lograr la interacción de un usuario con un programa, es decir ingresar datos por teclado se usa la función <span style="color:limegreen">input()</span>.

Caracteristicas:

+ La funcion siempre entrega un string como resultado (<span style="color:limegreen">str</span>).

+ Siempre se debe utilizar un conversión explicita de datos cuando se requiera ingresar algún tipo numérico.

+ Admite un string como argumento que se usa para especificar un mensaje.

```python
# Ejemplo: Strings 20
a = input()
print('El tipo del dato ingresado es:', type(a))
print('Y su valor es:', a)
```

```python
b = input("Ingrese un valor: ")
print('El tipo del dato ingresado es:', type(b))
print('Su valor es:', b*2)
```

```python
c = int(input("Ingrese un entero: "))
print('El tipo del dato ingresado es:', type(c))
print('Su valor es:', c)
```

```python
d = float(input("Ingrese un real: "))
print('El tipo del dato ingresado es:', type(d))
print('Su valor es:', d)
```
