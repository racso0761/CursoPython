# El Paquete Pandas

[*Pandas*](https://pandas.pydata.org/docs/) es un paquete open-source que proporciona una forma sencilla y potente de trabajar con estructuras de datos a través de múltiples herramientas para su análisis.

La manera mas sencilla de trabajar con este paquete es importar el paquete completo:

<span style="color:limegreen">import pandas as pd</span>

En <span style="color:yellow">Pandas</span> existen dos estructuras de datos sobre las que se basan todas las operaciones:

+ <span style="color:yellow">Series temporales</span>
+ <span style="color:yellow">Dataframes</span>

## Series Temporales

+ Se puede hacer una analogía de una <span style="color:yellow">serie</span> con un <span style="color:yellow">ndarray</span> de <span style="color:yellow">Numpy</span> en el que cada valor tiene asignada una etiqueta (<span style="color:yellow">índice</span>) y además admite un título (<span style="color:yellow">nombre</span>) opcional.

![Estructura de un Dataframe](./Figuras/SerieTemporal.png)

### Creación de una Serie

+ Se utiliza principalmente la función <span style="color:limegreen">pd.Series()</span> con diferentes modalidades.

```python
# Ejemplo: Series 1
import pandas as pd #type:ignore
```

```python
# Creación usando Listas
serie1 = pd.Series([8, 12, 23], name = 'Enteros')
print(f'Serie 1: \n{serie1}')
print('-------------')
```

```python
# Creación especificando un índice personalizado
serie2 = pd.Series(range(1, 4), index = ['a', 'b', 'c'], name = 'Enteros')
print(f'Serie 2: \n{serie2}')
print('-------------')
```

```python
# Creación especificando un diccionario
items = {'a': 8.2, 'b': 19.45, 'c': 45.76}
serie3 = pd.Series(items, name = 'Longitudes')
print(f'Serie 3: \n{serie3}')
print('-------------')
```

### Atributos de una Serie

Las <span style="color:limegreen">series</span> en <span style="color:yellow">Pandas</span> tienen una gran cantidad de atributos.

Algunos de ellos son:

+ Los Indices de la serie: <span style="color:limegreen">serie.index</span>

+ Los Valores de la serie: <span style="color:limegreen">serie.values</span>

+ Los Tipo de datos de la serie: <span style="color:limegreen">serie.dtype</span>

+ El Nombre de la serie: <span style="color:limegreen">serie.name</span>

+ La Memoria ocupada por la serie: <span style="color:limegreen">serie.nbytes</span>

+ El Número de registros en la serie (cantidad de datos): <span style="color:limegreen">serie.size</span>

```python
# Ejemplo: Series 2
import pandas as pd
# Diccionario con nombres de empresas y cantidad de empleados
datos = {'Apple': 147_000, 'Samsung': 267_937, 'Google': 135_301,
         'Microsoft': 163_000, 'Huawei': 197_000, 'Dell': 158_000,
         'Facebook': 58_604, 'Foxconn': 878_429, 'Sony': 109_700
        }
```

```python
# Creación de la Serie
empleados = pd.Series(datos, name = 'Empleados')
print(type(empleados))
print('------------')

print(f'Serie: \n{empleados}')
```

```python
# Indice
print(f'Indices: \n{empleados.index}\n')
print('------------')
```

```python
# Valores
print(f'Valores: \n{empleados.values}\n')
print('------------')
```

```python
# Tipo
print(f'Tipo de datos: {empleados.dtype}')
print('------------')
```

```python
# Nombre
print(f'Nombre: {empleados.name}')
print('------------')
```

```python
# Memoria en bytes
print(f'Memoria en bytes: {empleados.nbytes}')
print('------------')
```

```python
# Tamaño (numero de registros)
print(f'Numero de Registros: {empleados.size}')
print('------------')
```

### Selección de Registros

La selección de los datos de la <span style="color:yellow">Serie</span> se puede hacer de múltiples formas:

+ Usando indexado numérico: con el método <span style="color:limegreen">.iloc[ ]</span>

+ Usando indexado slicing: con el método <span style="color:limegreen">.iloc[ : : ]</span>

+ Usando etiquetas: con el método <span style="color:limegreen">.loc[ ] con etiquetas</span> o <span style="color:limegreen">.loc[ : : ]</span>

+ Accediendo a la parte inicial o final de la serie: con los métodos <span style="color:limegreen">.head()</span> o <span style="color:limegreen">.tail()</span>

```python
# Ejemplo: Series 3
print(f'Serie Empleados:\n{empleados}')
print('------------')
```

```python
# Indexado numérico
print(f'Empleados en posición 0: {empleados.iloc[0]}')
print(f'Empleados en la última posición: {empleados.iloc[-1]}')
print('------------')
```

```python
# Indexado slicing
print(f'Empleados entre las posiciones 2 y 5:\n{empleados[2:5]}\n')
print(f'Empleados entre las posiciones 1 a 6 cada 2: \n{empleados[1:6:2]}\n')
print(f'Empleados entre las posiciones 1 a 6 cada 2 con iloc: \n{empleados.iloc[1:6:2]}') # iloc es un alias, mejor esta opción
print('------------')
```

```python
# Usando etiquetas
print(f'Empleados de Apple: {empleados["Apple"]}\n')
print(f'Empleados entre Apple y Huawei: \n{empleados["Google" : "Huawei"]}\n')
print(f'Empleados entre Apple y Huawei cada 2 empress: \n{empleados["Google": "Huawei" : 2]}\n')
print(f'Empleados entre Apple y Huawei cada 2 empresas con iloc: \n{empleados.loc["Google" : "Huawei" : 2]}') # se debe usar esta opción
print('------------')
```

```python
# Fragmentos de comienzo y fin
print(f'Primeros 3 registros de la Serie Empleados: \n{empleados.head(3)}\n') # Primeros 3 registros
print(f'Ultimos 3 registros de la Serie Empleados: \n{empleados.tail(3)}') # Ultimos 3 registros
```

### Operaciones con Series

Teniendo en cuenta que una <span style="color:yellow">Serie</span> contiene valores en un formato similar a los <span style="color:magenta">ndarray</span> de <span style="color:yellow">Numpy</span> entonces se puede concluir que las operaciones definidas para los <span style="color:yellow">ndarray</span> se pueden utilizar también con las series temporales.

+ Operaciones Lógicas sobre Series

+ Ordenación de una Serie

+ Contar valores de una Serie

+ Operaciones Aritméticas

    + Operaciones con escalares

    + Operaciones entre Series

+ Funciones Estadísticas

+ Máximos y Mínimos

+ Exportación de Series

```python
# Ejemplo: Series 4
print(f'Serie Empleados: \n{empleados}')
print('---------------')
```

```python
# Operaciones lógicas: filtrado
print(f'Operacion de Filtrado (numero de emplados > 200000):\n{empleados[empleados > 200_000]}')
print('---------------')
```

```python
# Ordenacion
#     Admiten el argumento ascending como True o False
#     Admiten el argumento inplace para indicar si modifica o no la serie original

# Ordenacion por valor
print(f'Serie Ordenada por Valor: \n{empleados.sort_values(ascending = True)}')
print('---------------')
```

```python
# Ordenacion por indice
print(f'Serie Ordenada por Indice: \n{empleados.sort_index()}')
print('---------------')
```

```python
# Contar valores (tabla de frecuencias)
print(f'Tabla de Frecuencias: \n{empleados.value_counts()}')
# Numero de valores únicos en la serie
print(f'Numero de valores unicos en la Serie: \n{empleados.nunique()}')
# Numero de valores distinto de cero
print(f'Numero de valores distinto de cero en la Serie: \n{empleados.count()}')
print('---------------')
```

## DataFrames

Un <span style="color:yellow">DataFrame</span> es una estructura tabular compuesta por varias <span style="color:yellow">Series Temporales</span>. Se trata del tipo de datos fundamental en <span style="color:yellow">Pandas</span> y sobre el que actúan la mayoría de operaciones que se pueden realizar.

![Estructura de un Dataframe](./Figuras/dataframe_1.png)

### Creación de un DataFrame

Existen mútliles formas de crear un <span style="color:yellow">DataFrame</span> en <span style="color:yellow">Pandas</span>.

+ A partir de un <span style="color:yellow">diccionario de listas</span>

+ A partir de una <span style="color:yellow">lista de diccionarios</span>

+ A partir de una <span style="color:yellow">lista de listas</span>

+ A partir de <span style="color:yellow">Series</span>

```python
# Ejemplo: Dataframe 1
import pandas as pd

# Desde diccionarios de listas
#  Cada elemento del diccionario se convierte en una columna
#  La clave es el nombre de la serie.
datos = {'A':[1, 2, 3], 'B':[4, 5, 6]}
q = pd.DataFrame(datos)
print(f'Dataframe desde Diccionario de Listas: \n{q}')
print('-------------')
```

```python
# Desde lista de diccionarios
#   Cada elemento de la lista se convierte en una fila.
#   Las claves del diccionario son los nombres de las columnas
datos = [{'A':1, 'B': 2, 'C': 3}, {'A':4, 'B': 5, 'C': 6}]
p = pd.DataFrame(datos)
print(f'Dataframe desde Lista de Diccionarios: \n{p}')
print('-------------')
```

```python
# Desde lista de listas
#   Cada elemento de la lista se convierte en una fila
#   Los nombres de las columnas se pueden pasar como un argumento opcional
datos = [[1, 2], [3, 4], [5, 6]]
r = pd.DataFrame(datos, columns = ['A', 'B']) # Nombre de cada columna
print(f'Dataframe desde Lista de Listas: \n{r}')
print('-------------')
```

```python
# Desde Series
datos = {'Apple': 147000, 'Samsung': 267937, 'Google': 135301,
         'Microsoft': 163000, 'Huawei': 197000, 'Dell': 158000,
         'Facebook': 58604, 'Foxconn': 878429, 'Sony': 109700
         }

empleados = pd.Series(datos, name='Empleados')
datos_i = {'Apple': 274515, 'Samsung': 200734, 'Google': 182527,
            'Microsoft': 143015, 'Huawei': 129184, 'Dell': 92224,
            'Facebook': 85965, 'Foxconn': 181945, 'Sony': 84893
            }
ingresos = pd.Series(datos_i)

s = pd.DataFrame({'Empleados': empleados, 'Ingresos': ingresos})
print(f'Dataframe desde Series: \n{s}')
print('-------------')
```

### Gestión de Indices

+ Cuando se crea un <span style="color:yellow">Dataframe</span>, <span style="color:yellow">Pandas</span> autocompleta el índice con un valor entero autoincremental comenzando desde <span style="color:yellow">0</span>.

+ Si se desea que alguna columna se convierta en índice se utiliza el método <span style="color:limegreen">.set_index(columna)</span>.

+ También se puede agregar un argumento para especificar los valores deseados en el índice.

```python
# Ejemplo: Dataframe 2

# valores autoincrementales
q = pd.DataFrame({'A': [1, 2], 'B': [3, 4]})
print(f'Dataframe: \n{q}\n')
print('----------')
```

```python
# Conversion de una columna en indice
print(f'Dataframe con columna A como indice: \n{q.set_index("A")}')
print('----------')
```

```python
# Agregando índices como argumentos
q = pd.DataFrame({'A': [1, 2], 'B': [3, 4]}, index=['R1', 'R2'])
print(f'Dataframe con valores de indices arbitrarios: \n{q}')
print('----------')
```

### Acceso a un DataFrame

Se debe conocer la estructura interna de un <span style="color:yellow">DataFrame</span> para un adecuado manejo de la información que contiene:

![Componentes de un DataFrame](./Figuras/DataFrame_forma.png)

#### Acceso a Filas

+ Acceso a filas mediante índice numérico: <span style="color:limegreen">.iloc[numero]</span>.
+ Acceso a filas mediante la etiqueta del registro: <span style="color:limegreen">.loc['etiqueta']</span>.

#### Acceso a Columnas

En este caso se obtiene una Serie:

+ Se accede con el operador de indexación (<span style="color:limegreen">[ ]</span>) como en un diccionario.
+ Se pueden seleccionar varias columnas a la vez pasando una lista.
+ Esta técnica se puede utilizar para reordenar las columnas.

#### Acceso a Filas y Columnas

+ Se pueden mezclar los accesos anteriores para seleccionar datos de manera precisa.

```python
# Ejemplo: DataFrame 6
# Lee el archivo y crea el DataFrame
df = pd.read_csv('Archivos/empresas.csv', delimiter=',')
df = df.set_index('Company')
print(f'Dataframe Empresas: \n{df}')
```

```python
# Acceso a Filas
# Indice numerico
print(f'Acceso a filas mediante índice numérico (primera): \n{df.iloc[0]}\n')
print(f'Acceso a filas mediante índice numérico (ultima): \n{df.iloc[-1]}\n')
print(f'Acceso a filas mediante índice numérico (entre la 3 y la 5): \n{df.iloc[3 : 5]}\n')
print(f'Acceso a filas mediante índice numérico (todas de 5 en 5): \n{df.iloc[: : 5]}')
print('--------------------')
```

```python
# Etiquetas
print(f'Acceso a filas mediante etiquetas (primera): \n{df.loc["Apple"]}\n')
print(f'Acceso a filas mediante etiquetas (la numero 12): \n{df.loc["IBM"]}\n')
print(f'Acceso a filas mediante etiquetas (desde Sony hasta Intel): \n{df.loc["Sony" : "Intel"]}')
print('--------------------')
```

## Manejando Categorías

Hasta ahora se han visto tipos de datos numéricos, cadenas de texto y fechas. ¿Pero qué ocurre con las categorías?

Las categorías pueden ser tanto datos numéricos como textuales, con la característica de tener un número discreto (relativamente pequeño) de elementos y, en ciertas ocasiones, un orden preestablecido. Ejemplos de variables categóricas son: género, idioma, meses del año, color de ojos, nivel de estudios, grupo sanguíneo, valoración, etc.


### Tipos de Datos

Cuando se trabaja con estadísticas, es importante reconocer los diferentes tipos de datos: <span style="color:yellow">numéricos</span> (discretos y continuos), <span style="color:yellow">categóricos</span> y <span style="color:yellow">ordinales</span>. Los datos no son más que observaciones del mundo en que vivimos, por tanto, los mismos pueden venir en diferentes formas, no solo numérica. Por ejemplo, se le puede preguntar a un grupo de personas ¿cuántas mascotas tienen? y pueden responder: 0, 1, 2, 4, 3, 8; esta información por sí misma puede ser útil, pero para el análisis de las mascotas, podría servir también otro tipo de información, como por ejemplo el género de cada uno de esas personas; de esta forma se obtendría la siguiente información: hombre, mujer, mujer, mujer, hombre, mujer. Como se ve, se puede incluir a los datos dentro de tres categorías fundamentales: **datos cuantitativos** o **numéricos**, **datos cualitativos** o **categóricos** y **datos ordinales**.

### Datos Cuantitativos

Los datos cuantitativos se representan por números; estos números serán significativos si representan la medida o la cantidad observada de cierta característica. Dentro de esta categoría se puede encontrar por ejemplo: cantidades de dólares, cuentas, tamaños, número de empleados, y kilómetros por hora. Con los datos cuantitativos, se puede hacer todo tipo de tareas de procesamiento de datos numéricos, tales como sumarlos, calcular promedios, o medir su variabilidad. Asimismo, se puede dividir a los datos cuantitativos en *discretos* y *continuos*, dependiendo de los valores potencialmente observables.

+ Los datos discretos solo van a poder asumir un valor de una lista de números específicos. Representan ítems que pueden ser contados; todos sus posibles valores pueden ser listados. Suele ser relativamente fácil trabajar con este tipo de dato.

+ Los datos continuos representan mediciones; sus posibles valores no pueden ser contados y sólo pueden ser descritos usando intervalos en la recta de los números reales. Por ejemplo, la cantidad de kilómetros recorridos no puede ser medida con exactitud, puede ser que hayamos recorrido 1.7 km o 1.6987 km; en cualquier medida que tomemos del mundo real, siempre pueden haber pequeñas o grandes variaciones. Generalmente, los datos continuos se suelen redondear a un número fijo de decimales para facilitar su manipulación.

### Datos Cualitativos

En estadística, una variable categórica (también llamada variable cualitativa) es una variable que puede tomar uno de un número limitado, y normalmente fijo, de valores posibles, asignando cada valor individual u otra unidad de observación a un grupo particular o categoría nominal sobre la base de alguna propiedad cualitativa. En informática y algunas ramas de las matemáticas, las variables categóricas se denominan enumeraciones o tipos enumerados. Comúnmente, cada uno de los posibles valores de una variable categórica se denomina nivel. La distribución de probabilidad asociada a una variable aleatoria categórica se denomina distribución categórica.

Los datos categóricos son el tipo de datos estadísticos que consisten en variables categóricas o en datos que se han convertido a esa forma, por ejemplo como datos agrupados. Más concretamente, los datos categóricos pueden proceder de observaciones realizadas de datos cualitativos que se resumen como recuentos o tabulaciones cruzadas, o de observaciones de datos cuantitativos agrupados dentro de intervalos determinados. A menudo, los datos puramente categóricos se resumen en forma de tabla de contingencia. Sin embargo, sobre todo al considerar el análisis de datos, es habitual utilizar el término "datos categóricos" para aplicarlos a conjuntos de datos que, si bien contienen algunas variables categóricas, también pueden contener variables no categóricas.

Una variable categórica que puede tomar exactamente dos valores se denomina variable binaria o variable dicotómica. Las variables categóricas con más de dos valores posibles se denominan variables politómicas; a menudo se supone que las variables categóricas son politómicas a menos que se especifique lo contrario. La discretización consiste en tratar los datos continuos como si fueran categóricos. La dicotomización consiste en tratar los datos continuos o las variables politómicas como si fueran variables binarias. El análisis de regresión suele tratar la pertenencia a una categoría con una o más variables ficticias cuantitativas.

Entonces, si los datos dicen en cual de determinadas categorías no numéricas los ítems van a caer, entonces se está hablando de datos cualitativos o categóricos; ya que los mismos van a representar determinada cualidad que los ítems poseen. Dentro de esta categoría se pueden encontrar datos como: el sexo de una persona, el estado civil, la ciudad de nacimineto, o los tipos de películas que le gustan. Los datos categóricos pueden tomar valores numéricos (por ejemplo, "1" para indicar "masculino" y "2" para indicar "femenino"), pero esos números no tienen un sentido matemático.

### Datos Ordinales

Una categoría intermedia entre los dos tipos de datos anteriores, son los datos ordinales. En este tipo de datos, va a existir un orden significativo, se puede clasificar un primero, segundo, tercero, etc. es decir, se puede establecer un ranking para estos datos, el cual posiblemente luego tenga un rol importante en la etapa de análisis. Los datos se dividen en categorías, pero los números colocados en cada categoría tienen un significado. Por ejemplo, la calificación de un restaurante en una escala de 0 (bajo) a 5 (más alta) estrellas representa datos ordinales. Los datos ordinales son a menudo tratados como datos categóricos, en el sentido que se suelen agrupar y ordenar. Sin embargo, a diferencia de los datos categóricos, los números sí tienen un significado matemático.

Pandas facilita el tratamiento de datos categóricos mediante un tipo específico denominado <span style="color:yellow">Categorical</span>. Siguiendo con el <span style="color:yellow">dataset</span> de empresas tecnológicas, se va a añadir el continente al que pertenece cada empresa. En primera instancia mediante valores de texto:

```python
# Ejemplo Dataframe 18
import pandas as pd
df = pd.read_csv('./Archivos/empresas.csv', delimiter=',')
df = df.set_index('Company')

df['Continent'] = ['America', 'Asia', 'America', 'Asia',
                  'America', 'Asia', 'America', 'America',
                  'Asia', 'Asia', 'America', 'America',
                  'Asia', 'Asia', 'Asia', 'America', 'Asia'
                  ]
print(df["Continent"])
```

Ahora se puede convertir la columna <span style="color:yellow">Continent</span> a tipo <span style="color:yellow">Categorical</span>.

```python
df["Continent"].astype("category")
```

En este caso, al ser una conversión <span style="color:yellow">automática</span>, las categorías no han incluido ningún tipo de orden. Pero suponga que se desea establecer un orden para las categorías de continentes basadas, por ejemplo, en su población, es decir: Asia, África, Europa, América, Australia

```python
from pandas.api.types import CategoricalDtype

continentes = ('Asia', 'Africa', 'Europa', 'America', 'Australia')

categ_cont = CategoricalDtype(categories = continentes, ordered=True)
df['Continent'] = df["Continent"].astype(categ_cont)
print(f'Categorías Ordenadas: \n{df["Continent"].astype(categ_cont)}\n')
```

El propósito de trabajar con categorías ordenadas es que permite entre otras cosas operaciones como las siguientes:

```python
print(df['Continent'].min())
print('-----------------')
print(df['Continent'].max())
print('-----------------')
print(df['Continent'].sort_values())
```

> En condiciones normales (categorías sin ordenar) el mínimo hubiera sido <span style="color:yellow">America</span> y el máximo hubiera sido <span style="color:yellow">Asia</span> ya que se habrían ordenado alfabeticamente.

## Agrupando Datos en Dataframes

El método de pandas <span style="color:limegreen">.DataFrame.groupby()</span> tiene una funcionalidad en la que es necesario indicar el eje que contiene el criterio por el que se va a realizar el agrupamiento. Comencemos con un ejemplo sencillo. Partimos del siguiente Dataframe:


```python
import pandas as pd
ventas = pd.DataFrame({
	"Productos" : ["A", "B", "C", "B", "A", "A"],
	"Ventas" : [6, 2, 1, 4, 5, 2]	
})
print(ventas)
```

En el caso de los Dataframes, el parámetro <span style="color:limegreen">by</span> puede hacer referencia a una función, a un diccionario, a una etiqueta o a una lista de etiquetas. Si pasamos simplemente la etiqueta <span style="color:yellow">'Producto'</span> para indicar que el agrupamiento se realice según los valores de esta columna, se tiene:

```python
ventas.groupby(by = "Productos").mean()
```

Si se quisiéra realizar el agrupamiento por más de una columna, bastaría con pasar como argumento una lista con las etiquetas en cuestión. Por ejemplo, considere el siguiente caso en el que se tiene las ventas clasificadas por categoría y producto:

```python
ventas = pd.DataFrame({
	"Categoria" : [1, 2, 1, 1, 2, 1],
	"Productos" : ["A", "B", "C", "B", "A", "A"],
	"Ventas" : [6, 2, 1, 4, 5, 2]	
})
print(ventas)
```

Si se aplica ahora el método <span style="color:limegreen">groupby</span> con el argumento <span style="color:limegreen">by = ["Categoría", "Productos"]</span>, se tiene:

```python
print(ventas.groupby(by = ["Categoria", "Productos"]).mean())
```

Este ejemplo tiene pocos datos para ser significativo, pero aun así es posible ver que el método ha agrupado todas las ventas según la combinación de categoría y producto, y se ha calculado el valor medio. Por ejemplo, hay dos ventas de categoría 1 y producto A, de valores 6 y 2. La media, tal y como se muestra es de 4.

También se podría usar el parámetro <span style="color:yellow">level</span>. En el caso de estar trabajando con Dataframes con índices no jerárquicos, basta pasar como valor para este argumento el 0 para que el agrupamiento se realice según las etiquetas del índice. Por ejemplo, considere el siguiente Dataframe:

```python
ventas = pd.DataFrame({
	"Ventas" : [6, 2, 1, 4, 5, 2]	
}, index = ["A", "B", "C", "B", "A", "A"])
print(ventas)
```

Si se ejecuta el método con el argumento <span style="color:limegreen">level = 0</span>, se obtendría el siguiente resultado:

```python
print(ventas.groupby(level = 0).mean())
```

Las operaciones de agregado son muy utilizadas y nos permiten extraer información relevante, que, a simple vista, quizás no sea tan evidente.

En el siguiente ejemplo se calcula la suma de los ingresos de las empresas, agrupados por país:

```python
df = pd.read_csv('./Archivos/empresas.csv', delimiter=',')
print(df.groupby('Pais')['Ingresos'].sum())
```

También es posible realizar un agrupamiento en varios niveles. En el siguiente ejemplo se consiguen los datos agrupados por <span style="color:yellow">Pais</span> y por <span style="color:yellow">Ciudad</span>:

```python
print(df.groupby(['Pais', 'Ciudad'])['Ingresos'].sum())
```

Cuando se realiza un agrupamiento por varias columnas, el resultado contiene un índice de múltiples niveles. Se puede aplanar el Dataframe utilizando <span style="color:limegreen">.unstack()</span>.

Incluso se pueden aplicar distintas funciones de agregación a cada columna. Suponga que se necesita calcular la media de los ingresos y la mediana del número de empleados/as, con las empresas agrupadas por país (Se utiliza la función <span style="color:limegreen">agg()</span> pasando un diccionario cuyas claves son nombres de columnas y cuyos valores son funciones a aplicar):

### Tablas de Frecuencias

Para hacer una imagen de los datos, lo primero que se tiene que hacer es agruparlos. Al armar diferentes grupos se puede acercar la comprensión de los datos. La idea es agrupar las cosas que parecen ir juntas, para poder ver como se distribuyen a través de las diferentes categorías. Para los datos categóricos, agrupar es fácil; simplemente se debe contar el número de ítems que corresponden a cada categoría y apilarlos. Una forma en la que se puede agrupar el dataset del Titanic es contando las diferentes clases de pasajeros. PSe pueden organizar estos conteos en una tabla de frecuencia, que registra los totales y los nombres de las categorías utilizando la función <span style="color:limegreen">value_counts</span> que proporciona Pandas del siguiente modo:

```python
# tabla de frecuencia de clases de pasajeros
print(titanic['class'].value_counts())
```

Contar las cantidad de apariciones de cada categoría puede ser útil, pero a veces puede resultar más útil saber la fracción o proporción de los datos de cada categoría, así que se podría entonces dividir los recuentos por el total de casos para obtener los porcentajes que representa cada categoría.

Una tabla de frecuencia relativa muestra los porcentajes, en lugar de los recuentos de los valores en cada categoría. Ambos tipos de tablas muestran cómo los casos se distribuyen a través de las categorías. De esta manera, ellas describen la distribución de una variable categórica, ya que enumeran las posibles categorías y nos dicen con qué frecuencia se produce cada una de ellas.

```python
# tabla de frecuencias relativa de pasajeros
print(100 * titanic['class'].value_counts() / len(titanic['class']))
```

### Gráficos de Torta y de Barras

Ahora que ya se conocen las tablas de frecuencia se está en condiciones de crear visualizaciones que realmente den una imagen de los datos, sus propiedades y sus relaciones. En este punto, se debe tener mucho cuidado, ya que una mala visualización puede llegar a distorsionar la comprensión, en lugar de ayudarla. Las mejores visualizaciones de datos siguen un principio fundamental llamado el principio del área. Este principio dice que el área ocupada por cada parte del gráfico se debe corresponder con la magnitud del valor que representa. Violaciones del principio de área son una forma común de mentir con estadísticas. Dos gráficos útiles que se pueden utilizar para representar datos y que cumplen con este principio son el gráfico de barras y el gráfico de torta.

#### Gráfico de Barras

El gráfico de barras ayuda a obtener una impresión visual más precisa de la distribución de los datos. La altura de cada barra muestra el recuento de su categoría. Los barras tienen el mismo ancho, por lo que sus alturas determinan sus áreas, y estas áreas son proporcionales a los recuentos en cada categoría. De esta forma, se pueden ver fácilmente que había más del doble de pasajeros de tercera clase, que de primera o segunda clase. Los gráficos de barras hacen que este tipo de comparaciones sean fáciles y naturales. Se crearán de forma sencilla utilizando el método <span style="color:limegreen">plot()</span> dentro de un <span style="color:yellow">DataFrame</span> de Pandas.

```python
# Grafico de barras de pasajeros del Titanic
plot = titanic['class'].value_counts().plot(kind='bar', title='Pasajeros del Titanic')
```

Si se quisiéra enfocar en la proporción relativa de los pasajeros de cada una de las clases, simplemente se pueden sustituir los recuentos con porcentajes y utilizar un gráfico de barras de frecuencias relativas.

Los gráfico de torta dan una impresión rápida de cómo todo un grupo se divide en grupos más pequeños. Para este tipo de comparaciones, son mucho más útiles los gráficos de barras.

#### Relacionando Variables Categóricas

Al analizar la tragedia del Titanic, una de las preguntas que se podría hacer es ¿existe alguna relación entre la clase de pasajeros y la posibilidad de alcanzar un bote salvavidas y sobrevivir a la tragedia? Para poder responder a esta pregunta, se necesitan analizar las variables <span style="color:yellow">class</span> y <span style="color:yellow">survived</span> del dataset en forma conjunta. Una buena forma de analizar dos variables categóricas en forma conjunta, es agrupar los recuentos en una tabla de doble entrada; este tipo de tablas se conocen en estadística con el nombre de tabla de contingencia. Se crea esta tabla utilizando la función <span style="color:limegreen">crosstab()</span> de Pandas.

```python
# Tabla de contingencia class / survived
print(pd.crosstab(index=titanic['survived'], columns = titanic['class'], margins=True))
```

Los márgenes de la tabla, tanto en la derecha y en la parte inferior, muestran los totales. La línea inferior de la tabla representa la distribución de frecuencia de la clase de pasajeros. La columna derecha de la tabla es la distribución de frecuencia de la variable supervivencia. Cuando se presenta la información de este modo, cada celda de cada uno de los márgenes de la tabla representa la distribución marginal de esa variable en particular. Cada celda muestra el recuento para la combinación de los valores de las dos variables categóricas, en este caso <span style="color:yellow">class</span> y <span style="color:yellow">survived</span>.

De la misma manera que se hizo con las tablas de frecuencia, también podría ser útil representar las tablas de contingencia con porcentajes relativos; esto se puede realizar utilizando el método <span style="color:limegreen">apply()</span> del siguiente modo:

```python
# tabla de contingencia en porcentajes relativos total
print(pd.crosstab(index=titanic['survived'], columns= titanic['class'], margins=True).apply(lambda r: r/len(titanic)*100, axis=1))
```

Con esta tabla se puede ver fácilmente que solo el $37.91\%$ de los pasajeros sobrevivió a la tragedia y que este $37\%$ se compone de la siguiente forma: *del total de pasajeros sobrevivió un $15.42\%$ de pasajeros que eran de primera clase*, *un $8.97\%$ que eran de segunda clase* y *un $13.52\%$ que eran pasajeros de tercera clase*.

Volviendo a la pregunta inicial sobre la posibilidad de sobrevivir según la clase de pasajero, podría ser útil armar la tabla de porcentajes como un porcentaje relativo sobre el total de cada fila, es decir calcular el porcentaje relativo que cada clase tiene sobre haber sobrevivido o no. Esto se puede realizar del siguiente modo:

```python
# tabla de contingencia en porcentajes relativos segun sobreviviente
print(pd.crosstab(index=titanic['survived'], columns=titanic['class']).apply(lambda r: r/r.sum() * 100, axis=1))
```

Aquí se puede ver que de los pasajeros que sobrevivieron a la tragedia, el $40.68\%$ correspondían a primera clase, el $35.67\%$ a tercera clase y el $23.65\%$ a segunda clase. Por tanto se puede inferir que los pasajeros de primera clase tenían más posibilidades de sobrevivir.

Es más, también se podría armar la tabla de porcentaje relativos en relación al total de cada clase de pasajero y así poder ver que de los pasajeros de primera clase, logró sobrevivir un $62.46\%$.

```python
# tabla de contingencia en porcentajes relativos segun clase
print(pd.crosstab(index=titanic['survived'], columns=titanic['class']).apply(lambda r: r/r.sum() * 100, axis=0))
```

Este último resultado se podría representar visualmente con simples gráficos de barras del siguiente modo:

```python
# Grafico de barras de sobrevivientes segun clase
plot = pd.crosstab(index=titanic['class'], columns=titanic['survived']).apply(lambda r: r/r.sum() * 100, axis=1).plot(kind='bar')
```

```python
# Grafico de barras de sobrevivientes segun clase
plot = pd.crosstab(index=titanic['survived'], columns=titanic['class']).apply(lambda r: r/r.sum() * 100, axis=0).plot(kind='bar', stacked=True)
```

Para las variables cuantitativas, a diferencia de lo que pasa con las variables categóricas, no existe una forma obvia de agrupar los datos; por tal motivo lo que se suele hacer es, dividir los posibles valores en diferentes contenedores del mismo tamaño y luego contar el número de casos que caen dentro de cada uno de los contenedores. Estos contenedores junto con sus recuentos, proporcionan una imagen de la distribución de la variable cuantitativa y constituyen la base para poder graficar el histograma. Para construir el gráfico, simplemente se deben representar los recuentos como barras y graficarlas contra los valores de cada uno de los contenedores.

Con Python se puede representar fácilmente el histograma de la variable <span style="color:yellow">eruptions</span> utilizando el método <span style="color:limegreen">hist()</span> del Dataframe de Pandas del siguiente modo:

```python
# histograma duracion de erupciones con 8 barras
faithful['eruptions'].hist(bins=8)
plt.xlabel("Duracion en minutos")
plt.ylabel("Frecuencia")
plt.show()
```

Como se puede observar en este gráfico, la duración más frecuente de las erupciones del géiser ronda en alrededor de cuatro minutos y medio. Una cosa que se debe hacer notar es que en los histogramas, los contenedores dividen a todos los valores de la variable cuantitativa, por lo que no se deberían encontrar espacios entre las barras (a diferencia de lo que pasaba con los gráficos de barras que se observaron con variables categóricas). Cualquier espacio entre las barras es una brecha en los datos, que indica un región para la que no existen valores.

#### Distribución de Frecuencia

Un tema íntimamente relacionado con los histogramas son las tablas de distribución de frecuencia, en definitiva los histogramas no son más que gráficos de tablas de distribución de frecuencia. La distribución de frecuencia de una variable cuantitativa consiste en un resumen de la ocurrencia de un dato dentro de una colección de categorías que no se superponen. Estas categorías se deben poder agrupar según la conveniencia y lo que se desea analizar. Por ejemplo si se quisiera armar la distribución de frecuencia de la variable <span style="color:yellow">eruptions</span> se podrían realizar las siguiente manipulaciones con Pandas:

```python
# Distribución de Frecuencia
# Primero se crea un rango de valores para las categorías
contenedores = np.arange(1.5, 6., 0.5)

# Ahora se dividen los datos en cada contenedor
frec = pd.cut(faithful['eruptions'], contenedores)

# Por ultimo se hace el recuento de los contenedores para armar la tabla de frecuencias.
tabla_frec = frec.value_counts()
tabla_frec
```

Como se muestra, en esta tabla de distribución de frecuencia, la duración que más veces ocurre para las erupciones, se encuentran en el rango de 4 a 4.5 minutos.

#### Diagrama de Tallos y Hojas

Los histogramas permiten apreciar la distribución de los datos de una forma sencilla, pero los mismos no muestran los valores del dataset en sí mismos. Para solucionar esto, existe el *diagrama de tallos y hojas*, el cual es similar al histograma pero muestra los valores individuales del dataset. Para que quede más claro, se analizará un ejemplo sencillo. Suponga que se tiene una muestra sobre el ritmo cardíaco de 24 mujeres, las observaciones son las siguientes:

```python
pulso = [88, 80, 76, 72, 68, 56, 64, 60, 64, 68, 64, 68, 72, 76, 80, 84, 68, 80, 76, 
         72, 84, 80, 72, 76]
```

Se podría graficar el histograma de estas observaciones del siguiente modo:

```python
plt.hist(pulso, bins=7)
plt.xlabel("Latidos por minuto")
plt.ylabel("Cantidad de mujeres")
plt.show()
```

Este histograma muestra la forma en que se distribuyen los datos, pero no muestra los datos individuales. Para esto se podría graficar su diagrama de tallos y hojas de la siguiente manera:

```python
# Diagrama de Tallos y Hojas
def tallos(d):
    "Genera un simple diagrama de tallos y hojas"
    l, t = np.sort(d), 10
    O = range(int(l[0] - l[0]%t), int(l[-1]+11), t)
    I = np.searchsorted(l, O)
    for e, a, f in zip(I, I[1:], O):
        print('%3d|'%(f/t), *(l[e:a]-f), sep='')

tallos(pulso)
```

Como se muestra, la distribución del diagrama de tallos y hojas es similar a la del histograma, pero en este caso si se pueden ver los valores de las observaciones. El diagrama se lee así: Por un lado se tienen las decenas de los latidos, las cuales constituyen los tallos del diagrama (los valores antes del pipe o barra vertical "|") y luego se van agregando hojas a estos tallos , representadas por las unidades de cada latido. De esta forma 5|6, significa que solo aparece el valor 56 una sola vez, en cambio 8|0000, significa que tenemos el valor 80 observado en 4 oportunidades.

#### Medidas de Tendencia Central

Una vez que se tiene una buena idea visual de como se distribuyen los datos y de las relaciones que pueden existir entre los mismos, se pueden calcular medidas numéricas propias de la estadística descriptiva. En general, suele ser interesante conocer cual puede ser el promedio o valor central al que tiende la distribución de los datos, para esto se utilizan las medidas de tendencia central, entre las que podemos encontrar a:

+ La media aritmética
+ La media ponderada
+ La media geométrica
+ La media armónica
+ La mediana
+ La media truncada
+ La moda

**Media Aritmética**

La media aritmética es el valor obtenido al sumar todos los datos y dividir el resultado entre el número total elementos. Se calcula con el método <span style="color:limegreen">mean()</span>.

```python
faithful['eruptions'].mean()
```

**Media Ponderada**

La media ponderada es apropiada cuando en un dataset cada dato tiene una importancia relativa (o peso) respecto de los demás. Como esta media no aplica para el dataset del ejemplo no se la puede calcular.

**Media Geométrica**

La media geométrica es útil cuando se desean comparar cosas con propiedades muy diferentes; también es recomendada para datos de progresión geométrica, para promediar razones, interés compuesto y números índices. Se calcula tomando la raíz n-ésima del producto de todos los datos. Se calcula con la función <span style="color:limegreen">gmean()</span> de SciPy.

```python
stats.gmean(faithful['eruptions'])
```

**Media Armónica**

La media armónica promedia el número de elementos y los divide por la suma de sus inversos. La media armónica es siempre la media más baja y es recomendada para promediar velocidades. Se calcula con la función <span style="color:limegreen">hmean()</span> de SciPy.

```python
stats.hmean(faithful['eruptions'])
```

**Mediana**

La mediana representa el valor de posición central en un conjunto de datos ordenados. Se la puede calcular utilizando el método <span style="color:limegreen">median()</span> de Pandas

```python
faithful['eruptions'].median()
```

**Media Truncada**

La media truncada es una mezcla entre la media aritmética y la mediana. Para calcular el promedio previamente se descartan porciones en el extremo inferior y superior de la distribución de los datos. En Python se puede utilizar la función <span style="color:limegreen">trim_mean()</span> de SciPy.

```python
stats.trim_mean(faithful['eruptions'], .10)
```

**Moda**

Por último, la moda es el valor que tiene mayor frecuencia absoluta. Son los picos que vemos en el histograma. Dependiendo de la distribución de los datos puede existir más de una, como en el caso de la variable <span style="color:yellow">eruptions</span>. Se la calcula con el método <span style="color:limegreen">mode()</span>.

```python
faithful['eruptions'].mode()
```

### Medidas de Dispersión

Las medidas de tendencia central no son las únicas medidas de resumen estadístico que se pueden calcular; otras medidas también de gran importancia son las medidas de dispersión. Las medidas de dispersión, también llamadas medidas de variabilidad, muestran la variabilidad de una distribución, indicando por medio de un número si las diferentes puntuaciones de una variable están muy alejadas de la media. Cuanto mayor sea ese valor, mayor será la variabilidad y cuanto menor sea, más homogénea será a la media. Así se sabe si todos los casos son parecidos o varían mucho entre ellos. Las principales medidas de dispersión son:

+ La varianza
+ La desviación estándar
+ Los cuartiles
+ La covarianza
+ El coeficiente de correlación


**Varianza**

La varianza intenta describir la dispersión de los datos. Se define como la esperanza del cuadrado de la desviación de dicha variable respecto a su media. Una varianza pequeña indica que los puntos de los datos tienden a estar muy cerca de la media y por lo tanto el uno al otro, mientras que una varianza alta indica que los puntos de los datos están muy dispersos alrededor de la media y lejos uno del otro. Se puede calcular con el método <span style="color:limegreen">var()</span>.

```python
faithful['eruptions'].var()
```

**Desviación Estándar**

La desviación estándar o desviación típica es una medida que se utiliza para cuantificar la cantidad de variación o dispersión de un conjunto de valores o de datos. Una desviación estándar cercana a 0 indica que los puntos de los datos tienden a estar muy cerca de la media del conjunto, mientras que una desviación estándar alta indica que los puntos de los datos se extienden a lo largo de un amplio rango de valores. Se calcula como la raíz cuadrada de la varianza y con Pandas se puede calcular con el método <span style="color:limegreen">std()</span>.

```python
faithful['eruptions'].std()
```

**Cuartiles**

Los cuartiles son los tres puntos que dividen el conjunto de datos en cuatro grupos iguales, cada grupo comprende un cuarto de los datos. El <span style="color:yellow">(Q1)</span> se define como el número medio entre el número más pequeño y la mediana del conjunto de datos. El segundo cuartil <span style="color:yellow">(Q2)</span> es la mediana de los datos. El tercer cuartil <span style="color:yellow">(Q3)</span> es el valor medio entre la mediana y el valor más alto del conjunto de datos. Para dividir un dataset en sus cuartiles se utiliza el método <span style="color:limegreen">quantile()</span>.

```python
faithful['eruptions'].quantile([.25, .5, .75])
```

Un gráfico relacionado a los cuartiles y que describe varias características importantes al mismo tiempo, tales como la dispersión y simetría es el <span style="color:yellow">diagrama de caja</span>. Para su realización se representan los tres cuartiles y los valores mínimo y máximo de los datos, sobre un rectángulo, alineado horizontal o verticalmente. Se puede utilizar el método <span style="color:limegreen">boxplot()</span> de Seaborn para generarlo.

```python
cajas = sns.boxplot(list(faithful['eruptions']))
```

Hasta aquí se han calculado medidas de dispersión para una sola variable, pero normalmente un dataset tiene dos o mas variables cuantitativas. Para poder calcular medidas combinadas para dos variables se utiliza la covarianza y la correlación.

**Covarianza**

La covarianza es el equivalente de la varianza aplicado a una variable bidimensional. Es la media aritmética de los productos de las desviaciones de cada una de las variables respecto a sus medias. La covarianza indica el sentido de la correlación entre las variables; Si es mayor que cero la correlación es directa, en caso de ser menor, la correlación es inversa. Se puede calcular utilizando el método <span style="color:limegreen">cov()</span>.

```python
faithful.cov()
```

**Correlación**

Por último, el coeficiente de correlación es una medida del grado de dependencia lineal entre dos variables. El coeficiente de correlación oscila entre $-1$ y $1$. Un valor de $1$ significa que una ecuación lineal describe la relación entre las dos variables a la perfección, con todos los puntos cayendo sobre una línea recta de pendiente positiva. Un valor de $-1$ implica que todos los puntos se encuentran en una línea con pendiente negativa. Un valor de $0$ implica que no existe una correlación lineal entre las variables. Se la puede calcular con el método <span style="color:limegreen">corr()</span>.

```python
faithful.corr()
```

Como se puede apreciar las dos variables tienen una correlación bastante alta, lo que sugiere que están íntimamente relacionadas; a la misma conclusión se había llegado al observar el diagrama de dispersión.
